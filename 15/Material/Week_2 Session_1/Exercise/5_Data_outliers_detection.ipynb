```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

# --- Configuration ---
NUM_SAMPLES = 100
NUM_OUTLIERS = 6
path='/content/drive/MyDrive/MScDS&AI/DataPreprocessing-II/'
FILENAME = "data_with_outliers.csv"
# Read the data back from the file into a new DataFrame
df = pd.read_csv(path+FILENAME)
print("Data read successfully.")
print("\nFirst 5 rows of the loaded data:")
print(df.head())

# --- Step 1: Detecting Outliers with Box Plots ---
print("\n--- 1. Detecting Outliers with Box Plots ---")
# Create a figure to hold the subplots
plt.figure(figsize=(14, 6))

# Subplot 1: Box plot for Age from the loaded data
plt.subplot(1, 2, 1)  # (1 row, 2 columns, 1st subplot)
plt.boxplot(df['age'])
plt.title('Box Plot of Age')
plt.ylabel('Age')
plt.grid(True)

# Subplot 2: Box plot for Salary from the loaded data
plt.subplot(1, 2, 2)  # (1 row, 2 columns, 2nd subplot)
plt.boxplot(df['salary'])
plt.title('Box Plot of Salary')
plt.ylabel('Salary ($)')
plt.grid(True)

print("Displaying box plots. Close the plot window to see the scatter plot.")
plt.show()


# --- Step 2: Visualizing Outliers with a Scatter Plot ---
print("\n--- 2. Visualizing Outliers with a Scatter Plot ---")
# Create a new figure for the scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df['age'], df['salary'])
plt.title('Age vs. Salary with Outliers')
plt.xlabel('Age')
plt.ylabel('Salary ($)')
plt.grid(True)

print("Displaying scatter plot. Close the plot window to exit the script.")
plt.show()
```
