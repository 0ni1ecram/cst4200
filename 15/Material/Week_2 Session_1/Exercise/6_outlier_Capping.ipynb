```python
import pandas as pd
import numpy as np

# Sample data with an outlier
data = {'score': [85, 92, 78, 88, 95, 150]}
df = pd.DataFrame(data)
```

```python

# Calculate IQR bounds
Q1 = df['score'].quantile(0.25)
Q3 = df['score'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR

print(f"Upper Bound for outliers: {upper_bound}")

# Cap the outliers
df['score_capped'] = np.where(df['score'] > upper_bound, upper_bound, df['score'])

print("\nOriginal vs Capped:\n", df)
```
