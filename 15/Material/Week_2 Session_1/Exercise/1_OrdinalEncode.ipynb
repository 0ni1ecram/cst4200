```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Sample Data
df = pd.DataFrame({'size': ['Medium', 'Small', 'Large', 'Medium']})
print("Original Data:\n", df)
```

```python

# Define the desired order of categories.
size_categories = ['Small', 'Medium', 'Large']

# Initialize the encoder with the specific order
encoder = OrdinalEncoder(categories=[size_categories])

# Fit and transform the data.
df['size_encoded'] = encoder.fit_transform(df[['size']])
df['size_encoded'] = df['size_encoded'].astype(int)

print("\nEncoded Data:\n", df)
```
