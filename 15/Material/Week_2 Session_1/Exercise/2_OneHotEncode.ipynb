```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Sample Data
df = pd.DataFrame({'Country': ['USA', 'UK', 'Germany', 'UK']})
print("Original Data:\n", df)
```

```python

# Initialize encoder. sparse_output=False returns a readable numpy array.
encoder = OneHotEncoder(sparse_output=False)

# Reshape data, fit, and transform
encoded_data = encoder.fit_transform(df[['Country']])

# Create a new DataFrame with meaningful column names
encoded_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(['Country']))

print("\nData after One-Hot Encoding:\n", encoded_df)
```
