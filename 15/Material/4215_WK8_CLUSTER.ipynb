```python
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. Load DAta
# csv_file = "F:/Users/MSR/Documents/CST4200/15/Datasets/clusters.csv"
csv_file = "F:/Users/MSR/Documents/CST4200/15/Datasets/Wholesale customers data.csv"

# --- 2.Read Data from CSV File
df0 = pd.read_csv(csv_file)

print(df0.dtypes)
print(df0.head())
print(df0.columns.tolist)


df0.shape
```

```python
# --- Analyze Data to determine Clusters
df0['Channel'].value_counts(normalize=True)
df0['Region'].value_counts(normalize=True)
# df0['Fresh'].value_counts(normalize=True)
```

```python
# --- 3.Select DAta for each Cluster
df = df0.copy()
# Filter the DataFrame, to get X,Y coordinates for each label as numpy arrays
# We use ".values" to Convert the Pandas DAtasFrame columns to numpy.Arrays

c1_data = df[df0['Region'] == 1][['Fresh', 'Frozen' ]].values
c2_data = df[df0['Region'] == 2][['Fresh', 'Frozen']].values
c3_data = df[df0['Region'] == 3][['Fresh', 'Frozen']].values

# print(c1_data) 'Channel', 'Grocery', 'Frozen','Detergents_Paper', 'Delicassen',
```

```python
# ---  3. Plot Data
plt.figure(figsize=(10, 8))

# Plot for each: Color, Label for Legend
plt.scatter(c1_data[:,0], c1_data[:, 1], s=15, color='#f2b705', label='c1') #yellow
plt.scatter(c2_data[:,0], c2_data[:, 1], s=15, color='#e41a1c', label='c2') #Red
plt.scatter(c3_data[:,0], c3_data[:, 1], s=15, color='#4daf4a', label='c3') #Green

# --- 4. Customize Plot
plt.title('Cluster Data Visualization from CSV')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.axis('equal')  # Ensure x and y axes having same scale

# Show the plot
plt.show()
```
