```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_inverse(y):
    return np.log(y / (1 - y))  # Inverse sigmoid for better weight initialization

# Input data (5 samples, each with 3 features)
X = np.array([[1, 0, 1],
              [0, 1, 0],
              [1, 1, 1],
              [0, 0, 1],
              [1, 0, 0]]).T  # Transposed to match matrix dimensions
print("Input:")
print(X.T)  # Transposed back for readability

```

```python

# Adjusted Encoder weights to reduce dimension to 2
W_enc = np.array([[5, -5, 5],
                  [-5, 5, -5]])  # 2D latent space

# Encoding step
Z_enc = np.dot(W_enc, X)
A_enc = sigmoid(Z_enc)

print("\nEncoded Representation (2D):")
print(A_enc.T)

```

```python

# Adjusted Decoder weights to map back to 3 dimensions
W_dec = np.array([[5, -5],
                  [-5, 5],
                  [5, -5]])


# Decoding step
Z_dec = np.dot(W_dec, A_enc)
A_dec = sigmoid(Z_dec)



print("\nReconstructed Output:")
print(A_dec.T)
```
