```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D # Required for 3D plotting

# --- 1. Define the Initial Values from the Example ---
x = np.array([0.3, 0.8, 0.1])
w2_initial = np.array([0.4, 0.9, 0.2])
alpha = 0.5
h_bmu = 1.0

print("--- SOM Single Weight Update Step for the BMU ---")
print(f"Input Vector (x):          {x}")
print(f"Initial BMU Weight (w2_t): {w2_initial}")
print(f"Learning Rate (alpha):     {alpha}")

```

```python

# --- 2. Perform the Update Calculation Step-by-Step ---
print("Step 1: Calculate the difference vector (x - w2_t)")
difference_vector = x - w2_initial
print(f"Difference: {x} - {w2_initial} = {np.round(difference_vector, 2)}")

print("\nStep 2: Calculate the update amount (alpha * h * difference)")
update_amount = alpha * h_bmu * difference_vector
print(f"Update Amount: {alpha} * {h_bmu} * {np.round(difference_vector, 2)} = {np.round(update_amount, 3)}")

print("\nStep 3: Calculate the new weight (w2_t+1 = w2_t + update_amount)")
w2_new = w2_initial + update_amount
print(f"New Weight (w2_t+1): {w2_initial} + {np.round(update_amount, 3)} = {np.round(w2_new, 3)}")
print("-" * 50)
print("The BMU's weight vector has moved closer to the input vector.")
```

```python

# --- 3. Visualize the Weight Update in 3D ---
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the input vector
ax.scatter(x[0], x[1], x[2], c='red', marker='*', s=250, label='Input Vector (x)', depthshade=False)

# Plot the initial position of the BMU's weight vector
ax.scatter(w2_initial[0], w2_initial[1], w2_initial[2], c='blue', s=100, label='BMU Initial Position (w2_t)', depthshade=False)

# Plot the new position of the BMU's weight vector
ax.scatter(w2_new[0], w2_new[1], w2_new[2], c='green', s=100, label='BMU New Position (w2_t+1)', depthshade=False)

# Draw an arrow from the initial position to the new position to show the update
ax.quiver(w2_initial[0], w2_initial[1], w2_initial[2],
          update_amount[0], update_amount[1], update_amount[2],
          color='black', label='Weight Update Vector', arrow_length_ratio=0.3)

# Draw a dashed line from input to the initial BMU position for context
ax.plot([x[0], w2_initial[0]], [x[1], w2_initial[1]], [x[2], w2_initial[2]], 'k--', linewidth=1, alpha=0.5)


# Formatting the plot
ax.set_title("Visualizing the BMU's Weight Update")
ax.set_xlabel('Dimension 1')
ax.set_ylabel('Dimension 2')
ax.set_zlabel('Dimension 3')
ax.legend()
plt.grid(True)
plt.show()
```
