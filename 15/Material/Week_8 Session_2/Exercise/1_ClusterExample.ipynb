```python
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. Load Data ---
csv_filename = '/content/drive/MyDrive/MScDS&AI/DeepLearning/clusters.csv'

# Read the data from the CSV file
df = pd.read_csv(csv_filename)

# --- 2. Select Data for Each Cluster ---

# Filter the DataFrame to get x, y coordinates for each label as numpy arrays
# We use .values to convert the pandas DataFrame columns to numpy arrays
c1_data = df[df['label'] == 'c1'][['x', 'y']].values
c2_data = df[df['label'] == 'c2'][['x', 'y']].values
c3_data = df[df['label'] == 'c3'][['x', 'y']].values

# --- 3. Plot Data ---
plt.figure(figsize=(10, 8))

# Plot each cluster with the specified colors and labels for the legend
plt.scatter(c1_data[:, 0], c1_data[:, 1], s=15, color='#f2b705', label='c1') # Yellow
plt.scatter(c2_data[:, 0], c2_data[:, 1], s=15, color='#e41a1c', label='c2') # Red
plt.scatter(c3_data[:, 0], c3_data[:, 1], s=15, color='#4daf4a', label='c3') # Green

# --- 4. Customize Plot ---
plt.title('Cluster Data Visualization from CSV')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend() # Show the legend with cluster labels
plt.grid(True)
plt.axis('equal') # Ensure x and y axes have the same scale

# Show the plot
plt.show()
```
