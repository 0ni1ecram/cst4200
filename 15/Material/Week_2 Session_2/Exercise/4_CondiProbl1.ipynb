```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/MScDS&AI/Statistics/1.employment_records.csv')


# 1. Filter the DataFrame to create our condition (only Business Analysts)
business_analysts = df[df['Job Title'] == 'Business Analyst']

# 2. Within that filtered group, count who is 'Married'
married_business_analysts = business_analysts[business_analysts['Marital Status'] == 'Married']
```

```python

# 3. Calculate the conditional probability
# P(A|B) = Count(A and B) / Count(B)
prob_married_given_ba = len(married_business_analysts) / len(business_analysts)

print(f"Total number of Business Analysts: {len(business_analysts)}")
print(f"Number of married Business Analysts: {len(married_business_analysts)}")
print(f"The probability of being married given the employee is a Business Analyst is: {prob_married_given_ba:.2%}")
```
