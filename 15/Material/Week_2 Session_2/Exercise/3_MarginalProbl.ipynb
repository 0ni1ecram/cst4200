```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/drive/MyDrive/MScDS&AI/Statistics/1.employment_records.csv')

# --- Marginal Probability Calculation ---

# 1. Choose a categorical column
column_of_interest = 'Marital Status'

# 2. Count the occurrences of each category in the column
category_counts = df[column_of_interest].value_counts()

# 3. Count the total number of employees
total_employees = len(df)
```

```python

# 4. Calculate the marginal probability for each category
marginal_probabilities = category_counts / total_employees

# --- Display the Results ---
print(f"--- Marginal Probability for {column_of_interest} ---")
print(f"\nTotal Employees: {total_employees}")
print("\nCategory Counts:")
print(category_counts)
print("\nMarginal Probabilities:")

print(marginal_probabilities)
```
