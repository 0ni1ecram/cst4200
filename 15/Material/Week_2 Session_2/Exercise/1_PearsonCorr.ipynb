```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

# --- 1. Load the Dataset ---
# Load the data from the CSV file into a pandas DataFrame.
df = pd.read_csv('/content/drive/MyDrive/MScDS&AI/Statistics/1.employment_records.csv')

# --- 2. Extract Data Columns ---
# Ensure the 'Salary' and 'Bonus' columns exist and are numeric
salary = df['Salary']
bonus = df['Bonus']
```

```python

# --- 3. Visualize the Data ---
# A scatter plot is used to visualize the relationship between two variables.
plt.figure(figsize=(10, 6))

# Create the scatter plot using matplotlib
plt.scatter(salary, bonus, alpha=0.6)

# Add titles and labels for clarity
plt.title('Relationship between Salary and Bonus', fontsize=16)
plt.xlabel('Salary', fontsize=12)
plt.ylabel('Bonus', fontsize=12)
plt.grid(True)

print("Displaying scatter plot...")
plt.show()
```

```python

# --- 4. Compute Pearson Correlation ---
# The Pearson correlation test measures the linear relationship between two variables.
# It returns the correlation coefficient (r) and the p-value.
correlation_coefficient, p_value = stats.pearsonr(salary, bonus)

# --- 5. Print the Results ---
print("\n--- Pearson Correlation Results ---")
print(f"Correlation Coefficient (r): {correlation_coefficient:.4f}")
print(f"P-value: {p_value:.4f}")
print("-----------------------------------")
```

```python

# Interpretation of the correlation coefficient
if abs(correlation_coefficient) > 0.7:
    strength = "strong"
elif abs(correlation_coefficient) > 0.4:
    strength = "moderate"
elif abs(correlation_coefficient) > 0.1:
    strength = "weak"
else:
    strength = "very weak or no"

direction = "positive" if correlation_coefficient > 0 else "negative"

print(f"Interpretation: There is a {strength} {direction} linear relationship between Salary and Bonus.")

```
