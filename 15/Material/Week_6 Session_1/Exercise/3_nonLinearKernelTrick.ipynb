```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
path='/content/drive/MyDrive/MScDS&AI/SVM/'



# --- Read the DataFrame from the CSV file ---
print("Reading DataFrame from 'non_linear.csv'...")
df = pd.read_csv(path+'non_linear.csv')
print("DataFrame loaded successfully.\n")


# --- 2. Visualize the Original 1D Data (Simplified Plot) ---

# This plot shows that the data is not linearly separable in 1D.
plt.figure()
plt.title('Original 1D Data')
# Use a stripplot to visualize the 1D distribution
sns.stripplot(x='X', y='Class', data=df, hue='Class', orient='h', jitter=0.1)
plt.yticks([0, 1], ['Class 0', 'Class 1'])
plt.xlabel('Feature X')
plt.ylabel('Class')
plt.show()


# --- 3. Perform the Square Transformation (The "Kernel Trick") ---

# We manually apply a transformation to map the data into a higher dimension.
# The new dimension is created by squaring the original feature X.
# This simulates what a polynomial kernel (degree=2) does implicitly.
df['X_squared'] = df['X']**2

# Print the head of the updated DataFrame
print("\n--- Data After Transformation (in the same DataFrame) ---")
print(df.head())
print("\n")


# --- 4. Visualize the Transformed 2D Data (Simplified Plot) ---

# This plot shows that after the transformation, the data becomes linearly separable.
plt.figure()
plt.title('Transformed 2D Data')

# Create a scatter plot of the new 2D data
sns.scatterplot(x='X', y='X_squared', data=df, hue='Class')

# Draw the separating hyperplane
plt.axhline(y=1.5, color='green', linestyle='--', label='Separating Hyperplane')

plt.xlabel('Original Feature (X)')
plt.ylabel('Transformed Feature (X^2)')
plt.legend()
plt.show()


```
