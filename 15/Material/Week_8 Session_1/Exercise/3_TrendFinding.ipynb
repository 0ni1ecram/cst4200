```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
file_path = "/content/drive/MyDrive/MScDS&AI/AutoregressionANN/1.Stock_Data_for_XYZ_Company.xlsx"  # Update the path if needed
stock_data = pd.read_excel(file_path, parse_dates=['Date'])

# Ensure the data is sorted by date
stock_data = stock_data.sort_values(by="Date")

# Plot the stock prices
plt.figure(figsize=(12, 6))
plt.plot(stock_data['Date'], stock_data['Stock Price'], label='XYZ Stock Price', color='blue')

# Labels and title
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.title('Stock Price Trend for XYZ Company')
plt.legend()
plt.grid()

# Show the plot
plt.show()
```

```python
# Compute Moving Average (e.g., 30-day moving average)
stock_data['Moving_Avg'] = stock_data['Stock Price'].rolling(window=30, min_periods=1).mean()

# Plot stock prices and moving average
plt.figure(figsize=(12, 6))
plt.plot(stock_data['Date'], stock_data['Stock Price'], label='Stock Price', color='blue', alpha=0.5)
plt.plot(stock_data['Date'], stock_data['Moving_Avg'], label='30-Day Moving Average', color='red', linewidth=2)

# Labels and title
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.title('Upward Trend Detection using Moving Average')
plt.legend()
plt.grid()

# Show the plot
plt.show()

# Determine if there is an upward trend
if stock_data['Moving_Avg'].iloc[-1] > stock_data['Moving_Avg'].iloc[0]:
    print("Upward trend detected! 📈")
else:
    print("No clear upward trend detected.")
```
