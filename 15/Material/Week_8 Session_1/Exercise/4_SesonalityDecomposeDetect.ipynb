```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
file_path = "/content/drive/MyDrive/MScDS&AI/AutoregressionANN/2.Sales_Data.xlsx"  # Ensure the file is in the correct directory
sales_data = pd.read_excel(file_path, parse_dates=['Date'],sheet_name="Sales_Data")

# Sort data by date (if not already sorted)
sales_data = sales_data.sort_values(by="Date")

# Plot the sales data
plt.figure(figsize=(12, 6))
plt.plot(sales_data['Date'], sales_data['Sales'], label='Sales Data', color='green')

# Labels and title
plt.xlabel('Date')
plt.ylabel('Sales')
plt.title('Sales Data Over Time')
plt.legend()
plt.grid()

# Show the plot
plt.show()
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose

# Perform seasonal decomposition (additive model)
decomposition = seasonal_decompose(sales_data['Sales'], model='additive', period=30)  # Assuming monthly seasonality

# Plot decomposition results
plt.figure(figsize=(12, 8))

# Trend component
plt.subplot(3, 1, 1)
plt.plot(sales_data['Date'], decomposition.trend, label='Trend', color='blue')
plt.legend()

# Seasonal component
plt.subplot(3, 1, 2)
plt.plot(sales_data['Date'], decomposition.seasonal, label='Seasonality', color='orange')
plt.legend()

# Residual component
plt.subplot(3, 1, 3)
plt.plot(sales_data['Date'], decomposition.resid, label='Residuals', color='gray')
plt.legend()

# Display the plots
plt.tight_layout()
plt.show()

# Print a message based on seasonal detection
if decomposition.seasonal.std() > decomposition.resid.std():
    print("Seasonality detected in the sales data! 🎉")
else:
    print("No significant seasonality detected.")
```
