```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
file_path = "/content/drive/MyDrive/MScDS&AI/AutoregressionANN/2.Sales_Data.xlsx"  # Ensure the file is in the correct directory
sales_data = pd.read_excel(file_path, parse_dates=['Date'],sheet_name='Sales_Data')

# Sort data by date (if not already sorted)
sales_data = sales_data.sort_values(by="Date")

# Plot the sales data
plt.figure(figsize=(12, 6))
plt.plot(sales_data['Date'], sales_data['Sales'], label='Sales Data', color='green')

# Labels and title
plt.xlabel('Date')
plt.ylabel('Sales')
plt.title('Sales Data Over Time')
plt.legend()
plt.grid()

# Show the plot
plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_acf

# Plot the Autocorrelation Function (ACF)
plt.figure(figsize=(12, 6))
plot_acf(sales_data['Sales'], lags=50)  # Checking up to 50 lags for seasonality
plt.title("Autocorrelation Function (ACF) of Sales Data")
plt.show()
```
