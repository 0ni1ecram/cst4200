```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
file_path = "/content/drive/MyDrive/MScDS&AI/AutoregressionANN/1.Stock_Data_for_XYZ_Company.xlsx"
stock_data = pd.read_excel(file_path, parse_dates=['Date'])

# Ensure the data is sorted by date
stock_data = stock_data.sort_values(by="Date")

# Plot the stock prices
plt.figure(figsize=(12, 6))
plt.plot(stock_data['Date'], stock_data['Stock Price'], label='XYZ Stock Price', color='blue')

# Labels and title
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.title('Stock Price Trend for XYZ Company')
plt.legend()
plt.grid()

# Show the plot
plt.show()
```

```python
# Plot the bar chart for stock prices
plt.figure(figsize=(12, 6))
plt.bar(stock_data['Date'], stock_data['Stock Price'], label='XYZ Stock Price', color='blue', width=2)

# Labels and title
plt.xlabel('Date')
plt.ylabel('Stock Price')
plt.title('Stock Price Bar Chart for XYZ Company')
plt.legend()
plt.grid()

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Show the plot
plt.show()
```
