```python
import pandas as pd
from geopy.distance import geodesic
import numpy as np

# --- Step 1: Load the Data ---

df = pd.read_csv('/content/drive/MyDrive/MScDS&AI/FeatureEngineering/6.gloalLocation.csv')
print("--- Successfully loaded 6.gloalLocation.csv ---")
```

```python


# --- Part 1: Calculate the Distance Between Two Points ---
print("\n--- Part 1: Distance Calculation ---")

# Select two specific locations from the dataset
point1_name = 'Eiffel Tower'
point2_name = 'Buckingham Palace'

# Get the coordinates for each point
loc1 = df[df['specific_location'] == point1_name]
loc2 = df[df['specific_location'] == point2_name]

if not loc1.empty and not loc2.empty:
    coords1 = (loc1.iloc[0]['latitude'], loc1.iloc[0]['longitude'])
    coords2 = (loc2.iloc[0]['latitude'], loc2.iloc[0]['longitude'])

    # Calculate the distance using the geodesic formula (most accurate)
    distance_km = geodesic(coords1, coords2).kilometers

    print(f"The distance between {point1_name} and {point2_name} is: {distance_km:.2f} km")
else:
    print(f"Could not find one or both locations: '{point1_name}', '{point2_name}'")
```

```python


# --- Part 2: Simple Geospatial Clustering (Hemispheres) ---
print("\n--- Part 2: Simple Geospatial Clustering ---")

# Create two clusters based on the Prime Meridian (longitude 0)
# Locations with longitude >= 0 are in the Eastern Hemisphere
# Locations with longitude < 0 are in the Western Hemisphere
df['cluster'] = np.where(df['longitude'] >= 0, 'Eastern Hemisphere', 'Western Hemisphere')

print("\nSuccessfully grouped locations into two clusters (hemispheres).")
print("Displaying the first 10 rows with their assigned cluster:")
print(df.head(10))

# Display the number of locations in each identified cluster
print("\n--- Cluster Distribution ---")
print(df['cluster'].value_counts())

```
