```python
import pandas as pd

# Define the file path for the uploaded CSV
path='/content/drive/MyDrive/MScDS&AI/FeatureEngineering/'
file_path = path+'2.loan_sample_data.csv'

# Read the CSV file into a pandas DataFrame
df = pd.read_csv(file_path)
```

```python

# --- Step 1: Print the head of the original DataFrame ---
print("--- Original DataFrame Head ---")
print(df.head())
print("\n" + "="*50 + "\n")

# --- Step 2: Compute the new financial ratios ---
# Debt-to-Income Ratio (DTI)
df['DTI'] = df['Total_Debt'] / df['Annual_Income']

# Loan-to-Income Ratio
df['Loan_to_Income_Ratio'] = df['Loan_Amount'] / df['Annual_Income']

# Credit Utilization Ratio
df['Credit_Utilization'] = df['Credit_Card_Balance'] / df['Credit_Limit']

# --- Step 3: Print the head of the updated DataFrame with new features ---
print("--- DataFrame Head with New Features ---")
# Displaying all columns to see the newly added ones
pd.set_option('display.max_columns', None)
print(df.head())

```
