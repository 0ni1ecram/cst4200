```python
import pandas as pd

# --- Step 1: Read the CSV Data ---
# Read the data from the specified CSV file.
df = pd.read_csv('/content/drive/MyDrive/MScDS&AI/FeatureEngineering/5.customer_purchases.csv')


# --- Step 2: Convert to Datetime and Sort ---
# Convert timestamp string to datetime objects, specifying the exact format
df['timestamp'] = pd.to_datetime(df['timestamp'], format='%d-%m-%y %H:%M')
print(df['timestamp'].head())
```

```python

# Sort data to correctly calculate time since last purchase
df = df.sort_values(by=['customer_id', 'timestamp']).reset_index(drop=True)

```

```python

# --- Step 3: Generate New Features ---

# a) Temporal Features
df['hour'] = df['timestamp'].dt.hour
df['minute'] = df['timestamp'].dt.minute

# b) Cyclical Features
df['day_of_week'] = df['timestamp'].dt.day_name()
df['month'] = df['timestamp'].dt.month_name()

# c) Event-Based Features
df['is_weekend'] = df['timestamp'].dt.dayofweek >= 5 # Monday=0, Sunday=6

# For holidays, you would define a list of holiday dates
# Example: holidays = ['2025-09-23']
holidays = [] # No major public holidays in UAE in Sept 2025 on weekdays
df['is_holiday'] = df['timestamp'].dt.date.astype(str).isin(holidays)

# d) Relational Feature: Time Since Last Purchase (per customer)
# This calculates the difference between the current and previous purchase time for each customer
df['time_since_last_purchase'] = df.groupby('customer_id')['timestamp'].diff()


# --- Step 4: Display the Enhanced DataFrame ---
print("--- DataFrame with Newly Engineered Features ---")

print(df)

```
