```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# --- Step 1: Sample Review Data ---
# In a real-world scenario, this would be a column from your DataFrame.
# For example: df['review_text']
corpus = [
    'The pizza was great, and the service was good.',
    'The service was very slow, but the food was great.',
    'I loved the sushi, the best sushi I have ever had.',
    'The pizza was just okay, not great.',
    'Good food, good service, great experience overall.'
]

print(corpus)
```

```python

# --- Step 2: Create and Fit the TfidfVectorizer ---
# TfidfVectorizer will handle tokenizing, counting, and calculating the scores.
vectorizer = TfidfVectorizer()

# The fit_transform() method learns the vocabulary and calculates the TF-IDF scores.
# The result is a sparse matrix, which is efficient for storage.
tfidf_matrix = vectorizer.fit_transform(corpus)

```

```python

# --- Step 3: Display the Results in a DataFrame ---
# To make the results human-readable, we'll convert the sparse matrix to a DataFrame.
# The columns will be the unique words (the "vocabulary"), and the rows will be our documents.
feature_names = vectorizer.get_feature_names_out()
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)

print("--- TF-IDF Matrix ---")
print("Each row represents a review, and each column represents a word.")
print("The values are the TF-IDF scores for that word in that review.")
print(df_tfidf.round(2)) # Rounding for cleaner display
```
