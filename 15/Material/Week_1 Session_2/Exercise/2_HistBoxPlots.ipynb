```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

path='/content/drive/MyDrive/MScDS&AI/DataPreprocessing-I/'
#Read the data back from the CSV file
df = pd.read_csv(path+'1.salaryData.csv')
display(df)


# Histogram for the 'age' column
sns.histplot(df['age'], kde=True) # kde adds a density curve
plt.title('Age Distribution')
plt.show()

# Box plot for the 'salary' column
sns.boxplot(x=df['salary'])
plt.title('Salary Distribution')
plt.show()


```
