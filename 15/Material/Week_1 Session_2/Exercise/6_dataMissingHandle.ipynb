```python
import pandas as pd
import numpy as np
path='/content/drive/MyDrive/MScDS&AI/DataPreprocessing-I/'
df = pd.read_csv(path+'5.data_missing_val.csv')

# --- Option 1: Deletion ---
# Drop all rows that have any missing value
df_dropped = df.dropna()
print(f"Original rows: {len(df)}, After dropna: {len(df_dropped)}")

# --- Option 2: Imputation ---
# Make a fresh copy to work with for imputation
df_imputed = df.copy()

# Fill missing 'salary' with the median salary
median_salary = df_imputed['salary'].median()
df_imputed['salary'].fillna(median_salary, inplace=True)

# Fill missing 'age' with the mean age
mean_age = df_imputed['age'].mean()
df_imputed['age'].fillna(mean_age, inplace=True)

# Verify there are no more missing values
print("\nMissing values after imputation:")
print(df_imputed.isnull().sum())

# Check for missing values
missing_counts = df.isnull().sum()
print("Missing values per column:")
print(missing_counts[missing_counts > 0])
```
