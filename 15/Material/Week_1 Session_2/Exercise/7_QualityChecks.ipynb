```python
import pandas as pd
import numpy as np
import re
from datetime import datetime
path='/content/drive/MyDrive/MScDS&AI/DataPreprocessing-I/'
df=pd.read_csv(path+'7.messy_employee_data.csv')


print("--- Identifying Data Quality Issues ---")

# 1. Count missing/null values per column
print("\n1. Count of Missing/Null Values per Column:")
print(df.isnull().sum())

# 2. Check for inconsistent formats (categorical labels & typos)
print("\n2. Unique Values in 'location' (showing inconsistencies):")
print(df['location'].unique())

# 3. Use validation rules (age >= 18)
print("\n3. Validation Rule Checks:")
# Rule: Age must be a positive number, let's say >= 18
invalid_age = df[df['age'] < 18]
print(f"\nRows with invalid age (< 18):\n{invalid_age[['employee_id', 'age']]}")

# 4. Spot impossible values (negative salary, future dates)
print("\n4. Impossible Value Checks:")
# Spot negative salaries
negative_salary = df[df['salary'] < 0]
print(f"\nRows with negative salary:\n{negative_salary[['employee_id', 'salary']]}")

# Spot future hire dates
# First, convert date strings to datetime objects, coercing errors
df['hire_date_dt'] = pd.to_datetime(df['hire_date'], errors='coerce')
future_dates = df[df['hire_date_dt'] > datetime.now()]
print(f"\nRows with future hire dates:\n{future_dates[['employee_id', 'hire_date']]}")

# 5. & 6. Inconsistent Formatting and Capitalization are already shown in step 2
print("\nUnique Values in 'location' (showing inconsistencies):")
print(df['location'].unique())

```
