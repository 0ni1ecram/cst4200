```python
import pandas as pd
import numpy as np

path='/content/drive/MyDrive/MScDS&AI/DataPreprocessing-I/'
df=pd.read_csv(path+'Real_estate.csv')

print("--- Successfully loaded 'Real_estate.csv' ---")
print(f"Original DataFrame shape: {df.shape}")

# --- Step 1: Remove Duplicate Values ---

num_duplicates_before = df.duplicated().sum()
print(f"Number of duplicate rows found: {num_duplicates_before}")

if num_duplicates_before > 0:
    # Remove duplicates, keeping the first occurrence
    df_cleaned = df.drop_duplicates(keep='first', inplace=False)
    print(f"Removed {num_duplicates_before} duplicate rows.")
else:
    print("No duplicate rows to remove.")
    df_cleaned = df.copy()

```

```python
# --- Step 2: Perform Data Profiling ---
print("--- Step 2: Perform Data Profiling ---")
print("Generating a profile of the cleaned data...")

print("\n1. Data Shape (Rows, Columns):")
print(df_cleaned.shape)

print("\n2. Data Types, Column Names, and Non-Null Counts:")
# .info() provides a concise summary of the DataFrame's structure
df_cleaned.info()

print("\n3. Descriptive Statistics for Numerical Columns:")
print(df_cleaned.describe().round(2))
print("\n" + "="*50 + "\n")
```

```python

# --- Step 3: Outlier Detection using the Interquartile Range (IQR) Rule ---
#'Y house price of unit area' column

# Define the target column for outlier detection
col = 'Y house price of unit area'

# Calculate Q1 (25th percentile) and Q3 (75th percentile) for the target column
Q1 = df_cleaned[col].quantile(0.25)
Q3 = df_cleaned[col].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Define the lower and upper bounds for outlier detection
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify any data points that fall outside the bounds
outliers = df_cleaned[(df_cleaned[col] < lower_bound) | (df_cleaned[col] > upper_bound)]

if not outliers.empty:
    print(f"\nOutliers found in '{col}':")
    print(f"  - IQR: {IQR:.2f}")
    print(f"  - Bounds: Lower < {lower_bound:.2f}, Upper > {upper_bound:.2f}")
    print(f"  - Number of outliers detected: {len(outliers)}")
    # Displaying up to the first 5 detected outlier rows for brevity
    print("  - Example outlier rows:")
    print(outliers.head())
else:
    print(f"\nNo outliers found in '{col}'.")

print("\n" + "="*50 + "\n")
print("Data analysis script finished.")

```
