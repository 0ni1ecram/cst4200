```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostRegressor, GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_squared_error

# --- 1. Read the CSV file ---
# Load the dataset using a semicolon delimiter
df = pd.read_csv('/content/drive/MyDrive/MScDS&AI/EnsembleRandomForest/student-mat.csv', sep=';')
print("--- CSV file loaded successfully! ---")
print(f"Dataset shape: {df.shape}\n")


# --- 2. Display columns and data types ---
print("--- Columns and their Data Types ---")
print(df.info())
print("\n")
```

```python
# --- 3. Separate Input Features (X) and Target Variable (y) ---
# The target variable is 'G3', the final grade.
# All other columns will be used as input features.
X = df.drop('G3', axis=1)
y = df['G3']

print(f"--- Features (X) and Target (y) separated ---")
print(f"Shape of X: {X.shape}")
print(f"Shape of y: {y.shape}\n")
```

```python
# --- 4. Encode Categorical Features ---
# pandas.get_dummies() will automatically find and encode
# columns with 'object' or 'category' data types.
X_encoded = pd.get_dummies(X, drop_first=True)

print("--- Categorical features encoded ---")
print(f"Shape of X after encoding: {X_encoded.shape}")
print("New columns created from categorical features:")
# Show columns that were added by get_dummies
original_cols = set(X.columns)
encoded_cols = set(X_encoded.columns)
new_cols = list(encoded_cols - original_cols)
print(new_cols[:10]) # Print first 10 new columns
print("\n")
```

```python

# --- 5. Perform Train-Test Split (80% training, 20% testing) ---
# We use a random_state for reproducibility of the split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

print("--- Data split into training and testing sets ---")
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}\n")
```

```python
# --- 6. Fit AdaBoost Regressor and Evaluate ---
print("--- Training AdaBoost Regressor ---")
# Initialize the model with a random state for reproducibility
adaboost_model = AdaBoostRegressor(n_estimators=100, random_state=42)

# Train the model on the training data
adaboost_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_ada = adaboost_model.predict(X_test)

# Calculate R2 score and Mean Squared Error (MSE)
r2_ada = r2_score(y_test, y_pred_ada)
mse_ada = mean_squared_error(y_test, y_pred_ada)

print("AdaBoost Model Evaluation:")
print(f"R-squared (R2) Score: {r2_ada:.4f}")
print(f"Mean Squared Error (MSE): {mse_ada:.4f}\n")
```

```python
# --- 7. Fit Gradient Boosting Regressor and Evaluate ---
print("--- Training Gradient Boosting Regressor ---")
# Initialize the model with a random state
gradient_boost_model = GradientBoostingRegressor(n_estimators=100, random_state=42)

# Train the model
gradient_boost_model.fit(X_train, y_train)

# Make predictions
y_pred_gb = gradient_boost_model.predict(X_test)

# Calculate R2 score and MSE
r2_gb = r2_score(y_test, y_pred_gb)
mse_gb = mean_squared_error(y_test, y_pred_gb)

print("Gradient Boosting Model Evaluation:")
print(f"R-squared (R2) Score: {r2_gb:.4f}")
print(f"Mean Squared Error (MSE): {mse_gb:.4f}\n")
```
