```python
import pandas as pd

# Load the dataset
file_path = '/content/drive/MyDrive/MScDS&AI/ANN/Data.csv'  # Replace with the correct path if necessary
data = pd.read_csv(file_path)

# Initialize weights and parameters
w1 = -0.5
w2 = 0.5
alpha = 0.3  # Learning rate

# Hebbian learning for each row, explicitly
# Row 1
F1, F2, label = data.iloc[0]['F1'], data.iloc[0]['F2'], data.iloc[0]['Label']
weighted_sum = w1 * F1 + w2 * F2
print(f"Row 1: Weighted sum = {weighted_sum}")
step_output = 1 if weighted_sum >= 0 else 0
print(f"Row 1: Step activation output = {step_output}")
error = label - step_output
if error != 0:
    w1 += alpha * error * F1
    w2 += alpha * error * F2
    print(f"Row 1: Weights updated -> w1 = {w1}, w2 = {w2}")
else:
    print("Row 1: No weight update required.")
```
