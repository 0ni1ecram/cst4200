```python
import pandas as pd

# Load the dataset
file_path = '/content/drive/MyDrive/MScDS&AI/ANN/Data.csv'  # Replace with the correct path if necessary
data = pd.read_csv(file_path)

# Initialize weights and parameters
w1 = -0.5
w2 = 0.5
alpha = 0.3  # Learning rate

# Hebbian learning
print("Starting Hebbian learning:")
for index, row in data.iterrows():
    # Extract features and label
    F1 = row['F1']
    F2 = row['F2']
    label = row['Label']

    # Calculate weighted sum
    weighted_sum = w1 * F1 + w2 * F2
    print(f"Row {index+1}: Weighted sum = {weighted_sum}")

    # Step activation function
    step_output = 1 if weighted_sum >= 0 else 0
    print(f"Row {index+1}: Step activation output = {step_output}")

    # Check if label matches the step output
    error = label - step_output  # Error calculation
    if error != 0:
        # Update weights if there is an error
        w1_new = w1 + alpha * error * F1
        w2_new = w2 + alpha * error * F2
        print(f"Row {index+1}: Weights updated -> w1 = {w1_new}, w2 = {w2_new}")
        w1, w2 = w1_new, w2_new  # Assign new weights
    else:
        print(f"Row {index+1}: No weight update required.")

    print()  # Blank line for readability

# Final weights
print(f"Final weights: w1 = {w1}, w2 = {w2}")
```
