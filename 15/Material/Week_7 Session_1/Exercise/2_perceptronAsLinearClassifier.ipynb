```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the dataset
data = pd.read_csv("/content/drive/MyDrive/MScDS&AI/ANN/Data.csv")

# Define weights and bias
w1 = 0.5  # Weight for F1
w2 = 0.1  # Weight for F2
bias = 0  # Bias term

# Single calculation for the first row
first_row = data.iloc[0]
weighted_sum = w1 * first_row['F1'] + w2 * first_row['F2'] + bias
print(f"Weighted sum for the first row: {weighted_sum}")
activation_op = 1 if weighted_sum >= 0 else 0
print(f"After activation function: {activation_op}")
```

```python

# Calculate perceptron output for all rows
# Step 1: Calculate the weighted sum
data['Weighted_Sum'] = data['F1'] * w1 + data['F2'] * w2 + bias

# Step 2: Apply the step activation function
data['Perceptron_Output'] = data['Weighted_Sum'].apply(lambda x: 1 if x >= 0 else 0)
```

```python

# Visualize the data and the decision boundary
plt.figure(figsize=(8, 6))

# Plot data points with different labels
for label, group in data.groupby('Label'):
    plt.scatter(group['F1'], group['F2'], label=f'Class {label}')

# Plot the decision boundary
x_values = np.linspace(data['F1'].min(), data['F1'].max(), 100)
y_values = -(w1 / w2) * x_values - (bias / w2)
plt.plot(x_values, y_values, color='red', label='Decision Boundary')

plt.xlabel('F1')
plt.ylabel('F2')
plt.title('Perceptron as a Linear Classifier')
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')
plt.legend()
plt.grid(True)
plt.show()

# Display the updated DataFrame
print(data)
```
