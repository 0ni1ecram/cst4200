```python
# 2. Load and preProcess an input image from local file path
img_path = "F:/Users/MSR/Downloads/FULL.jpg"

print(f"Loading form: {img_path}")

# Open image using PIL (python Imaging Library)
original_image = Image.open(img_path).convert('RGB')

# Resize to Target Size VGG16 expects (224 x 224)
target_size = (224, 224)
resized_omg = original_image.resize(target_size)

# Convert the PIL image to a NumPy Array
img_array = image.img_to_array(resized_omg)

# Adding a fourth dimension, batch size of 1
img_array_expanded = np.expand_dims(img_array, axis=0)

# PreProcess the image for VGG16 (handles color channel ordering and normalization)
preprocessed_img = preprocess_input(img_array_expanded)
```

```python
# 3. Make a Prediction
print("Classify the image...")
predictions = model.predict(preprocessed_img)

# 4. Decode prediction into human readable labels
# This function converts the model's output vector into names
decoded_predictions = decode_predictions(predictions, top=1)[0]

# Get the single top predictions
_, label, score = decoded_predictions[0]

print("\n---- Top Predictions ----- ")
prediction_text = f"{label}({score:.2%})"
print(prediction_text)

# 5. Visualize the result
plt.figure(figsize=(8,6))
plt.imshow(original_image)
plt.title(f"Prediction: {prediction_text}", fontsize=14)
plt.xticks([])
plt.yticks([])
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import matplotlib.pyplot as plt
from PIL import Image
print("Loading VGG16 model image NetWeights...")

# LOAD THE Pre-Trained VGG16 model
# The model downloaded automatically on the first run.
model  = VGG16(weights="imagenet")   
# 2. Load and preProcess an input image from local file path
img_path = "F:/Users/MSR/Downloads/FULL.jpg"

print(f"Loading form: {img_path}")

# Open image using PIL (python Imaging Library)
original_image = Image.open(img_path).convert('RGB')

# Resize to Target Size VGG16 expects (224 x 224)
target_size = (224, 224)
resized_omg = original_image.resize(target_size)

# Convert the PIL image to a NumPy Array
img_array = image.img_to_array(resized_omg)

# Adding a fourth dimension, batch size of 1
img_array_expanded = np.expand_dims(img_array, axis=0)

# PreProcess the image for VGG16 (handles color channel ordering and normalization)
preprocessed_img = preprocess_input(img_array_expanded)

# 3. Make a Prediction
print("Classify the image...")
predictions = model.predict(preprocessed_img)

# 4. Decode prediction into human readable labels
# This function converts the model's output vector into names
decoded_predictions = decode_predictions(predictions, top=1)[0]

# Get the single top predictions
_, label, score = decoded_predictions[0]

print("\n---- Top Predictions ----- ")
prediction_text = f"{label}({score:.2%})"
print(prediction_text)

# 5. Visualize the result
plt.figure(figsize=(8,6))
plt.imshow(original_image)
plt.title(f"Prediction: {prediction_text}", fontsize=14)
plt.xticks([])
plt.yticks([])
plt.tight_layout()
plt.show()
```
