```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram

# --- Main Script ---

# Sample data: 5 points in a 2D space (0 to 20) with pre-assigned values
X = np.array([[4, 5], [10, 12], [3, 15], [16, 17], [14, 6]], dtype=float)

# Point labels for the dendrogram
point_labels = [f'P{i}' for i in range(len(X))]

# --- Perform Hierarchical Clustering ---
# The linkage function calculates the distances between clusters and records the merge history.
# Method 'single' uses the minimum distance between points in the two clusters (nearest neighbor).

Z = linkage(X,method='single', metric='euclidean')

# --- Create and Display the Dendrogram ---
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(10, 7))

# The dendrogram function takes the linkage matrix Z and plots the hierarchy.
dendrogram(
    Z,
    labels=point_labels,
    orientation='top',  # Can be 'top', 'bottom', 'left', 'right'
    ax=ax
)

# --- Customize the Plot ---
ax.set_title('Hierarchical Clustering Dendrogram', fontsize=16)
ax.set_xlabel('Data Points', fontsize=12)
ax.set_ylabel('Euclidean Distance', fontsize=12)

# Add a threshold line (optional) to show a potential cut-off for forming clusters
# For example, a distance of 10
# ax.axhline(y=10, color='r', linestyle='--')

plt.tight_layout()
plt.show()
```
