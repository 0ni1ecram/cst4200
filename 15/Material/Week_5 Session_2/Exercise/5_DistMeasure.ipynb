```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# --- 1. Euclidean Distance (L2 Norm) ---
# The straight-line distance between two points in Euclidean space.
# Formula: sqrt(sum((p_i - q_i)^2))

print("--- 1. Euclidean Distance (L2 Norm) ---")
point_a = np.array([2, 3])
point_b = np.array([8, 11])

# Calculate Euclidean distance using NumPy's norm function
euclidean_dist = np.linalg.norm(point_a - point_b)

print(f"Point A: {point_a}")
print(f"Point B: {point_b}")
print(f"Euclidean Distance: {euclidean_dist:.4f}")
print("This is the 'as the crow flies' distance between the two points.\n")
#

# --- 2. Manhattan Distance (L1 Norm) ---
# The distance between two points measured along the axes at right angles.
# Also known as "city block distance".
# Formula: sum(|p_i - q_i|)

print("--- 2. Manhattan Distance (L1 Norm) ---")
point_c = np.array([2, 3])
point_d = np.array([8, 11])

# Calculate Manhattan distance
manhattan_dist = np.sum(np.abs(point_c - point_d))

print(f"Point C: {point_c}")
print(f"Point D: {point_d}")
print(f"Manhattan Distance: {manhattan_dist:.4f}")
print("This is the distance you would travel in a city grid to get from C to D.\n")
#

# --- 3. Cosine Similarity ---
# Measures the cosine of the angle between two non-zero vectors.
# It determines the orientation, not the magnitude, of two vectors.
# Result ranges from -1 (exactly opposite) to 1 (exactly the same). 0 indicates orthogonality.

print("--- 3. Cosine Similarity ---")

# Sample text documents
text_1 = "The cat sat on the mat."
text_2 = "The dog sat on the log."
text_3 = "A cat and a dog are friends."

documents = [text_1, text_2, text_3]
print("Comparing the following documents:")
print(f'  Doc 1: "{text_1}"')
print(f'  Doc 2: "{text_2}"')
print(f'  Doc 3: "{text_3}"\n')

# Step 1: Convert text to numerical vectors
# We use a TfidfVectorizer to create vectors based on term frequency-inverse document frequency.
# This gives more weight to more important words.
# First, create an instance of the vectorizer
tfidf_vectorizer = TfidfVectorizer()
# Then, fit and transform the documents
vectors_sparse = tfidf_vectorizer.fit_transform(documents)
vectors = vectors_sparse.toarray()


# Step 2: Calculate Cosine Similarity
# We compare the first document to the second and the first to the third.
cosine_sim_1_2 = cosine_similarity(vectors[0].reshape(1, -1), vectors[1].reshape(1, -1))
cosine_sim_1_3 = cosine_similarity(vectors[0].reshape(1, -1), vectors[2].reshape(1, -1))

print("Cosine Similarity Results:")
print(f"Similarity between Doc 1 and Doc 2: {cosine_sim_1_2[0][0]:.4f}")
print("  -> These texts are quite similar because they share many of the same words ('the', 'sat', 'on').")

print(f"Similarity between Doc 1 and Doc 3: {cosine_sim_1_3[0][0]:.4f}")
print("  -> These texts are less similar as they share fewer common words.")
print("\nA similarity of 1 means the texts are identical, and 0 means they have no words in common.")

```
