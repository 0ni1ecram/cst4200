```python
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score

# --- 1. Load and Inspect Data ---
print("--- 1. Loading and Inspecting Data ---")
# Load the dataset with the correct semicolon delimiter
df = pd.read_csv('/content/drive/MyDrive/MScDS&AI/HierarchicalClustering/student-mat.csv', sep=';')
print("Dataset loaded successfully.")

# Display initial information about all columns
print("\nInitial DataFrame Info:")
df.info()

```

```python


# --- 2. Data Preprocessing ---
print("\n--- 2. Preprocessing Data ---")
# Create a copy for preprocessing to keep the original data intact
df_processed = df.copy()

# Convert categorical features to one-hot encoding
# The 'get_dummies' function automatically handles columns with object or category dtypes
df_processed = pd.get_dummies(df_processed)
print(f"Shape after one-hot encoding: {df_processed.shape}")

# Scale all features using StandardScaler
# This is crucial for distance-based algorithms like agglomerative clustering
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df_processed)
print("All features have been scaled using StandardScaler.")
```

```python

# --- 3. Initial 3D Visualization ---
print("\n--- 3. Displaying Initial 3D Scatter Plot ---")
# Use the original (unscaled) G1, G2, G3 columns for better interpretability
fig_before = px.scatter_3d(
    df,
    x='G1',
    y='G2',
    z='G3',
    title='Initial 3D Scatter Plot of Student Grades (G1, G2, G3)',
    labels={'G1': 'First Period Grade', 'G2': 'Second Period Grade', 'G3': 'Final Grade'},
    opacity=0.8
)
fig_before.update_traces(marker=dict(size=5))
fig_before.show()
print("The initial 3D scatter plot has been generated.")
```

```python

# --- 4. Agglomerative Clustering ---
print("\n--- 4. Performing Agglomerative Clustering ---")
# We will create 3 clusters as requested
# The model is fitted on the fully preprocessed and scaled data
# Note: 'affinity' is not used with 'ward' linkage as it implies euclidean distance.
agg_cluster = AgglomerativeClustering(n_clusters=3, linkage='ward') #change to single, complete, average, ward and check scores.
cluster_labels = agg_cluster.fit_predict(scaled_features)

# Add the cluster labels to the original DataFrame for visualization
df['cluster'] = cluster_labels
print("Agglomerative clustering complete. 3 clusters have been formed.")

```

```python


# --- 5. Post-Clustering Visualization ---
print("\n--- 5. Displaying 3D Scatter Plot with Clusters ---")
# Create a 3D scatter plot and color the points by their assigned cluster
fig_after = px.scatter_3d(
    df,
    x='G1',
    y='G2',
    z='G3',
    color='cluster',
    title='3D Scatter Plot of Grades with 3 Clusters',
    labels={'G1': 'First Period Grade', 'G2': 'Second Period Grade', 'G3': 'Final Grade'},
    opacity=0.9
)
fig_after.update_traces(marker=dict(size=5))
fig_after.show()
print("The 3D scatter plot with cluster colors has been generated.")

```

```python


# --- 6. Cluster Evaluation ---
print("\n--- 6. Evaluating Cluster Performance ---")
# Calculate evaluation metrics using the scaled data and the predicted labels

# Silhouette Score: Measures how similar an object is to its own cluster compared to other clusters.
# The best value is 1 and the worst value is -1. Values near 0 indicate overlapping clusters.
silhouette = silhouette_score(scaled_features, cluster_labels)
print(f"Silhouette Score: {silhouette:.4f}")

# Davies-Bouldin Index: The average similarity measure of each cluster with its most similar cluster.
# Lower values indicate better clustering. The minimum score is 0.
davies_bouldin = davies_bouldin_score(scaled_features, cluster_labels)
print(f"Davies-Bouldin Index: {davies_bouldin:.4f}")


```
