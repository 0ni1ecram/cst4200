```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Tips': [20, 7.5, 12.5],
    'Bill': [220, 90, 150]
}
# Create the DataFrame
df = pd.DataFrame(data)
# Display the DataFrame
print(df)
```

```python
plt.scatter(df['Bill'], df['Tips'], color='blue')

# Adding labels and title
plt.xlabel('Bill')
plt.ylabel('Tips')
plt.title('Scatter Plot of Tips vs Bill')

# Display the plot
plt.show()
```

```python
y = np.array(data['Tips'])
x = np.array(data['Bill'])
b0 =-0.75
b1 =-1.25
y_pred = b0 + b1 * x
print(y_pred)
loss = np.sum((y - y_pred) ** 2)
print(loss)
```

```python
y = np.array(data['Tips'])
x = np.array(data['Bill'])

# b0 and b1 values
b_values = np.array([
    [-0.75, -1.25],
    [-0.5, -0.75],
    [-0.25, -0.25],
    [0, 0.25],
    [0.25, 0.75],
    [0.5, 1.25],
    [0.75, 1.75]
])

# Calculate the loss for each combination of b0 and b1
loss_table = []
for b0, b1 in b_values:
    y_pred = b0 + b1 * x
    loss = np.sum((y - y_pred) ** 2)
    loss_table.append([b0, b1, loss])

# Convert to DataFrame for better visualization
loss_df = pd.DataFrame(loss_table, columns=["b0", "b1", "Loss"])

# Display the DataFrame
print(loss_df)
```
