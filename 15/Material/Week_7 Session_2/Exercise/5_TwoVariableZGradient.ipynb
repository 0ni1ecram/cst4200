```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Initialize (x, y), learning rate, and number of steps
x, y = 8, 8
learning_rate = 0.05
steps = 5

# Define gradients of the function Z = 2x^2 + 2y^2 - 10
# Gradient with respect to x: dZ/dx = 4x
# Gradient with respect to y: dZ/dy = 4y
def gradient_x(x):
    return 4 * x

def gradient_y(y):
    return 4 * y

z = 2 * x**2 + 2 * y**2 - 10
grad_x = gradient_x(x)
grad_y = gradient_y(y)
print(f'z={z}, grad_x={grad_x}, grad_y={grad_y}')

```

```python
# Create a list to store the data for the DataFrame
data = []

# Perform gradient descent for the specified number of steps
for step in range(steps):
    z = 2 * x**2 + 2 * y**2 - 10
    grad_x = gradient_x(x)
    grad_y = gradient_y(y)
    new_x = x - learning_rate * grad_x
    new_y = y - learning_rate * grad_y
    new_z = 2 * new_x**2 + 2 * new_y**2 - 10
    data.append([x, y, z, grad_x, grad_y, new_x, new_y, new_z])
    x, y = new_x, new_y  # Update x and y for the next step

# Create a DataFrame
columns = ["x", "y", "z", "Grad. x", "Grad. y", "New x ", "New y ", "New z "]
df = pd.DataFrame(data, columns=columns)

# Display the DataFrame
print(df)
```

```python

# Plot the function and the gradient descent steps
x_range = np.linspace(-10, 10, 100)
y_range = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x_range, y_range)
Z = 2 * X**2 + 2 * Y**2 - 10

plt.figure(figsize=(10, 6))
contour = plt.contour(X, Y, Z, levels=30, cmap="viridis")
plt.colorbar(contour)
plt.scatter(df["x"], df["y"], color="red", label="Gradient Descent Steps")
plt.title("Gradient Descent on Z = 2x^2 + 2y^2 - 10")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
