```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Input data
data = {
    'Tips': [20, 7.5, 12.5],
    'Bill': [220, 90, 150]
}
y = np.array(data['Tips'])
x = np.array(data['Bill'])
```

```python

# Define b0 and b1 ranges for a grid
b0_range = np.linspace(-0.75, 0.75, 50)
b1_range = np.linspace(-1.25, 1.75, 50)
b0_grid, b1_grid = np.meshgrid(b0_range, b1_range)
```

```python

# Calculate the loss for each combination of b0 and b1
loss_grid = np.zeros_like(b0_grid)
for i in range(b0_grid.shape[0]):
    for j in range(b0_grid.shape[1]):
        b0 = b0_grid[i, j]
        b1 = b1_grid[i, j]
        y_pred = b0 + b1 * x
        loss_grid[i, j] = np.sum((y - y_pred) ** 2)

# 3D Surface Plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surf = ax.plot_surface(b0_grid, b1_grid, loss_grid, cmap='viridis', edgecolor='none')

# Add labels and title
ax.set_title("3D Surface Plot of Loss")
ax.set_xlabel("b0")
ax.set_ylabel("b1")
ax.set_zlabel("Loss")

# Add color bar
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

plt.show()

# Display the grid dimensions
print("Loss grid dimensions:", loss_grid.shape)
```
