```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D   # noqa: F401
from sklearn.preprocessing import StandardScaler

CSV_PATH = "wholesale.csv"
UCI_URL  = ("https://archive.ics.uci.edu/ml/machine-learning-databases/"
            "00292/Wholesale%20customers%20data.csv")

if not os.path.exists(CSV_PATH):
    df = pd.read_csv(UCI_URL); df.to_csv(CSV_PATH, index=False)
else:
    df = pd.read_csv(CSV_PATH)
df = df.rename(columns={"Delicatessen": "Delicassen"})

SEED = 63
rng  = np.random.default_rng(SEED)
```

```python
# ---- 3-D features so the plot is meaningful ----
FEATURES_3D = ["Grocery", "Detergents_Paper", "Fresh"]
X3_raw = df[FEATURES_3D].to_numpy(dtype=np.float32)
X3     = StandardScaler().fit_transform(np.log1p(X3_raw)).astype(np.float32)

# Pick one real customer as x, and a *different* real customer as w2_t.
i_x, i_w = rng.choice(len(X3), size=2, replace=False)
x          = X3[i_x]              # input vector = one wholesale customer
w2_initial = X3[i_w].copy()       # "BMU weight" = another wholesale customer

alpha  = 0.5
h_bmu  = 1.0

print("----- SOM Single Weight Update Step from the BMU -----")
print(f"Input vector x           (row {i_x}): {np.round(x, 3)}")
print(f"Initial BMU weight w2_t  (row {i_w}): {np.round(w2_initial, 3)}")
print(f"Learning rate alpha:       {alpha}")
print(f"Neighbourhood h_bmu:       {h_bmu}")

# --- Step 1: difference vector ---
diff = x - w2_initial
print("\nStep 1: (x - w2_t) =", np.round(diff, 3))

# --- Step 2: update amount ---
update_amount = alpha * h_bmu * diff
print(f"\nStep 2: alpha * h * (x - w2_t) = "
      f"{alpha} * {h_bmu} * {np.round(diff, 3)} = {np.round(update_amount, 3)}")

# --- Step 3: new weight ---
w2_new = w2_initial + update_amount
print(f"\nStep 3: w2_t+1 = w2_t + update = "
      f"{np.round(w2_initial, 3)} + {np.round(update_amount, 3)} = {np.round(w2_new, 3)}")
print("-" * 55)
print("The BMU's weight vector has moved closer to the input vector.")

# Sanity: distance shrank?
d_before = np.linalg.norm(x - w2_initial)
d_after  = np.linalg.norm(x - w2_new)
print(f"Distance |x - w2_t|   = {d_before:.4f}")
print(f"Distance |x - w2_t+1| = {d_after:.4f}   (smaller is correct)")

# --- 4. 3-D visualization (same as your file) ---
fig = plt.figure(figsize=(10, 8))
ax  = fig.add_subplot(111, projection="3d")

ax.scatter(*x,          c="red",   marker="*", s=250,
           label=f"Input customer (row {i_x})", depthshade=False)
ax.scatter(*w2_initial, c="blue",  s=100,
           label=f"BMU initial w2_t (row {i_w})", depthshade=False)
ax.scatter(*w2_new,     c="green", s=100,
           label="BMU updated w2_t+1", depthshade=False)

ax.quiver(*w2_initial, *update_amount,
          color="black", label="Weight update vector",
          arrow_length_ratio=0.3)

ax.plot([x[0], w2_initial[0]], [x[1], w2_initial[1]], [x[2], w2_initial[2]],
        "k--", linewidth=1, alpha=0.5)

ax.set_title("BMU weight update on real Wholesale customers\n"
             f"features = {FEATURES_3D}")
ax.set_xlabel(FEATURES_3D[0]); ax.set_ylabel(FEATURES_3D[1]); ax.set_zlabel(FEATURES_3D[2])
ax.legend()
plt.grid(True); plt.tight_layout(); plt.show()
```

```python
SPEND = ["Fresh", "Milk", "Grocery", "Frozen", "Detergents_Paper", "Delicassen"]
X6_raw = df[SPEND].to_numpy(dtype=np.float32)
X6     = StandardScaler().fit_transform(np.log1p(X6_raw)).astype(np.float32)

# x = a grocery-heavy customer, w2_t = a fresh-heavy customer (deliberately different)
gro_idx   = np.argmax(df["Grocery"].to_numpy())
fresh_idx = np.argmax(df["Fresh"].to_numpy())

x          = X6[gro_idx].copy()
w2_initial = X6[fresh_idx].copy()

alpha, h_bmu = 0.5, 1.0

diff          = x - w2_initial
update_amount = alpha * h_bmu * diff
w2_new        = w2_initial + update_amount

print("\n----- 6-D single BMU update on Wholesale -----")
print(f"x         (grocery-heavy customer, row {gro_idx})   :", np.round(x,          2))
print(f"w2_t      (fresh-heavy customer,   row {fresh_idx}) :", np.round(w2_initial, 2))
print(f"update    = alpha*h*(x - w2_t)                      :", np.round(update_amount, 2))
print(f"w2_t+1                                              :", np.round(w2_new,      2))

d_before = np.linalg.norm(x - w2_initial)
d_after  = np.linalg.norm(x - w2_new)
print(f"\n|x - w2_t|   = {d_before:.4f}")
print(f"|x - w2_t+1| = {d_after:.4f}   (should be {alpha:.2f}× the before distance)")

# ---- Diagnostic bar chart: before vs after, per feature ----
fig, ax = plt.subplots(figsize=(10, 5))
idx  = np.arange(len(SPEND))
wbar = 0.38
ax.bar(idx - wbar/2, w2_initial, wbar, label="w2_t   (initial)",  color="tab:blue")
ax.bar(idx + wbar/2, w2_new,     wbar, label="w2_t+1 (updated)", color="tab:green")
ax.axhline(0, color="k", lw=0.6)

# overlay the target x as red dots so you can see the move *toward* x
ax.scatter(idx, x, color="tab:red", zorder=3, label="target x", marker="*", s=160)

ax.set_xticks(idx); ax.set_xticklabels(SPEND, rotation=20)
ax.set_ylabel("standardized log1p(spend)")
ax.set_title(f"BMU update in 6-D  (alpha*h = {alpha*h_bmu:.2f})")
ax.legend(); ax.grid(True, axis="y")
plt.tight_layout(); plt.show()
```

```python
# ---- grid of neurons ----
grid_h, grid_w = 5, 5
N, D = grid_h * grid_w, X6.shape[1]
neurons  = X6[rng.choice(len(X6), size=N, replace=False)].copy()  # (25, 6)
grid_pos = np.array([(r, c) for r in range(grid_h) for c in range(grid_w)])

def grid_dists(bmu):
    return np.linalg.norm(grid_pos - grid_pos[bmu], axis=1)

EPOCHS = 2000
lr0, lr1   = 0.5, 0.01
sig0, sig1 = 2.5, 0.3

for epoch in range(EPOCHS):
    t  = epoch / EPOCHS
    lr = lr0 * (1 - t) + lr1 * t
    sg = sig0 * (1 - t) + sig1 * t

    for i in rng.permutation(len(X6)):
        x = X6[i]

        # ---- find BMU (the `min(distances, key=...)` step from your first file) ----
        d   = np.linalg.norm(neurons - x, axis=1)
        bmu = int(d.argmin())

        # ---- update  <-- this IS your file's math, vectorized over neighbours ----
        h = np.exp(-(grid_dists(bmu) ** 2) / (2 * sg * sg))
        neurons += lr * h[:, None] * (x - neurons)     # matches alpha*h*(x - w)

    if epoch % 400 == 0:
        qe = np.mean(np.min(np.linalg.norm(X6[:, None, :] - neurons[None], axis=2),
                            axis=1))
        print(f"epoch {epoch:4d}  lr={lr:.3f}  sigma={sg:.2f}  QE={qe:.4f}")

# ---- how many customers land on each neuron, and Channel majority ----
d_all  = np.linalg.norm(X6[:, None, :] - neurons[None], axis=2)
bmu6   = d_all.argmin(axis=1)
counts = np.bincount(bmu6, minlength=N).reshape(grid_h, grid_w)
print("\nCustomers per neuron:\n", counts)

channel = df["Channel"].to_numpy()
maj = np.zeros_like(counts, dtype=float)
for k in range(N):
    m = bmu6 == k
    maj.flat[k] = np.nan if not m.any() else np.bincount(channel[m], minlength=3).argmax()

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(counts, cmap="Blues")
axes[0].set_title("Customers per neuron")
for k in range(N):
    r, c = divmod(k, grid_w)
    axes[0].text(c, r, str(counts[r, c]), ha="center", va="center")

axes[1].imshow(maj, cmap="coolwarm")
axes[1].set_title("Majority Channel per neuron\n(labels never used during training)")
plt.tight_layout(); plt.show()
```
