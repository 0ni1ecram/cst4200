```python
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

# Generate 100 data points
m = 100
X = 6 * np.random.rand(m, 1) - 3
y = 0.5 * X**2 + X + 2 + np.random.randn(m, 1)

# Plot the data
plt.scatter(X, y)
plt.title("Generated Non-Linear Data")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Transform features to include polynomial terms (degree=2)
poly_features = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly_features.fit_transform(X)

# X_poly now contains the original feature X and X^2
# X[0] -> (X[0], X[0]**2)

# Fit a linear regression model on the transformed features
lin_reg = LinearRegression()
lin_reg.fit(X_poly, y)
```

```python
# Generate points for plotting the curve
X_new = np.linspace(-3, 3, 100).reshape(100, 1)
X_new_poly = poly_features.transform(X_new)
y_new = lin_reg.predict(X_new_poly)

# Plot the original data and the polynomial fit
plt.scatter(X, y, label="Data Points")
plt.plot(X_new, y_new, "r-", linewidth=2, label="Polynomial Fit (degree=2)")
plt.title("Polynomial Regression Fit")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```
