```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10, 10, 100)
y = x
#y=x**3
plt.plot(x, y, label='y = x')
```

```python
noise = np.random.normal(0, 4, x.size)
y = x + noise
```

```python
#noise = np.random.normal(0, 20, x.size)
#y = x**3 + noise
```

```python
plt.plot(x, x, label='y = x', color='blue')
plt.scatter(x, y, color='red', label='y = x + noise', s=10)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of y = x with Added Noise')
```
