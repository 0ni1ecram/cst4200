```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D # REquired for 3D-printing

# --- 1. Define thew Input Vector and Neuron Weights --
#  Use the Excat VAlues form th example
x = np.array([0.3, 0.8, 0.1])

neuron_weights = {
    'w1': np.array([0.1, 0.2, 0.7]),
    'w2': np.array([0.4, 0.9, 0.2]),
    'w3': np.array([0.8, 0.1, 0.3]),
    'w4': np.array([0.5, 0.5, 0.5])
}

print("--- Finding the Best Matching Unit (BMU)---")
print(f"Input vector (x): {x}\n")
```

```python
# ---- 2. Calculate the Euclidean Distance from Each Neurone
distances = {}
for name, w in neuron_weights.items():
    #  Eucilidean distance is the square root of the sum of squared differences.
    # np.linalg.norm(a - b) is a convinient way to calculate this
    dist = np.linalg.norm(x - w)
    distances[name] = dist
    print(f"Distance to {name} ({w}): {dist:.4f}")
```

```python
# --- 3. Identify the best matching unit
# The BMU is the neuron with the minimum distance to the input vector
bmu_name = min(distances, key=distances.get)
bmu_weight = neuron_weights[bmu_name]
min_distances = distances[bmu_name]

print(f"\nMinimum distance is: {min_distances:.4f}.")
print(f"Therefor, the Best Matching Unit (BMU) is : {bmu_name}")
```

```python
# --- 4. Visualize the Concept in 3D
fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111, projection='3d')

# Plot the input vector as a prominint star
ax.scatter(x[0], x[1], x[2], c='red', marker='*', s=200, label='Input Vector (x)')

# Plot the neuron weight vectors
for name, w in neuron_weights.items():
    ax.scatter(w[0], w[1], w[2], c='blue', s=50, label=f'Neuron {name}')
    # Draw line from input vector to each neuron
    ax.plot([x[0], w[0]], [x[1], w[1]], [x[2], w[2]], 'k--', linewidth=1, alpha=0.5)

# Highlight the line to the BMU
ax.plot([x[0], bmu_weight[0]], [x[1], bmu_weight[1]], [x[2], bmu_weight[2]],
        'g-', linewidth=3, label=f'Distance to BMU ({bmu_name})')

# Formating the Plot
ax.set_title('Finding the Best Matching Unit (BMU) in 3D space')
ax.set_xlabel('Dimension 1')
ax.set_ylabel('Dimension 2')
ax.set_zlabel('Dimension 3')

handles, labels =ax.get_legend_handles_labels()
unique_labels = dict(zip(labels, handles))
ax.legend(unique_labels.values(), unique_labels.keys())
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D   # noqa: F401  (enables 3d)

from sklearn.preprocessing import StandardScaler

# ---- Load wholesale.csv (with UCI fallback so it's runnable anywhere) ----
CSV_PATH = "wholesale.csv"
UCI_URL  = ("https://archive.ics.uci.edu/ml/machine-learning-databases/"
            "00292/Wholesale%20customers%20data.csv")

import os
if not os.path.exists(CSV_PATH):
    df = pd.read_csv(UCI_URL); df.to_csv(CSV_PATH, index=False)
else:
    df = pd.read_csv(CSV_PATH)
df = df.rename(columns={"Delicatessen": "Delicassen"})

# ---- For a *3-D* BMU demo (matching the shape of your snippet) ----
# Pick three high-variance spend columns so the 3-D plot is meaningful.
FEATURES_3D = ["Grocery", "Detergents_Paper", "Fresh"]

X_raw = df[FEATURES_3D].to_numpy(dtype=np.float32)
X = StandardScaler().fit_transform(np.log1p(X_raw)).astype(np.float32)

print("Wholesale (3 features):", X.shape)
print("Sample rows:\n", np.round(X[:5], 3))
```

```python
# --- 1. Input vector + neuron weights --------------------------------
rng = np.random.default_rng(63)
sample_idx = int(rng.integers(len(X)))
x = X[sample_idx]                     # one customer's standardized spend

# Same structure as your snippet — 4 neurons, 3 dims
neuron_weights = {
    'w1': np.array([0.1, 0.2, 0.7]),
    'w2': np.array([0.4, 0.9, 0.2]),
    'w3': np.array([0.8, 0.1, 0.3]),
    'w4': np.array([0.5, 0.5, 0.5]),
}

print("\n--- Finding the Best Matching Unit (BMU) ---")
print(f"Input vector (x)  [row {sample_idx}]: {np.round(x, 4)}\n")

# --- 2. Euclidean distance to each neuron ----------------------------
distances = {}
for name, w in neuron_weights.items():
    dist = np.linalg.norm(x - w)
    distances[name] = dist
    print(f"Distance to {name} ({w}): {dist:.4f}")

# --- 3. BMU -----------------------------------------------------------
bmu_name   = min(distances, key=distances.get)
bmu_weight = neuron_weights[bmu_name]
min_dist   = distances[bmu_name]

print(f"\nMinimum distance is: {min_dist:.4f}.")
print(f"Therefore, the Best Matching Unit (BMU) is: {bmu_name}")

# --- 4. 3-D visualization (same layout as your file) -----------------
fig = plt.figure(figsize=(10, 8))
ax  = fig.add_subplot(111, projection='3d')

ax.scatter(x[0], x[1], x[2], c='red', marker='*', s=250,
           label=f'Input customer (row {sample_idx})')

for name, w in neuron_weights.items():
    ax.scatter(w[0], w[1], w[2], c='blue', s=60, label=f'Neuron {name}')
    ax.plot([x[0], w[0]], [x[1], w[1]], [x[2], w[2]],
            'k--', linewidth=1, alpha=0.5)

ax.plot([x[0], bmu_weight[0]], [x[1], bmu_weight[1]], [x[2], bmu_weight[2]],
        'g-', linewidth=3, label=f'Distance to BMU ({bmu_name})')

ax.set_title(f'BMU search on Wholesale customer #{sample_idx}\n'
             f'features = {FEATURES_3D}')
ax.set_xlabel(FEATURES_3D[0]); ax.set_ylabel(FEATURES_3D[1]); ax.set_zlabel(FEATURES_3D[2])

handles, labels = ax.get_legend_handles_labels()
uniq = dict(zip(labels, handles))
ax.legend(uniq.values(), uniq.keys())
plt.grid(True); plt.tight_layout(); plt.show()
```

```python
# Stack neuron weights into a (n_neurons, 3) matrix
names = list(neuron_weights)
W     = np.stack([neuron_weights[n] for n in names])   # (4, 3)

# Pairwise Euclidean distance: (n_samples, n_neurons)
d  = np.linalg.norm(X[:, None, :] - W[None, :, :], axis=2)
bmu_idx   = d.argmin(axis=1)          # index into names for each customer
bmu_dist  = d[np.arange(len(X)), bmu_idx]

# Summary table
summary = pd.DataFrame({
    "neuron":    names,
    "customers": [np.sum(bmu_idx == i) for i in range(len(names))],
    "mean_dist": [float(np.mean(bmu_dist[bmu_idx == i])) if np.any(bmu_idx == i)
                  else float("nan") for i in range(len(names))],
})
print("\nBMU assignment counts (hand-set neurons):")
print(summary.to_string(index=False))

# Attach back to the dataframe for inspection
df_view = df.copy()
df_view["bmu"]       = [names[i] for i in bmu_idx]
df_view["bmu_dist"]  = bmu_dist
print("\nFirst 10 customers with their BMU:")
print(df_view[["Channel", "Grocery", "Detergents_Paper", "Fresh",
               "bmu", "bmu_dist"]].head(10).to_string(index=False))
```

```python
# ---------- Data: all 6 spend columns ----------
SPEND = ["Fresh", "Milk", "Grocery", "Frozen", "Detergents_Paper", "Delicassen"]
X6_raw = df[SPEND].to_numpy(dtype=np.float32)
X6     = StandardScaler().fit_transform(np.log1p(X6_raw)).astype(np.float32)

# ---------- Grid of neurons (e.g., 5x5) ----------
grid_h, grid_w = 5, 5
n_neurons = grid_h * grid_w
n_dim     = X6.shape[1]

rng = np.random.default_rng(63)
# Initialize neuron weights from data (much better than random)
neurons = X6[rng.choice(len(X6), size=n_neurons, replace=False)].copy()  # (25, 6)
grid_pos = np.array([(i, j) for i in range(grid_h) for j in range(grid_w)])

# ---------- SOM training ----------
n_epochs   = 2000
lr0, lr_end = 0.5, 0.01
sig0, sig_end = 2.5, 0.3

def grid_distances(i):
    return np.linalg.norm(grid_pos - grid_pos[i], axis=1)

for epoch in range(n_epochs):
    t  = epoch / n_epochs
    lr = lr0  * (1 - t) + lr_end * t
    sg = sig0 * (1 - t) + sig_end * t

    # shuffled order (SOM = stochastic)
    for idx in rng.permutation(len(X6)):
        x = X6[idx]

        # ---- BMU  (exactly your `min(distances, key=...)` logic, vectorized) ----
        d = np.linalg.norm(neurons - x, axis=1)
        bmu = int(d.argmin())

        # ---- neighbourhood update ----
        h = np.exp(-(grid_distances(bmu) ** 2) / (2 * sg * sg))
        neurons += lr * h[:, None] * (x - neurons)

    if epoch % 400 == 0:
        qe = np.mean(np.min(np.linalg.norm(X6[:, None, :] - neurons[None], axis=2),
                            axis=1))
        print(f"epoch {epoch:4d}  lr={lr:.3f}  sigma={sg:.2f}  QE={qe:.4f}")

# ---------- Assign every customer to its BMU ----------
d_all  = np.linalg.norm(X6[:, None, :] - neurons[None], axis=2)   # (440, 25)
bmu6   = d_all.argmin(axis=1)
bmu6_d = d_all[np.arange(len(X6)), bmu6]

print("\nCustomers per winning neuron (of 25):")
counts = np.bincount(bmu6, minlength=n_neurons).reshape(grid_h, grid_w)
print(counts)

# ---------- Side-by-side with Channel labels ----------
for ch in sorted(np.unique(df["Channel"])):
    m = df["Channel"].to_numpy() == ch
    print(f"Channel {ch}: n={m.sum():3d}  "
          f"mean BMU distance={bmu6_d[m].mean():.3f}")

# ---------- Visualize the 5x5 U-matrix and Channel votes ----------
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Majority Channel per neuron
maj = np.zeros((grid_h, grid_w))
for k in range(n_neurons):
    m = bmu6 == k
    maj.flat[k] = np.nan if not m.any() else np.bincount(
        df["Channel"].to_numpy()[m], minlength=3).argmax()

axes[0].imshow(maj, cmap="coolwarm")
axes[0].set_title("Majority Channel per neuron")
axes[0].set_xlabel("grid col"); axes[0].set_ylabel("grid row")
for k in range(n_neurons):
    r, c = divmod(k, grid_w)
    axes[0].text(c, r, str(int(counts[r, c])), ha="center", va="center", color="k")

axes[1].imshow(counts, cmap="Blues")
axes[1].set_title("Customers per neuron (hit map)")
axes[1].set_xlabel("grid col"); axes[1].set_ylabel("grid row")
plt.tight_layout(); plt.savefig("wholesale_som.png", dpi=120); plt.show()
```

```python
That second script is what your snippet grows into: the same min(distances, ...) BMU call,
wrapped in a training loop that pulls the neurons toward real customers. After training, a BMU lookup answers 
"which type of customer is this?" rather than "which of four arbitrary points is closest?"

What you'll observe on Wholesale

Signal	                                            What to expect
___________________________________________________|_________________________________________________________________
Hand-set neurons (Stage 2)	                       | Extremely uneven counts — one or two neurons absorb most of the dataset. 
                                                   | Shows why fixed prototypes fail.
 --------------------------------------------------|---------------------------------------------------------------------
                                                   |
Quantization error during SOM training (Stage 3)   | Falls from ~2.5 to ~0.6 over 2000 epochs.
    --------------------------------------------------------------------------------------------------
                                                   |
Hit map	                                           | Non-uniform — the "retail-heavy" corner of the grid 
                                                   |  (high Grocery/Detergents/Milk, low Fresh) gets many customers; 
                                                   | "fresh-heavy" corner (high Fresh, low Grocery) gets a smaller, distinct set.
     ----------------------------------------------|----------------------------------------------------------------------
                                                   |     
Majority-Channel grid	                           | A clean spatial split — one half of the grid is dominated by Channel 1, the other by Channel 2,
                                                   | even though Channel was never seen by the algorithm. This is the same emergent separation the 
                                                   | autoencoder produced, just with a different unsupervised method.
                -----------------------------------|-----------------------------------------------------------

If the majority-Channel map looks like random noise, the usual cause is skipping log1p + StandardScaler — the SOM will get dragged by the huge Fresh values and forget the other 5 dimensions. Same failure mode as the autoencoder.

TL;DR
Your neuron_weights + np.linalg.norm + min(distances, key=...) is exactly the BMU step of a SOM.

Applied literally (Stage 1) it just answers "which of 4 fixed points is nearest this customer?"

Applied across the dataset (Stage 2) it gives you a partition — but a bad one if the prototypes are hand-set.

Wrapped in the SOM update rule (Stage 3) the neurons learn to sit on the Wholesale customer types, 
and the same BMU call then gives a meaningful customer-type assignment that lines up with Channel even though Channel was held out.
```
