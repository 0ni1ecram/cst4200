QUADRATIC EXAMPLE

```python
# ---- Define Quadrati Parameters
# We'll Create data for the equation: y = b2*x^2 + b1*x + b0
b2 = 2.5  # Formerly a
b1 = -3.0 # Formerly b
b0 = 10.0 # Formerly C


# Generate data
#  Create 100 values from -10 to 10
x_data = np.linspace(-10, 10, 100)

# Calculate the corresponding y-values based on quadratic equations
y_true = b2 * x_data**2 + b1 * x_data + b0
```

```python
# Adding random noise to simulate real world
# This makes the fitting process more interesting
noise = np.random.normal(0, 15, x_data.shape) # Mean=0, StdDev=15
y_data = y_true + noise

# Plot the data
# Create scater plot of the generated data
plt.figure(figsize=(10, 6))
plt.scatter(x_data, y_data, label='Generated Noisy DAta', color='blue', alpha=0.7)

#  Customize and show plot
plt.title('Scatter Plot of Generated Quadratic DAta', fontsize=16)
plt.xlabel('X Values', fontsize=14)
plt.ylabel('Y Values', fontsize=14)
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Plot Noisy DAta and True Equation
#  Create new figure to show both the Noisy data and the true underlying equation
plt.figure(figsize=(10,6))

# Sacatter Plot for the Noisy DAta in Blue
plt.scatter(x_data, y_data, label='Generated Noisy DAta', color='blue', alpha=0.6)

# Line plot for the true equation in red
plt.plot(x_data, y_true, label='True Equation (y_true)', color='red', linewidth=3)

# Customize and show the new plot
plt.title('Scatter Plot of Generated Quadratic DAta', fontsize=16)
plt.xlabel('X Values', fontsize=14)
plt.ylabel('Y Values', fontsize=14)
plt.legend()
plt.grid(True)
plt.show()
```

QUADRATIC AND HIGHER ORDER

```python
#  Define Quadratic parameters
b2 = 2.5
b1 = -3.0
b0 = 10.0

# Generate DAta
# Create 100 x-values


# Demo for 6th-Order Polynomial
# Manually define coefficients for a 6th order
b6_6 = 0.001   # coefficient for x^6
b6_5 = -0.005  # coefficient for x^5
b6_4 = -0.05   # coefficient for x^4
b6_3 = 0.1     # coefficient for x^3
b6_2 = 1.5     # coefficient for x^2
b6_1 = -2.0    # coefficient for x^1
b6_0 = 8.0     # coefficient for x^0 (intercept)

#  Calculate the y-values for the 6th order polynomial
y_poly6 = (b6_6 * x_data**6 +
           b6_5 * x_data**5 +
           b6_4 * x_data**4 +
           b6_3 * x_data**3 +
           b6_2 * x_data**2 +
           b6_1 * x_data +
           b6_0)

# --- 6. Plot the Data and the High-Order Polynomial ---
plt.figure(figsize=(12, 7))

# Plot the original noisy data
plt.scatter(x_data, y_data, label='Original Noisy Data', color='blue', alpha=0.6)

# Plot the 6th-order polynomial curve
plt.plot(x_data, y_poly6, label='6th-Order Polynomial', color='purple', linewidth=2)

# --- 7. Customize and Show the New Plot ---
plt.title('Data vs. Manually Defined 6th-Order Polynomial', fontsize=16)
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)
plt.legend()
plt.grid(True)
plt.ylim(min(y_data) - 20, max(y_data) + 20) # Adjust y-axis
plt.show()
```

POLY_FIT

```python
# Set seed for reproducibility
np.random.seed(42)

# Generate 100 data points
m = 100
x = 6 * np.random.rand(m, 1) - 3
y = 0.5 * x**2 + x + 2 + np.random.randn(m, 1)

# Plot the data
plt.scatter(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Generated Non-linear Data')
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Transform features to include polynomial terms
poly_features = PolynomialFeatures(degree=2, include_bias=False)
x_poly = poly_features.fit_transform(x)
```

```python
# Poly now contains the original features x and x^2
# x[0]-> (x[0], x[0]**2)

# Fit Linear Regression model on the transformed features
lin_reg = LinearRegression()
lin_reg.fit(x_poly, y)
```

```python
#  Generate points for plotting the curve
x_new = np.linspace(-3, 3, 100).reshape(100, 1)
x_new_poly = poly_features.transform(x_new)
y_new = lin_reg.predict(x_new_poly)

# pLOT THE Original DAta and Polynomial fit
plt.scatter(x, y, label='Original Data', color='blue')
plt.plot(x_new, y_new, "-r", linewidth=2, label='Polynomial Fit (deg=2)', color='red')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Polynomial Regression Fit')
plt.legend()
plt.show()
```

RIDGE REGRESSION

```python
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
```

```python
# Generate 100 data points
m = 100
x = 6 * np.random.rand(m, 1) - 3
y = 0.5 * x**2 + x + 2 + np.random.randn(m, 1)

# Plot the DAta
plt.scatter(x,y)
plt.title("Generate nonn-linear Data")
plt.show()
```

```python
#  Application of Ridge Regression
# Create High Degree Polynomial Features (degree 20)

polyFeat = PolynomialFeatures(degree=20, include_bias=False)
x_poly = polyFeat.fit_transform(x)

# Scale the Features
scaler = StandardScaler()
x_poly_scaled = scaler.fit_transform(x_poly)

# Apply Ridge Regression
ridgeReg = Ridge(alpha=1, solver='cholesky', random_state=42)
ridgeReg.fit(x_poly_scaled, y)

#
```
