```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Initialize x, learning rate and number of steps
x = 8
learn_rate = 0.05
steps = 5

# Define Gradient of Function y = 2x**2 + 3x - 10
def gradient(x):
    return 4 * x + 3

# Create a List to store data for the DataFrame
data =[]

# Performing Gradient desent for specified number of steps
x_val = [x]
y_val = [2 * x **2 + 3 * x - 10]

print(x_val, y_val)
```

```python
fst_x = 8
fst_y = 2 * x**2 + 3 * x - 10
print(f"Initial x: {fst_x}")
print(f"Initial y: {fst_y}")

fst_grad = gradient(fst_x)
fst_new_x = fst_x - learn_rate * fst_grad
fst_new_y = 2 * fst_new_x **2 + 3 * fst_new_x - 10

# Print initial Values
print(f"Gradient at initial x: {fst_grad}")
print(f"Updated x after step one: {fst_new_x}")
print(f"Updated y after one step: {fst_new_y}")
```

```python
for step in range(steps):
    y = 2 * x**2 +3 * x - 10
    grad = gradient(x)
    new_x = x - learn_rate * grad
    new_y = 2 * new_x**2 + 3 * new_x - 10
    data.append([step, x,y,grad, new_x, new_y])
    x_val.append(new_x)
    y_val.append(new_y)
    x = new_x #Updating x for every step

# Create DAtaFrame
columns = ["Step", "x", "y", "Gradient of X", "New X Value", "New Y Val"]
df = pd.DataFrame(data, columns=columns)

# Display DAtaFrame
print(df)
```

```python
# Plot Function and Gradient Descent Step

x_range = np.linspace(-10, 10, 500)
y_range = 2 * x_range**2 + 3 * x_range - 10

plt.figure(figsize=(10,6))
plt.plot(x_range, y_range, label="y=2x^2+3x-10", color="blue")
plt.scatter(x_val, y_val, color="red", label="Gradient Descent Step")
plt.axhline(0, color='black', linewidth=0.8, linestyle="--")  # x-axis
plt.axvline(0, color='black', linewidth=0.8, linestyle="--")  #y-axis
plt.title("Gradient Descent Optimization")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
