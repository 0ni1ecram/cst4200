```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return sigmoid(x) * (1 - sigmoid(x))

# Define simple weight matrix (same for all layers of simplicity)
W = np.array([[0.1, 0.2, 0.3],
             [0.1, 0.2, 0.3],
             [0.1, 0.2, 0.3]])

# Input Layer
X = np.array([[1], [1], [1]])

# Frward Pass
Z1 = np.dot(W, X)
A1 = sigmoid(Z1)

Z2 = np.dot(W, A1)
A2 = sigmoid(Z2)

Z3 = np.dot(W, A2)
A3 = sigmoid(Z3)

#  Assume Output Error at Output layer ( small error to simulte gradient Flow)
delta_3 = 0.01 * sigmoid_derivative(Z3)

delta_2 = sigmoid_derivative(Z2) * np.dot(W.T, delta_3)
delta_1 = sigmoid_derivative(Z1) * np.dot(W.T, delta_2)
```

```python
# Print REsults
print("Forward Propagation Results:")
print(f"Z1: {Z1.flatten()}")
print(f"A1: {A1.flatten()}")
print(f"Z2: {Z2.flatten()}")
print(f"A2: {A2.flatten()}")
print(f"Z3: {Z3.flatten()}")
print(f"A3: {A3.flatten()}")
```
