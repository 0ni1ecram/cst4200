```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 1. The Input Sentence ---
sentence = ["The", "animal", "didn't", "cross", "the", "street", "because", "it", "was", "too", "tired"]
num_words = len(sentence)

```

```python
# --- 2. Create a "Score" Matrix (No Calculation) ---
# We will manually create a grid to represent attention strength.
# Let's initialize it with very low scores (zeros).
attention_scores = np.zeros((num_words, num_words))

# --- 3. Manually Define the Conceptual Links ---
# This simulates what a trained model would learn to do.

# Rule A: A word always pays a little attention to itself.
for i in range(num_words):
    attention_scores[i, i] = 0.1 # Small self-attention

# Rule B: The most important link we want to show.
# Find the positions of our key words.
it_idx = sentence.index("it")
animal_idx = sentence.index("animal")
tired_idx = sentence.index("tired")

# Manually set a very high score to show "it" is attending to "animal".
attention_scores[it_idx, animal_idx] = 0.9 # STRONG connection

# Let's also say "it" pays some attention to "tired," as that's its state.
attention_scores[it_idx, tired_idx] = 0.4 # Moderate connection

print("--- Conceptual Model ---")
print(f"Manually setting the attention score from '{sentence[it_idx]}' to '{sentence[animal_idx]}' to be very high.")
print("This represents a learned contextual link, without any complex math.")

```

```python

# --- 4. Visualize the Manually-Defined Attention ---
plt.figure(figsize=(12, 10))
sns.heatmap(
    attention_scores,
    xticklabels=sentence,
    yticklabels=sentence,
    annot=True,        # Show the scores in the cells
    cmap="rocket_r",   # Use a visually clear color map
    fmt=".1f"          # Format numbers to one decimal place
)
plt.title("Conceptual Self-Attention: How 'it' finds 'animal'", fontsize=16)
plt.xlabel("Words Being 'Looked At' (Keys)", fontsize=12)
plt.ylabel("Words 'Doing the Looking' (Queries)", fontsize=12)
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
