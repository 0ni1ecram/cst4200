```python
import gensim
from gensim.models import Word2Vec
from gensim.scripts.glove2word2vec import glove2word2vec
from gensim.models import KeyedVectors
import string
import os

# ---------------------------------------------------------
# INPUT DATA
# ---------------------------------------------------------
# The sentence from your image
text = "The animal didn't cross the street because it was too tired"

# Preprocessing: Lowercase and remove punctuation for better matching
clean_text = text.lower().translate(str.maketrans('', '', string.punctuation))
tokens = clean_text.split()

# Word2Vec expects a list of sentences (which are lists of tokens)
# We replicate the sentence to simulate a "corpus" for this demo
corpus = [tokens]

print(f"Original Sentence: {text}")
print(f"Tokens: {tokens}\n")

```

```python

# ---------------------------------------------------------
# WORD2VEC DEMONSTRATION
# ---------------------------------------------------------
print("--- Word2Vec Demonstration ---")

# Initialize and train the model
# min_count=1: Accepts words that appear even once (usually this is 5+)
# vector_size=10: We create small 10-dimensional vectors for this demo
# window=5: Context window size
model_w2v = Word2Vec(sentences=corpus, vector_size=10, window=5, min_count=1, workers=4)

# Access the vector for a specific word, e.g., 'animal'
vector_animal = model_w2v.wv['animal']

print("Vector for 'animal' (Word2Vec, 10 dimensions):")
print(vector_animal)

# Calculate similarity (Note: On 1 sentence, this will be statistically meaningless but shows the API)
similarity = model_w2v.wv.similarity('animal', 'tired')
print(f"Similarity between 'animal' and 'tired': {similarity:.4f}")
print("-" * 30 + "\n")
```
