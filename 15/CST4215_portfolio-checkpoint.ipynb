```python
#  Creat dataset copies before permanently manipulating data
df_r = df_rvew.copy()
df_p = df_prod.copy()

#  To avoid Correlance Dropping columns with less than 90% NUll values

# Keep columns that have at least 80% non-null values



trsh1 = len(df_r) * 0.9
df_r_cln = df_r.dropna(axis=1, thresh=trsh1)
trsh2 = len(df_p) * 0.9
df_p_cln = df_p.dropna(axis=1, thresh=trsh2)

#  Checking for null values

print(df_p_cln.isna().sum(), "\n" + "_"*50 + "\n", df_r_cln.isna().sum())
print("\n" + "_"*50 + "\n")
print(f"Product Shape | Review Shape")
print(df_p_cln.shape, " ---| ", df_r_cln.shape)
print("\n" + "_"*50 + "\n")
```

```python
# Since trust is an issue, in the line of work.
# Our main focus are with the ones whose verified Purchase = True, since their review will actually mean somthing
df_r_cln['verifiedPurchase'].value_counts()

# Passing its self to drop the non-purchasing reviews (all part of cleaning data)
df_r_cln = df_r_cln[df_r_cln['verifiedPurchase']]
```

```python
import time
# There is a very huege possibilities that we could be dealing with  relational databases, so far similarity is pointing at 'ASIN'
# Finding Foreign key in the Database
# 1. Check if column is Unique

dup_cont = df_r_cln['productASIN'].duplicated().sum()
cont = len(df_r_cln['productASIN'])

print("Is the 'productASIN' column Unique: ")
time.sleep(3)
print(df_r_cln['productASIN'].is_unique)
time.sleep(3)
print("\n" + "_"*50 + "\n")
time.sleep(3)
# 2. Print UNique Value (Visualize what we dealing with)
# print(df_r_cln['productASIN'].unique())
print("What's the number of duplicated values:")
time.sleep(3)
print(f"{dup_cont} / {cont}")
time.sleep(3)
print("\n" + "_"*50 + "\n")
time.sleep(3)

# 3. Summarize: By determining the frequency of recurring duplicates
print(df_r_cln['productASIN'].value_counts())
print("\n" + "_"*50 + "\n")
```
