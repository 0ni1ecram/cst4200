````python
The construction project work chain (or project lifecycle) is the sequential workflow of interdependent phases that takes a built asset from an initial concept to final operation. Each stage relies on specific deliverables from the previous stage to control risk, budget, and scheduling. [1] (https://amcengineer.com/stages-of-a-construction-project/), [2] (https://www.rib-software.com/en/blogs/construction-project-phases), [3] (https://www.autodesk.com/blogs/construction/the-6-stages-of-construction/)

1. Conception and Initiation

The chain begins with defining the project's purpose and proving its viability before committing major capital. [1] (https://www.rib-software.com/en/blogs/construction-project-phases), [2] (https://amcengineer.com/stages-of-a-construction-project/)
• Feasibility Studies. Teams evaluate technical, economic, legal, and scheduling constraints to confirm the project makes financial and practical sense. [1] (https://www.rib-software.com/en/blogs/construction-project-phases)
• Site Selection and Evaluation. Surveyors and engineers inspect the land for zoning rules, soil conditions, and environmental impacts. [1] (https://www.bigrentz.com/blog/phases-of-construction), [2] (https://www.projectmanager.com/blog/construction-process)
• Project Initiation Document (PID). The owner establishes the initial scope, preliminary budget, high-level timeline, and key stakeholder roles. [1] (https://www.rib-software.com/en/blogs/construction-project-phases), [2] (https://www.bigrentz.com/blog/phases-of-construction)

2. Design and Engineering

Once approved, architects and engineers translate the owner's vision into actionable technical plans. [1] (https://www.projectmanager.com/blog/construction-process), [2] (https://amcengineer.com/stages-of-a-construction-project/)
• Conceptual and Schematic Design. Architects establish the building's layout, massing, aesthetics, and spatial relationships. [1] (https://www.autodesk.com/blogs/construction/the-6-stages-of-construction/), [2] (https://www.rib-software.com/en/blogs/construction-project-phases)
• Design Development. Structural, mechanical, electrical, and plumbing (MEP) engineers integrate their systems into the architectural plans. [1] (https://www.rib-software.com/en/blogs/construction-project-phases), [2] (https://www.bigrentz.com/blog/phases-of-construction)
• Construction Documents. The design team produces detailed blueprints and technical specifications used for permitting, bidding, and building. [1] (https://www.rib-software.com/en/blogs/construction-project-phases), [2] (https://www.bigrentz.com/blog/phases-of-construction)

3. Preconstruction and Permitting

This bridge phase prepares the legal, financial, and logistical groundwork before physical work starts. [1] (https://www.autodesk.com/blogs/construction/the-6-stages-of-construction/), [2] (https://www.rib-software.com/en/blogs/construction-project-phases)
• Regulatory Approvals. Teams submit construction documents to local authorities to secure building permits and utility clearances.
• Cost Estimation and BOQ. Quantity surveyors create a Bill of Quantities (BOQ) and refine cost estimates based on the final designs.
• Risk and Safety Planning. Project managers establish site-specific safety protocols, risk mitigation plans, and master schedules. [1] (https://www.autodesk.com/blogs/construction/the-6-stages-of-construction/), [2] (https://www.projectmanager.com/blog/construction-process), [3] (https://www.rib-software.com/en/blogs/construction-project-phases), [4] (https://www.bigrentz.com/blog/phases-of-construction)

4. Bidding and Procurement

The project shifts from planning to commercial execution by securing labor, materials, and contracts. [1] (https://www.autodesk.com/blogs/construction/the-6-stages-of-construction/), [2] (https://www.rib-software.com/en/blogs/construction-project-phases)
• Tendering and Contract Award. The owner selects a delivery method (such as Design-Bid-Build or Design-Build), solicits bids, and awards the contract to a general contractor.
• Subcontractor Onboarding. The general contractor divides the scope and hires specialized trades for concrete, steel, electrical, and finishes.
• Material Sourcing. Procurement teams order long-lead items—such as elevators, HVAC chillers, or custom facade glass—to prevent future site bottlenecks. [1] (https://www.linkedin.com/pulse/construction-project-managementa-step-by-step-guide-first-bit-dubai-qsaaf), [2] (https://www.rib-software.com/en/blogs/construction-project-phases), [3] (https://www.bigrentz.com/blog/phases-of-construction)

5. Physical Construction

This is the execution phase where field teams transform drawings into a physical structure. [1] (https://www.bigrentz.com/blog/phases-of-construction), [2] (https://www.projectmanager.com/blog/construction-process)
• Site Mobilization and Earthwork. Crews set up temporary offices, fencing, and utilities before clearing, excavating, and grading the site.
• Substructure and Superstructure. Workers pour foundations and erect the structural frame, including columns, slabs, load-bearing walls, and roofs.
• Building Envelope and MEP Rough-In. Teams install exterior cladding, windows, and insulation while routing internal HVAC ducts, plumbing pipes, and electrical conduits.
• Interior Finishes. Trades complete drywall, flooring, painting, cabinetry, and final fixture installations.
• Project Controls. Throughout execution, managers monitor daily logs, quality assurance checks, safety compliance, Requests for Information (RFIs), and cost-to-completion. [1] (https://www.rib-software.com/en/blogs/construction-project-phases), [2] (https://www.projectmanager.com/blog/construction-process), [3] (https://www.bigrentz.com/blog/phases-of-construction)

6. Commissioning and Closeout

As physical work finishes, the project undergoes rigorous verification before changing hands. [1] (https://amcengineer.com/stages-of-a-construction-project/), [2] (https://www.bigrentz.com/blog/phases-of-construction)
• Systems Commissioning. Engineers test and balance all mechanical, electrical, fire safety, and plumbing systems to ensure they perform to specification.
• Punch List and Inspections. The owner, architect, and local authorities walk the site to log minor defects (the punch list) for the contractor to fix.
• Handover and Occupancy. Upon passing final inspections, the authority issues a Certificate of Occupancy. The contractor demobilizes and delivers as-built drawings, warranties, and maintenance manuals. [1] (https://www.autodesk.com/blogs/construction/the-6-stages-of-construction/), [2] (https://www.bigrentz.com/blog/phases-of-construction), [3] (https://www.projectmanager.com/blog/construction-process)

7. Operation and Maintenance (Post-Construction)

In the full asset lifecycle, the chain concludes with the long-term use of the facility. [1] (https://amcengineer.com/stages-of-a-construction-project/)
• Defects Liability Period. The contractor remains obligated for a set period (typically 6 to 12 months) to repair any latent defects that emerge after handover.
• Facility Management. The owner or operations team manages ongoing maintenance, energy performance, and eventual renovations. [1] (https://amcengineer.com/stages-of-a-construction-project/)

How Stakeholders Connect in the Chain

Different parties lead different links in the work chain, as outlined in the AMC Engineer Construction Stage Guide: [1] (https://amcengineer.com/stages-of-a-construction-project/)

Stakeholder	Primary Focus in the Work Chain	Active Span
Project Owner / Developer	Defines vision, secures funding, and holds the long-term asset.	Conception through Operation
Architects & Engineers	Translates requirements into safe, code-compliant designs.	Design through Closeout
Project Manager (PM)	Controls scope, schedule, budget, and cross-team communication.	Initiation through Closeout
General Contractor	Manages site execution, safety, subcontractors, and physical delivery.	Bidding through Defects Liability
Suppliers & Subcontractors	Provides specialized trade labor, materials, and heavy equipment.	Procurement through Commissioning

For deeper detail on project controls and phase breakdowns, explore the RIB Software Construction Phases Overview or the Autodesk Digital Builder Stage Guide. [1] (https://www.rib-software.com/en/blogs/construction-project-phases), [2] (https://www.autodesk.com/blogs/construction/the-6-stages-of-construction/)
Would you like to explore a specific angle of this work chain, such as contract delivery models (Design-Build vs. EPC), critical path scheduling, or regional approval gates (such as in the UAE or KSA)?
````

````python
Across the seven construction lifecycle phases, teams can collect over 150 distinct categories of data points—which scale into millions of individual records depending on project size, duration, and digital adoption (such as BIM, IoT sensors, and ERP software).
Here is a breakdown of the key data points captured at each stage:

1. Conception and Initiation (~15–20 Data Types)

• Feasibility and Finance: Projected Return on Investment (ROI), Net Present Value (NPV), Internal Rate of Return (IRR), cap rate, and funding sources.
• Site and Land: Land parcel coordinates, plot area, Floor Area Ratio (FAR), zoning classifications, soil bearing capacity, water table depth, and environmental impact metrics.
• Project Baseline: Rough Order of Magnitude (ROM) budget, target start and completion dates, Gross Floor Area (GFA), and stakeholder contact registries.

2. Design and Engineering (~25–30 Data Types)

• Architectural and Spatial: Room schedules, net-to-gross area ratios, occupancy loads, acoustic ratings, and fire-rating classifications.
• Structural: Concrete strength grades (MPa/PSI), rebar tonnage, steel section weights, wind/seismic load calculations, and foundation depths.
• MEP Systems: Cooling/heating loads (BTUs/tons), electrical kVA demand, lighting lux levels, water flow rates, and drainage slopes.
• BIM and Design Controls: Clash detection counts, BIM Level of Development (LOD) metrics, drawing revision numbers, and design review approval timestamps.

3. Preconstruction and Permitting (~20–25 Data Types)

• Cost and Quantities: Bill of Quantities (BOQ) line items, unit material rates, labor productivity rates, contingency reserves, and value engineering savings.
• Scheduling: Work Breakdown Structure (WBS) codes, task durations, activity dependencies, critical path float days, and milestone target dates.
• Permitting and Risk: Permit application submission/approval dates, regulatory fee amounts, risk probability/impact scores, and safety hazard mitigation logs.

4. Bidding and Procurement (~20–25 Data Types)

• Tendering: Bid package counts, bidder prequalification scores, bid variance percentages, tender clarifications, and contract award values.
• Subcontractors: Trade package scopes, insurance/bond expiration dates, performance guarantees, and payment milestone schedules.
• Supply Chain: Purchase Order (PO) numbers, vendor lead times, factory fabrication progress, shipping ETAs, customs clearance dates, and material unit costs.

5. Physical Construction (~40–50 Data Types)

• Progress and Labor: Daily headcount by trade, man-hours worked, percent complete per WBS activity, Earned Value metrics (CPI, SPI), and equipment utilization hours.
• Quality and Engineering Controls: Request for Information (RFI) turnaround times, submittal approval statuses, concrete cube/cylinder break test results, weld inspection passes, and Non-Conformance Reports (NCRs).
• Health, Safety, and Environment (HSE): Toolbox talk attendance, near-miss reports, Lost Time Injury Frequency Rates (LTIFR), PPE compliance checks, and waste/carbon emissions tracking.
• Financials: Interim payment applications, change order counts and values, committed vs. actual costs, and retention amounts.
• Site Conditions: Daily weather logs (temperature, wind speed, rainfall) and drone/360-photo progress timestamps.

6. Commissioning and Closeout (~15–20 Data Types)

• Testing and Balancing: HVAC airflow CFM readings, water pressure test results, electrical insulation resistance values, and fire alarm response times.
• Defects and Inspections: Punch list/snag item counts, snag resolution turnaround times, municipal inspection pass/fail logs, and Certificate of Occupancy issue dates.
• Handover Assets: Asset tag IDs, equipment serial numbers, warranty start/expiration dates, spare parts inventory counts, and as-built model verification logs.

7. Operation and Maintenance (~15–20 Data Types)

• Defects Liability: Post-handover defect reports, contractor response times, repair completion dates, and retention release approvals.
• Facility Performance: IoT/BMS telemetry (energy kWh, water consumption, indoor air quality, occupancy patterns), preventive maintenance schedules, work order resolution times, and asset lifecycle replacement costs.

Summary Scale by Project Type


Project Scale	Typical Data Categories	Total Individual Records Generated
Small Residential / Fit-Out	~60–80 categories	Thousands of records
Mid-Size Commercial	~100–130 categories	Hundreds of thousands of records
Mega-Infrastructure / Smart Building	150+ categories	Millions to billions (with live IoT/BIM)

Would you like to narrow this down into a data collection schema/table for a specific software platform (like Procore, Autodesk Construction Cloud, or Power BI), or focus on KPIs for one particular phase?
````

````python
Below is a relational database design that links every data point in your seven-phase construction lifecycle. It normalizes the data into entities/tables, gives each data point a home, and shows the foreign-key relationships that connect phases (conception → design → preconstruction → procurement → construction → commissioning → operations). A short set of conventions and a relationship map follow the schema.

---

## Design conventions

- **Surrogate keys**: every table has an integer `*_id` primary key.
- **Natural/business keys** are kept as `UNIQUE` columns where they exist (e.g., `wbs_code`, `po_number`, `asset_tag_id`).
- **Phase anchors**: `project_id` is the root; most tables cascade from it. Design/procurement/construction objects reference the WBS, BOQ, or contract they belong to.
- **Time-series data** (progress, HSE, telemetry, weather) are stored in fact tables keyed by `(project_id, activity/asset, date)`.
- **Derived metrics** (CPI, SPI, LTIFR, NPV) are stored as computed/denormalized columns or views, not as independent entities.
- **Units** are recorded in column names or a `unit` column to avoid ambiguity (MPa vs PSI, CFM, kVA).

---

## 1. Core / project baseline

```sql
CREATE TABLE project (
  project_id            INTEGER PRIMARY KEY,
  project_name          VARCHAR(200) NOT NULL,
  rom_budget            NUMERIC(18,2),          -- ROM budget
  target_start_date     DATE,
  target_completion_date DATE,
  gfa_m2                NUMERIC(12,2),          -- Gross Floor Area
  created_at            TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

CREATE TABLE stakeholder (
  stakeholder_id        INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  name                  VARCHAR(200) NOT NULL,
  role                  VARCHAR(100),
  email                 VARCHAR(200),
  phone                 VARCHAR(50)
);
```

## 2. Feasibility & finance

```sql
CREATE TABLE funding_source (
  funding_source_id     INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  source_name           VARCHAR(200),
  amount                NUMERIC(18,2),
  terms                 TEXT
);

CREATE TABLE financial_metric (
  metric_id             INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  metric_type           VARCHAR(30) CHECK (metric_type IN ('ROI','NPV','IRR','CAP_RATE')),
  metric_value          NUMERIC(18,6),
  as_of_date            DATE,
  UNIQUE (project_id, metric_type, as_of_date)
);
```

## 3. Site & land

```sql
CREATE TABLE site (
  site_id               INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  parcel_coordinates    VARCHAR(200),           -- or GEOGRAPHY(POLYGON)
  plot_area_m2          NUMERIC(14,2),
  far                   NUMERIC(8,3),           -- Floor Area Ratio
  zoning_classification VARCHAR(100),
  soil_bearing_capacity_kpa NUMERIC(10,2),
  water_table_depth_m   NUMERIC(8,2),
  environmental_impact_score NUMERIC(6,2)
);
```

## 4. Design & engineering

```sql
CREATE TABLE design_discipline (
  discipline_id         INTEGER PRIMARY KEY,
  name                  VARCHAR(50)            -- Architectural, Structural, MEP
);

CREATE TABLE room (
  room_id               INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  room_number           VARCHAR(50),
  room_name             VARCHAR(200),
  net_area_m2           NUMERIC(10,2),
  gross_area_m2         NUMERIC(10,2),
  occupancy_load        INTEGER,
  acoustic_rating       VARCHAR(50),
  fire_rating_class     VARCHAR(50)
);

CREATE TABLE structural_element (
  element_id            INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  element_type          VARCHAR(100),
  concrete_strength_mpa NUMERIC(8,2),
  rebar_tonnage_t       NUMERIC(12,3),
  steel_section_weight_kg NUMERIC(12,3),
  wind_load_kpa         NUMERIC(10,3),
  seismic_load_kpa      NUMERIC(10,3),
  foundation_depth_m    NUMERIC(8,2)
);

CREATE TABLE mep_system (
  mep_id                INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  system_type           VARCHAR(50),            -- HVAC, Electrical, Plumbing, Drainage
  cooling_load_tons     NUMERIC(10,2),
  heating_load_btu      NUMERIC(14,2),
  electrical_demand_kva NUMERIC(12,2),
  lighting_lux          NUMERIC(8,2),
  water_flow_rate_lps   NUMERIC(10,2),
  drainage_slope_pct    NUMERIC(6,2)
);

CREATE TABLE design_control (
  design_control_id     INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  clash_detection_count INTEGER,
  bim_lod_level         VARCHAR(20),
  drawing_revision_no   VARCHAR(50),
  review_approval_ts    TIMESTAMP
);
```

## 5. Preconstruction & scheduling

```sql
CREATE TABLE wbs (
  wbs_id                INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  wbs_code              VARCHAR(50) NOT NULL UNIQUE,
  parent_wbs_id         INTEGER REFERENCES wbs(wbs_id),
  description           VARCHAR(300)
);

CREATE TABLE schedule_activity (
  activity_id           INTEGER PRIMARY KEY,
  wbs_id                INTEGER NOT NULL REFERENCES wbs(wbs_id),
  activity_name         VARCHAR(300),
  duration_days         NUMERIC(8,2),
  dependency_activity_id INTEGER REFERENCES schedule_activity(activity_id),
  critical_path_float_days NUMERIC(8,2),
  milestone_target_date DATE
);

CREATE TABLE boq_item (
  boq_item_id           INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  wbs_id                INTEGER REFERENCES wbs(wbs_id),
  description           VARCHAR(300),
  quantity              NUMERIC(16,3),
  unit                  VARCHAR(20),
  unit_material_rate    NUMERIC(14,4),
  labor_productivity_rate NUMERIC(12,4)
);

CREATE TABLE contingency (
  contingency_id        INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  reserve_amount        NUMERIC(18,2),
  value_engineering_savings NUMERIC(18,2)
);

CREATE TABLE permit (
  permit_id             INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  permit_type           VARCHAR(100),
  submitted_date        DATE,
  approved_date         DATE,
  regulatory_fee        NUMERIC(14,2)
);

CREATE TABLE risk_register (
  risk_id               INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  description           TEXT,
  probability_score     NUMERIC(5,2),
  impact_score          NUMERIC(5,2),
  mitigation_log        TEXT
);
```

## 6. Bidding & procurement

```sql
CREATE TABLE bid_package (
  bid_package_id        INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  trade_package_scope   TEXT,
  bidder_prequal_score  NUMERIC(6,2),
  bid_variance_pct      NUMERIC(8,3),
  tender_clarifications TEXT,
  award_value           NUMERIC(18,2)
);

CREATE TABLE contract (
  contract_id           INTEGER PRIMARY KEY,
  bid_package_id        INTEGER REFERENCES bid_package(bid_package_id),
  vendor_id             INTEGER,
  contract_value        NUMERIC(18,2),
  performance_guarantee TEXT,
  insurance_expiry_date DATE,
  bond_expiry_date      DATE
);

CREATE TABLE payment_milestone (
  milestone_id          INTEGER PRIMARY KEY,
  contract_id           INTEGER NOT NULL REFERENCES contract(contract_id),
  description           VARCHAR(300),
  due_date              DATE,
  amount                NUMERIC(18,2),
  status                VARCHAR(30)
);

CREATE TABLE purchase_order (
  po_id                 INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  po_number             VARCHAR(50) NOT NULL UNIQUE,
  vendor_lead_time_days INTEGER,
  unit_material_cost    NUMERIC(14,4),
  order_date            DATE
);

CREATE TABLE supply_shipment (
  shipment_id           INTEGER PRIMARY KEY,
  po_id                 INTEGER NOT NULL REFERENCES purchase_order(po_id),
  fabrication_progress_pct NUMERIC(6,2),
  shipping_eta          DATE,
  customs_clearance_date DATE
);
```

## 7. Physical construction (fact tables)

```sql
CREATE TABLE daily_progress (
  progress_id           INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  activity_id           INTEGER NOT NULL REFERENCES schedule_activity(activity_id),
  report_date           DATE NOT NULL,
  headcount_by_trade    INTEGER,
  man_hours             NUMERIC(10,2),
  percent_complete      NUMERIC(6,2),
  equipment_util_hours  NUMERIC(10,2),
  cpi                   NUMERIC(8,4),           -- Earned Value
  spi                   NUMERIC(8,4),
  UNIQUE (activity_id, report_date)
);

CREATE TABLE rfi (
  rfi_id                INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  submitted_ts          TIMESTAMP,
  response_ts           TIMESTAMP,
  turnaround_hours      NUMERIC(10,2)
);

CREATE TABLE submittal (
  submittal_id          INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  description           VARCHAR(300),
  approval_status       VARCHAR(30),
  approval_ts           TIMESTAMP
);

CREATE TABLE quality_test (
  test_id               INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  test_type             VARCHAR(50),            -- concrete break, weld inspection
  result_value          NUMERIC(14,4),
  pass_fail             BOOLEAN,
  test_date             DATE
);

CREATE TABLE ncr (
  ncr_id                INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  description           TEXT,
  raised_date           DATE,
  closed_date           DATE
);

CREATE TABLE hse_record (
  hse_id                INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  record_date           DATE NOT NULL,
  toolbox_attendance    INTEGER,
  near_miss_count       INTEGER,
  ltifr                 NUMERIC(8,4),
  ppe_compliance_pct    NUMERIC(6,2),
  waste_tons            NUMERIC(12,3),
  carbon_emissions_t    NUMERIC(12,3),
  UNIQUE (project_id, record_date)
);

CREATE TABLE payment_application (
  app_id                INTEGER PRIMARY KEY,
  contract_id           INTEGER NOT NULL REFERENCES contract(contract_id),
  period_end_date       DATE,
  amount_applied        NUMERIC(18,2),
  retention_amount      NUMERIC(18,2),
  approved_amount       NUMERIC(18,2)
);

CREATE TABLE change_order (
  change_order_id       INTEGER PRIMARY KEY,
  contract_id           INTEGER NOT NULL REFERENCES contract(contract_id),
  description           TEXT,
  value                 NUMERIC(18,2),
  committed_cost        NUMERIC(18,2),
  actual_cost           NUMERIC(18,2)
);

CREATE TABLE site_weather_log (
  weather_id            INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  log_date              DATE NOT NULL,
  temperature_c         NUMERIC(6,2),
  wind_speed_kph        NUMERIC(6,2),
  rainfall_mm           NUMERIC(8,2),
  drone_photo_ts        TIMESTAMP,              -- 360/drone progress capture
  UNIQUE (project_id, log_date)
);
```

## 8. Commissioning & closeout

```sql
CREATE TABLE commissioning_test (
  test_id               INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  system_type           VARCHAR(50),
  hvac_cfm              NUMERIC(10,2),
  water_pressure_kpa    NUMERIC(10,2),
  insulation_resistance_mohm NUMERIC(10,3),
  fire_alarm_response_s NUMERIC(8,3),
  test_date             DATE
);

CREATE TABLE punch_item (
  punch_item_id         INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  description           TEXT,
  raised_date           DATE,
  resolved_date         DATE,
  resolution_turnaround_days NUMERIC(8,2)
);

CREATE TABLE municipal_inspection (
  inspection_id         INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  inspection_type       VARCHAR(100),
  inspection_date       DATE,
  pass_fail             BOOLEAN,
  certificate_of_occupancy_date DATE
);

CREATE TABLE asset (
  asset_id              INTEGER PRIMARY KEY,
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  asset_tag_id          VARCHAR(50) NOT NULL UNIQUE,
  equipment_type        VARCHAR(100),
  serial_number         VARCHAR(100),
  warranty_start_date   DATE,
  warranty_expiry_date  DATE,
  spare_parts_count     INTEGER,
  as_built_verified     BOOLEAN,
  lifecycle_replacement_cost NUMERIC(18,2)
);
```

## 9. Operation & maintenance

```sql
CREATE TABLE defect (
  defect_id             INTEGER PRIMARY KEY,
  asset_id              INTEGER REFERENCES asset(asset_id),
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  reported_date         DATE,
  contractor_response_ts TIMESTAMP,
  repair_completion_date DATE,
  retention_release_approved BOOLEAN
);

CREATE TABLE maintenance_schedule (
  schedule_id           INTEGER PRIMARY KEY,
  asset_id              INTEGER NOT NULL REFERENCES asset(asset_id),
  task_description      VARCHAR(300),
  frequency_days        INTEGER,
  next_due_date         DATE
);

CREATE TABLE work_order (
  work_order_id         INTEGER PRIMARY KEY,
  asset_id              INTEGER NOT NULL REFERENCES asset(asset_id),
  opened_ts             TIMESTAMP,
  closed_ts             TIMESTAMP,
  resolution_hours      NUMERIC(10,2)
);

CREATE TABLE facility_telemetry (
  telemetry_id          INTEGER PRIMARY KEY,
  asset_id              INTEGER REFERENCES asset(asset_id),
  project_id            INTEGER NOT NULL REFERENCES project(project_id),
  reading_ts            TIMESTAMP NOT NULL,
  energy_kwh            NUMERIC(14,4),
  water_consumption_m3  NUMERIC(12,4),
  indoor_air_quality    NUMERIC(10,3),
  occupancy_count       INTEGER,
  UNIQUE (asset_id, reading_ts)
);
```

---

## Relationship map (text ERD)

```
project 1─* stakeholder
project 1─* funding_source
project 1─1 financial_metric (per type/date)
project 1─1 site
project 1─* room / structural_element / mep_system / design_control
project 1─* wbs 1─* schedule_activity
project 1─* boq_item ── wbs
project 1─1 contingency
project 1─* permit
project 1─* risk_register
project 1─* bid_package 1─* contract 1─* payment_milestone
contract 1─* payment_application
contract 1─* change_order
project 1─* purchase_order 1─* supply_shipment
schedule_activity 1─* daily_progress
project 1─* rfi / submittal / quality_test / ncr / hse_record / site_weather_log
project 1─* commissioning_test / punch_item / municipal_inspection
project 1─* asset 1─* defect / maintenance_schedule / work_order / facility_telemetry
```

---

## Coverage check (data points → table)

| Phase | Data points mapped |
|---|---|
| 1 Conception | ROI/NPV/IRR/cap rate → `financial_metric`; funding → `funding_source`; parcel/FAR/zoning/soil/water table/EIA → `site`; ROM/GFA/dates → `project`; stakeholders → `stakeholder` |
| 2 Design | rooms, N/G ratio, occupancy, acoustic, fire → `room`; concrete/rebar/steel/wind/seismic/foundation → `structural_element`; loads/kVA/lux/flow/slope → `mep_system`; clashes/LOD/revisions/approvals → `design_control` |
| 3 Preconstruction | BOQ/rates/productivity → `boq_item`; contingency/VE → `contingency`; WBS/durations/dependencies/float/milestones → `wbs`, `schedule_activity`; permits/fees → `permit`; risk/safety → `risk_register` |
| 4 Procurement | bid packages/prequal/variance/clarifications/award → `bid_package`; insurance/bond/guarantee → `contract`; payment milestones → `payment_milestone`; PO/lead time/unit cost → `purchase_order`; fabrication/shipping/customs → `supply_shipment` |
| 5 Construction | headcount/man-hours/%complete/EVM/equipment → `daily_progress`; RFI/submittal/tests/NCR → `rfi`, `submittal`, `quality_test`, `ncr`; HSE/LTIFR/waste/carbon → `hse_record`; payments/change orders/retention → `payment_application`, `change_order`; weather/drone → `site_weather_log` |
| 6 Commissioning | CFM/pressure/insulation/fire alarm → `commissioning_test`; punch/snag/municipal/CO → `punch_item`, `municipal_inspection`; asset tags/serials/warranty/spares/as-built → `asset` |
| 7 O&M | defects/response/repair/retention → `defect`; telemetry/PM/work orders/lifecycle cost → `facility_telemetry`, `maintenance_schedule`, `work_order`, `asset.lifecycle_replacement_cost` |

This schema captures all ~150 data types listed, keeps time-series facts separate from master data, and links phases through `project_id`, `wbs_id`, `contract_id`, and `asset_id` so you can trace any operational reading back to its design, procurement, and construction origin.
````

````python
# Interactive ERD Explorer — Construction Lifecycle Database

Below is a complete, self-contained HTML file that renders the schema from your document as an interactive ERD explorer. Open it in any browser (no build step, no external dependencies). It supports:

- **Pan & zoom** the canvas (drag to pan, wheel to zoom)
- **Click any table** to highlight its foreign-key links (both directions) and dim everything else
- **Hover a relationship** to see the FK→PK detail
- **Search / filter** tables by name or phase
- **Toggle phases** (Core, P1–2 Design, P3–4 Procure, P5 Construction, P6–7 O&M)
- **Side panel** showing full column list + FK targets for the selected table
- **Layout switcher** (phase-column layout vs. force-ish radial grouping)

```html
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="utf-8" />
<title>Construction Lifecycle — Interactive ERD Explorer</title>
<style>
  :root{
    --bg:#0f1419; --panel:#161d26; --panel2:#1e2833; --line:#2b3a48;
    --txt:#dce6f0; --muted:#7d90a3; --accent:#4fd1c5; --accent2:#f6ad55;
    --fk:#4fd1c5; --pk:#f6ad55; --sel:#63b3ed; --dim:#33414f;
    --core:#f6ad55; --d12:#63b3ed; --p34:#9f7aea; --p5:#48bb78; --p67:#ed64a6;
  }
  *{box-sizing:border-box}
  html,body{margin:0;height:100%;font-family:ui-monospace,Menlo,Consolas,monospace;background:var(--bg);color:var(--txt)}
  #app{display:grid;grid-template-columns:1fr 340px;height:100%}
  #stage{position:relative;overflow:hidden;background:radial-gradient(circle at 30% 20%,#182430,#0f1419 70%)}
  svg{width:100%;height:100%;cursor:grab;display:block}
  svg.dragging{cursor:grabbing}

  /* toolbar */
  #toolbar{position:absolute;top:12px;left:12px;right:12px;display:flex;gap:8px;flex-wrap:wrap;
    align-items:center;background:rgba(22,29,38,.92);border:1px solid var(--line);border-radius:10px;
    padding:8px 10px;backdrop-filter:blur(6px);z-index:5}
  #toolbar input[type=text]{background:var(--panel2);border:1px solid var(--line);color:var(--txt);
    padding:6px 10px;border-radius:6px;font-size:12px;width:190px;outline:none}
  #toolbar input:focus{border-color:var(--accent)}
  .chip{font-size:11px;padding:4px 9px;border-radius:20px;border:1px solid var(--line);
    background:var(--panel2);color:var(--muted);cursor:pointer;user-select:none;transition:.15s}
  .chip.on{color:#0f1419;font-weight:700}
  .chip[data-phase=core].on{background:var(--core);border-color:var(--core)}
  .chip[data-phase=d12].on{background:var(--d12);border-color:var(--d12)}
  .chip[data-phase=p34].on{background:var(--p34);border-color:var(--p34)}
  .chip[data-phase=p5].on{background:var(--p5);border-color:var(--p5)}
  .chip[data-phase=p67].on{background:var(--p67);border-color:var(--p67)}
  .chip[data-act].on{background:var(--accent);border-color:var(--accent)}
  .spacer{flex:1}
  #legend{font-size:10px;color:var(--muted);display:flex;gap:12px;align-items:center}
  .lg{display:inline-flex;align-items:center;gap:4px}
  .dot{width:8px;height:8px;border-radius:2px;display:inline-block}

  /* side panel */
  #panel{background:var(--panel);border-left:1px solid var(--line);overflow-y:auto;padding:16px}
  #panel h2{font-size:13px;margin:0 0 4px;color:var(--accent);letter-spacing:.5px;text-transform:uppercase}
  #panel h3{font-size:11px;margin:16px 0 6px;color:var(--muted);text-transform:uppercase;letter-spacing:.5px}
  .tbl-name{font-size:18px;font-weight:700;margin-bottom:2px;word-break:break-word}
  .badge{display:inline-block;font-size:10px;padding:2px 7px;border-radius:10px;margin-top:6px;font-weight:700;color:#0f1419}
  .cols{list-style:none;padding:0;margin:0;font-size:12px}
  .cols li{padding:5px 8px;border-radius:5px;display:flex;justify-content:space-between;gap:8px;
    border:1px solid transparent;margin-bottom:2px}
  .cols li:hover{background:var(--panel2)}
  .cols li.pk{border-color:rgba(246,173,85,.4);background:rgba(246,173,85,.08)}
  .cols li.fk{border-color:rgba(79,209,197,.35);background:rgba(79,209,197,.07)}
  .col-name{color:var(--txt);word-break:break-all}
  .col-type{color:var(--muted);font-size:10px;white-space:nowrap}
  .tag{font-size:9px;font-weight:700;padding:1px 5px;border-radius:4px;margin-left:4px}
  .tag.pk{background:var(--pk);color:#0f1419}
  .tag.fk{background:var(--fk);color:#0f1419}
  .rel{margin:6px 0;padding:7px 9px;background:var(--panel2);border-radius:6px;font-size:11px;
    border-left:3px solid var(--accent);cursor:pointer}
  .rel:hover{background:#26333f}
  .rel .arrow{color:var(--accent);font-weight:700}
  .rel .lbl{color:var(--muted);font-size:10px}
  .empty{color:var(--muted);font-size:12px;font-style:italic;margin-top:40px;text-align:center;line-height:1.6}

  /* svg styles */
  .link{fill:none;stroke:var(--line);stroke-width:1.2;transition:.15s}
  .link.active{stroke:var(--accent);stroke-width:2.4;filter:drop-shadow(0 0 4px rgba(79,209,197,.6))}
  .link.dim{opacity:.06}
  .link-hit{fill:none;stroke:transparent;stroke-width:12;cursor:pointer}
  .node{cursor:pointer;transition:.15s}
  .node rect{fill:var(--panel2);stroke:var(--line);stroke-width:1.2;rx:6;transition:.15s}
  .node:hover rect{stroke:var(--accent)}
  .node.active rect{stroke:var(--sel);stroke-width:2.5;filter:drop-shadow(0 0 10px rgba(99,179,237,.5))}
  .node.dim{opacity:.18}
  .node text{fill:var(--txt);font-size:10px;pointer-events:none;font-family:inherit}
  .node .tname{font-size:11px;font-weight:700}
  .node .tphase{font-size:8px;fill:var(--muted);text-transform:uppercase;letter-spacing:.5px}
  .node .bar{width:4px}
  .tooltip{position:absolute;pointer-events:none;background:#0b1016;border:1px solid var(--accent);
    border-radius:6px;padding:7px 10px;font-size:11px;max-width:320px;opacity:0;transition:.12s;
    box-shadow:0 6px 20px rgba(0,0,0,.5);z-index:10;line-height:1.5}
  .tooltip.show{opacity:1}
  .tooltip .tt-t{color:var(--accent);font-weight:700;margin-bottom:3px}
  .tooltip code{color:var(--accent2)}
  #hint{position:absolute;bottom:12px;left:12px;font-size:10px;color:var(--muted);
    background:rgba(22,29,38,.85);padding:6px 10px;border-radius:6px;border:1px solid var(--line);z-index:5}
</style>
</head>
<body>
<div id="app">
  <div id="stage">
    <div id="toolbar">
      <input id="search" type="text" placeholder="filter tables… (e.g. asset, wbs)" />
      <span class="chip on" data-phase="core">Core</span>
      <span class="chip on" data-phase="d12">P1–2 Design</span>
      <span class="chip on" data-phase="p34">P3–4 Procure</span>
      <span class="chip on" data-phase="p5">P5 Build</span>
      <span class="chip on" data-phase="p67">P6–7 O&amp;M</span>
      <span class="spacer"></span>
      <span class="chip" data-act="layout">layout: phase-columns</span>
      <span class="chip" data-act="reset">reset view</span>
      <div id="legend">
        <span class="lg"><span class="dot" style="background:var(--pk)"></span>PK</span>
        <span class="lg"><span class="dot" style="background:var(--fk)"></span>FK</span>
      </div>
    </div>
    <svg id="svg"><g id="viewport"></g></svg>
    <div id="tooltip" class="tooltip"></div>
    <div id="hint">drag to pan · scroll to zoom · click a table to trace its FK links</div>
  </div>
  <div id="panel">
    <h2>ERD Inspector</h2>
    <div id="panel-body">
      <div class="empty">Click any table in the diagram<br/>to inspect its columns<br/>and foreign-key links.</div>
    </div>
  </div>
</div>

<script>
/* ============================================================
   SCHEMA DEFINITION
   phase: core | d12 | p34 | p5 | p67
   cols: [name, type, pk?, fkTarget?]
   ============================================================ */
const SCHEMA = [
  /* ---------- CORE SPINE ---------- */
  {name:"project", phase:"core", cols:[
    ["project_id","INT",1],["name","VARCHAR"],["rom_budget","NUMERIC"],
    ["approved_budget","NUMERIC"],["gfa","NUMERIC"],["target_start_date","DATE"],
    ["target_end_date","DATE"],["current_phase","ENUM"],
    ["certificate_of_occupancy_date","DATE"]
  ]},
  {name:"site", phase:"core", cols:[
    ["site_id","INT",1],["project_id","INT",0,"project"],["parcel_id","VARCHAR"],
    ["coordinates","GEOGRAPHY"],["far","NUMERIC"],["zoning_code","VARCHAR"],
    ["soil_bearing_capacity","NUMERIC"],["water_table_depth","NUMERIC"],["eia_status","ENUM"]
  ]},
  {name:"wbs", phase:"core", cols:[
    ["wbs_id","INT",1],["project_id","INT",0,"project"],["parent_wbs_id","INT",0,"wbs"],
    ["wbs_code","VARCHAR"],["description","TEXT"]
  ]},
  {name:"spatial_zone_room", phase:"core", cols:[
    ["room_id","INT",1],["project_id","INT",0,"project"],["level_floor","INT"],
    ["room_number","VARCHAR"],["net_area","NUMERIC"],["gross_area","NUMERIC"],
    ["occupancy_load","INT"],["acoustic_rating","VARCHAR"],["fire_rating","VARCHAR"]
  ]},
  {name:"stakeholder", phase:"core", cols:[
    ["stakeholder_id","INT",1],["organization_name","VARCHAR"],["role","ENUM"],
    ["contact_info","TEXT"]
  ]},

  /* ---------- PHASE 1 & 2 ---------- */
  {name:"financial_metric", phase:"d12", cols:[
    ["metric_id","INT",1],["project_id","INT",0,"project"],["version","ENUM"],
    ["roi","NUMERIC"],["npv","NUMERIC"],["irr","NUMERIC"],["cap_rate","NUMERIC"],["calculated_at","TIMESTAMP"]
  ]},
  {name:"funding_source", phase:"d12", cols:[
    ["funding_id","INT",1],["project_id","INT",0,"project"],["stakeholder_id","INT",0,"stakeholder"],
    ["source_type","ENUM"],["committed_amount","NUMERIC"]
  ]},
  {name:"bim_element", phase:"d12", cols:[
    ["element_id","INT",1],["project_id","INT",0,"project"],["room_id","INT",0,"spatial_zone_room"],
    ["wbs_id","INT",0,"wbs"],["ifc_guid","UUID"],["discipline","ENUM"],["lod_level","VARCHAR"]
  ]},
  {name:"structural_element", phase:"d12", cols:[
    ["element_id","INT",1,"bim_element"],["concrete_grade","VARCHAR"],["rebar_tonnage","NUMERIC"],
    ["steel_weight","NUMERIC"],["wind_load","NUMERIC"],["seismic_zone","VARCHAR"],["foundation_depth","NUMERIC"]
  ]},
  {name:"mep_system", phase:"d12", cols:[
    ["element_id","INT",1,"bim_element"],["system_type","ENUM"],["cooling_heating_load","NUMERIC"],
    ["kva_demand","NUMERIC"],["lux_level","NUMERIC"],["flow_rate","NUMERIC"],["slope","NUMERIC"]
  ]},
  {name:"design_control", phase:"d12", cols:[
    ["control_id","INT",1],["project_id","INT",0,"project"],["element_id","INT",0,"bim_element"],
    ["revision_number","VARCHAR"],["clash_count","INT"],["status","ENUM"],["approved_by","INT",0,"stakeholder"]
  ]},

  /* ---------- PHASE 3 & 4 ---------- */
  {name:"boq_item", phase:"p34", cols:[
    ["boq_id","INT",1],["project_id","INT",0,"project"],["wbs_id","INT",0,"wbs"],
    ["element_id","INT",0,"bim_element"],["unit_of_measure","VARCHAR"],["estimated_qty","NUMERIC"],
    ["unit_material_rate","NUMERIC"],["unit_labor_rate","NUMERIC"],["productivity_norm","NUMERIC"]
  ]},
  {name:"contingency", phase:"p34", cols:[
    ["contingency_id","INT",1],["project_id","INT",0,"project"],["wbs_id","INT",0,"wbs"],
    ["type","ENUM"],["allocated_amount","NUMERIC"],["remaining_amount","NUMERIC"]
  ]},
  {name:"schedule_activity", phase:"p34", cols:[
    ["activity_id","INT",1],["project_id","INT",0,"project"],["wbs_id","INT",0,"wbs"],
    ["room_id","INT",0,"spatial_zone_room"],["planned_start","DATE"],["planned_end","DATE"],
    ["actual_start","DATE"],["actual_end","DATE"],["duration_days","NUMERIC"],
    ["total_float_days","NUMERIC"],["is_critical_path","BOOL"]
  ]},
  {name:"activity_dependency", phase:"p34", cols:[
    ["dependency_id","INT",1],["predecessor_id","INT",0,"schedule_activity"],
    ["successor_id","INT",0,"schedule_activity"],["dependency_type","ENUM"],["lag_days","INT"]
  ]},
  {name:"permit", phase:"p34", cols:[
    ["permit_id","INT",1],["project_id","INT",0,"project"],["wbs_id","INT",0,"wbs"],
    ["status","ENUM"],["fee_amount","NUMERIC"],["expiry_date","DATE"]
  ]},
  {name:"risk_register", phase:"p34", cols:[
    ["risk_id","INT",1],["project_id","INT",0,"project"],["wbs_id","INT",0,"wbs"],
    ["probability_score","NUMERIC"],["impact_score","NUMERIC"],["mitigation_log","TEXT"]
  ]},
  {name:"bid_package", phase:"p34", cols:[
    ["bid_package_id","INT",1],["project_id","INT",0,"project"],["trade_scope","TEXT"],
    ["prequal_score","NUMERIC"],["bid_variance_pct","NUMERIC"]
  ]},
  {name:"contract", phase:"p34", cols:[
    ["contract_id","INT",1],["project_id","INT",0,"project"],["bid_package_id","INT",0,"bid_package"],
    ["stakeholder_id","INT",0,"stakeholder"],["award_value","NUMERIC"],["retention_pct","NUMERIC"],
    ["insurance_expiry","DATE"],["bond_number","VARCHAR"]
  ]},
  {name:"purchase_order", phase:"p34", cols:[
    ["po_id","INT",1],["contract_id","INT",0,"contract"],["boq_id","INT",0,"boq_item"],
    ["unit_cost","NUMERIC"],["lead_time_days","INT"]
  ]},
  {name:"supply_shipment", phase:"p34", cols:[
    ["shipment_id","INT",1],["po_id","INT",0,"purchase_order"],["fabrication_pct","NUMERIC"],
    ["etd","DATE"],["eta","DATE"],["customs_cleared_at","TIMESTAMP"]
  ]},

  /* ---------- PHASE 5 ---------- */
  {name:"daily_progress", phase:"p5", cols:[
    ["progress_id","INT",1],["activity_id","INT",0,"schedule_activity"],["boq_id","INT",0,"boq_item"],
    ["log_date","DATE"],["headcount","INT"],["man_hours","NUMERIC"],["equipment_hours","NUMERIC"],
    ["percent_complete","NUMERIC"],["earned_value_ev","NUMERIC"],["actual_cost_ac","NUMERIC"]
  ]},
  {name:"rfi", phase:"p5", cols:[
    ["rfi_id","INT",1],["project_id","INT",0,"project"],["element_id","INT",0,"bim_element"],
    ["activity_id","INT",0,"schedule_activity"],["stakeholder_id","INT",0,"stakeholder"],
    ["submitted_at","TIMESTAMP"],["due_date","DATE"],["closed_at","TIMESTAMP"],["cost_schedule_impact_flag","BOOL"]
  ]},
  {name:"submittal", phase:"p5", cols:[
    ["submittal_id","INT",1],["project_id","INT",0,"project"],["element_id","INT",0,"bim_element"],
    ["activity_id","INT",0,"schedule_activity"],["stakeholder_id","INT",0,"stakeholder"],
    ["submitted_at","TIMESTAMP"],["due_date","DATE"],["closed_at","TIMESTAMP"],["cost_schedule_impact_flag","BOOL"]
  ]},
  {name:"quality_test", phase:"p5", cols:[
    ["test_id","INT",1],["project_id","INT",0,"project"],["wbs_id","INT",0,"wbs"],
    ["room_id","INT",0,"spatial_zone_room"],["element_id","INT",0,"bim_element"],
    ["test_type","ENUM"],["measured_value","NUMERIC"],["pass_fail","BOOL"]
  ]},
  {name:"ncr", phase:"p5", cols:[
    ["ncr_id","INT",1],["project_id","INT",0,"project"],["wbs_id","INT",0,"wbs"],
    ["room_id","INT",0,"spatial_zone_room"],["element_id","INT",0,"bim_element"],
    ["description","TEXT"],["ncr_status","ENUM"],["raised_date","DATE"],["closed_date","DATE"]
  ]},
  {name:"hse_record", phase:"p5", cols:[
    ["hse_id","INT",1],["project_id","INT",0,"project"],["log_date","DATE"],
    ["ltifr","NUMERIC"],["waste_tonnage","NUMERIC"],["carbon_kg_co2e","NUMERIC"],
    ["ppe_compliance_pct","NUMERIC"],["toolbox_attendance","INT"]
  ]},
  {name:"site_weather_log", phase:"p5", cols:[
    ["weather_id","INT",1],["project_id","INT",0,"project"],["log_date","DATE"],
    ["temp_c","NUMERIC"],["wind_kmh","NUMERIC"],["rainfall_mm","NUMERIC"],["drone_capture_url","TEXT"]
  ]},
  {name:"payment_application", phase:"p5", cols:[
    ["app_id","INT",1],["contract_id","INT",0,"contract"],["wbs_id","INT",0,"wbs"],
    ["claimed_amount","NUMERIC"],["certified_amount","NUMERIC"],["retention_held","NUMERIC"]
  ]},
  {name:"change_order", phase:"p5", cols:[
    ["change_order_id","INT",1],["contract_id","INT",0,"contract"],["wbs_id","INT",0,"wbs"],
    ["approved_variation_value","NUMERIC"],["description","TEXT"]
  ]},

  /* ---------- PHASE 6 & 7 ---------- */
  {name:"asset", phase:"p67", cols:[
    ["asset_id","INT",1],["element_id","INT",0,"bim_element"],["room_id","INT",0,"spatial_zone_room"],
    ["po_id","INT",0,"purchase_order"],["asset_tag_qr","VARCHAR"],["serial_number","VARCHAR"],
    ["manufacturer","VARCHAR"],["warranty_start","DATE"],["warranty_end","DATE"],
    ["lifecycle_replacement_cost","NUMERIC"]
  ]},
  {name:"commissioning_test", phase:"p67", cols:[
    ["comm_id","INT",1],["asset_id","INT",0,"asset"],["test_parameter","VARCHAR"],
    ["design_target","NUMERIC"],["measured_reading","NUMERIC"],["passed_boolean","BOOL"]
  ]},
  {name:"punch_item", phase:"p67", cols:[
    ["punch_id","INT",1],["room_id","INT",0,"spatial_zone_room"],["asset_id","INT",0,"asset"],
    ["description","TEXT"],["raised_date","DATE"],["resolved_date","DATE"]
  ]},
  {name:"municipal_inspection", phase:"p67", cols:[
    ["inspection_id","INT",1],["project_id","INT",0,"project"],["room_id","INT",0,"spatial_zone_room"],
    ["asset_id","INT",0,"asset"],["inspection_date","DATE"],["pass_fail","BOOL"]
  ]},
  {name:"defect", phase:"p67", cols:[
    ["defect_id","INT",1],["asset_id","INT",0,"asset"],["contract_id","INT",0,"contract"],
    ["reported_at","TIMESTAMP"],["sla_deadline","TIMESTAMP"],["resolved_at","TIMESTAMP"],
    ["retention_release_approved","BOOL"]
  ]},
  {name:"maintenance_schedule", phase:"p67", cols:[
    ["schedule_id","INT",1],["asset_id","INT",0,"asset"],["task_description","TEXT"],
    ["frequency_days","INT"],["next_due_date","DATE"]
  ]},
  {name:"work_order", phase:"p67", cols:[
    ["work_order_id","INT",1],["asset_id","INT",0,"asset"],["opened_at","TIMESTAMP"],
    ["closed_at","TIMESTAMP"],["downtime_hours","NUMERIC"],["repair_cost","NUMERIC"]
  ]},
  {name:"facility_telemetry", phase:"p67", cols:[
    ["telemetry_id","BIGINT",1],["asset_id","INT",0,"asset"],["timestamp","TIMESTAMPTZ"],
    ["metric_name","VARCHAR"],["metric_value","NUMERIC"]
  ]}
];

/* derived helpers */
const PHASE_COLOR = {core:"#f6ad55", d12:"#63b3ed", p34:"#9f7aea", p5:"#48bb78", p67:"#ed64a6"};
const PHASE_LABEL = {core:"Core", d12:"P1–2", p34:"P3–4", p5:"P5", p67:"P6–7"};

const NODE_W = 168, NODE_H_BASE = 26, ROW_H = 13, PAD = 8;

/* ============================================================
   LAYOUT — phase columns, with intra-column vertical stacking
   ============================================================ */
function layoutPhaseColumns(){
  const groups = {core:[], d12:[], p34:[], p5:[], p67:[]};
  SCHEMA.forEach(t => groups[t.phase].push(t));

  const order = ["core","d12","p34","p5","p67"];
  const colGap = 230, rowGap = 14, startX = 60, startY = 60;
  const pos = {};

  order.forEach((ph, ci) => {
    let y = startY;
    groups[ph].forEach(t => {
      const h = NODE_H_BASE + (t.cols.length * ROW_H) + PAD;
      pos[t.name] = {x: startX + ci*colGap, y, h, w: NODE_W, phase: ph};
      y += h + rowGap;
    });
  });
  return pos;
}

/* alternate: radial by phase wedge — keeps tables of same phase clustered in a ring sector */
function layoutRadial(){
  const cx = 620, cy = 620;
  const order = ["core","d12","p34","p5","p67"];
  const groups = {core:[], d12:[], p34:[], p5:[], p67:[]};
  SCHEMA.forEach(t => groups[t.phase].push(t));
  const pos = {};
  const radii = {core:0, d12:340, p34:600, p5:340, p67:600};
  const angles = {core:0, d12: 200, p34: 200, p5: 20, p67: 20}; // degrees
  order.forEach(ph => {
    const list = groups[ph];
    const n = list.length;
    list.forEach((t,i) => {
      let x,y;
      const h = NODE_H_BASE + (t.cols.length * ROW_H) + PAD;
      if(ph==="core"){
        // core in center vertical stack
        x = cx - NODE_W/2;
        y = cy - 300 + i*150;
      } else {
        const spread = 150; // degrees of arc
        const a = (angles[ph] - spread/2 + (spread/(Math.max(n-1,1)))*i) * Math.PI/180;
        x = cx + Math.cos(a)*radii[ph] - NODE_W/2;
        y = cy + Math.sin(a)*radii[ph] - h/2;
      }
      pos[t.name] = {x, y, h, w: NODE_W, phase: ph};
    });
  });
  return pos;
}

/* ============================================================
   EDGE BUILDER — one edge per FK column
   ============================================================ */
function buildEdges(){
  const edges = [];
  SCHEMA.forEach(t => {
    t.cols.forEach(c => {
      if(c[3]){ // fkTarget
        edges.push({from:t.name, to:c[3], col:c[0], targetPk:pkOf(c[3])});
      }
    });
  });
  return edges;
}
function pkOf(tableName){
  const t = SCHEMA.find(x=>x.name===tableName);
  if(!t) return "?";
  const pk = t.cols.find(c=>c[2]===1);
  return pk ? pk[0] : "?";
}

/* ============================================================
   SVG RENDERING
   ============================================================ */
const svg = document.getElementById("svg");
const viewport = document.getElementById("viewport");
const tooltip = document.getElementById("tooltip");
const panelBody = document.getElementById("panel-body");

let LAYOUT = layoutPhaseColumns();
let EDGES = buildEdges();
let selected = null;
let visiblePhases = new Set(["core","d12","p34","p5","p67"]);
let filterText = "";

// pan/zoom state
let tx=0, ty=0, scale=1;
let dragging=false, dragStart=null;

function applyTransform(){
  viewport.setAttribute("transform", `translate(${tx},${ty}) scale(${scale})`);
}

/* ---------- edge path (orthogonal-ish curved) ---------- */
function edgePath(a, b){
  // exit from right side of a, enter left side of b (phase-column layout)
  // if b is left of a, exit left / enter right
  const ax = a.x + a.w, ay = a.y + Math.min(a.h, 40)/2;
  const bx = b.x,        by = b.y + Math.min(b.h, 40)/2;
  if(bx >= ax){
    const mx = (ax+bx)/2;
    return `M${ax},${ay} C${mx},${ay} ${mx},${by} ${bx},${by}`;
  } else {
    const aLx = a.x, aLy = a.y + Math.min(a.h,40)/2;
    const bRx = b.x + b.w, bRy = b.y + Math.min(b.h,40)/2;
    const mx = (aLx+bRx)/2;
    return `M${aLx},${aLy} C${mx},${aLy} ${mx},${bRy} ${bRx},${bRy}`;
  }
}

/* ---------- draw ---------- */
function render(){
  viewport.innerHTML = "";
  applyTransform();

  // edges first (under nodes)
  const edgeLayer = document.createElementNS("http://www.w3.org/2000/svg","g");
  const nodeLayer = document.createElementNS("http://www.w3.org/2000/svg","g");
  viewport.appendChild(edgeLayer);
  viewport.appendChild(nodeLayer);

  const edgeEls = [];
  EDGES.forEach((e,i)=>{
    const a = LAYOUT[e.from], b = LAYOUT[e.to];
    if(!a || !b) return;
    if(!visiblePhases.has(a.phase) || !visiblePhases.has(b.phase)) return;
    const path = document.createElementNS("http://www.w3.org/2000/svg","path");
    path.setAttribute("d", edgePath(a,b));
    path.setAttribute("class","link");
    path.dataset.from = e.from; path.dataset.to = e.to;
    path.dataset.col = e.col;
    edgeLayer.appendChild(path);
    edgeEls.push(path);

    // invisible hit area for hover
    const hit = document.createElementNS("http://www.w3.org/2000/svg","path");
    hit.setAttribute("d", edgePath(a,b));
    hit.setAttribute("class","link-hit");
    hit.addEventListener("mousemove", ev => showTooltip(ev, e));
    hit.addEventListener("mouseleave", hideTooltip);
    edgeLayer.appendChild(hit);
  });

  // nodes
  const nodeEls = {};
  SCHEMA.forEach(t=>{
    if(!visiblePhases.has(t.phase)) return;
    if(filterText && !t.name.toLowerCase().includes(filterText)) return;
    const p = LAYOUT[t.name]; if(!p) return;
    const g = document.createElementNS("http://www.w3.org/2000/svg","g");
    g.setAttribute("class","node");
    g.setAttribute("transform", `translate(${p.x},${p.y})`);
    g.dataset.name = t.name;

    // box
    const rect = document.createElementNS("http://www.w3.org/2000/svg","rect");
    rect.setAttribute("width", p.w);
    rect.setAttribute("height", p.h);
    rect.setAttribute("rx", 6);
    g.appendChild(rect);

    // phase color bar
    const bar = document.createElementNS("http://www.w3.org/2000/svg","rect");
    bar.setAttribute("width", 4);
    bar.setAttribute("height", p.h);
    bar.setAttribute("rx", 2);
    bar.setAttribute("fill", PHASE_COLOR[t.phase]);
    g.appendChild(bar);

    // title
    const title = document.createElementNS("http://www.w3.org/2000/svg","text");
    title.setAttribute("x", 10); title.setAttribute("y", 15);
    title.setAttribute("class","tname");
    title.textContent = t.name;
    g.appendChild(title);

    // phase label (right aligned)
    const ph = document.createElementNS("http://www.w3.org/2000/svg","text");
    ph.setAttribute("x", p.w - 8); ph.setAttribute("y", 15);
    ph.setAttribute("text-anchor","end");
    ph.setAttribute("class","tphase");
    ph.textContent = PHASE_LABEL[t.phase];
    g.appendChild(ph);

    // columns (pk / fk highlighted)
    t.cols.forEach((c, i)=>{
      const y = 30 + i*ROW_H;
      const txt = document.createElementNS("http://www.w3.org/2000/svg","text");
      txt.setAttribute("x", 12); txt.setAttribute("y", y);
      let label = c[0];
      if(c[2]===1) label = "🔑 " + label;
      else if(c[3]) label = "↳ " + label;
      txt.textContent = label;
      if(c[2]===1) txt.setAttribute("fill","#f6ad55");
      else if(c[3]) txt.setAttribute("fill","#4fd1c5");
      else txt.setAttribute("fill","#9fb0c2");
      g.appendChild(txt);
    });

    g.addEventListener("click", ev => {
      ev.stopPropagation();
      selectTable(t.name);
    });
    g.addEventListener("mousemove", ev => showTooltipNode(ev, t));
    g.addEventListener("mouseleave", hideTooltip);

    nodeLayer.appendChild(g);
    nodeEls[t.name] = g;
  });

  // re-apply selection highlight
  applyHighlight();

  return {edgeEls, nodeEls};
}

/* ---------- selection / highlight ---------- */
function selectTable(name){
  selected = (selected === name) ? null : name;
  applyHighlight();
  renderPanel();
}

function applyHighlight(){
  const allLinks = viewport.querySelectorAll(".link");
  const allNodes = viewport.querySelectorAll(".node");
  if(!selected){
    allLinks.forEach(l=>l.classList.remove("active","dim"));
    allNodes.forEach(n=>n.classList.remove("active","dim"));
    return;
  }
  const connected = new Set([selected]);
  allLinks.forEach(l=>{
    const isRel = l.dataset.from===selected || l.dataset.to===selected;
    l.classList.toggle("active", isRel);
    l.classList.toggle("dim", !isRel);
    if(isRel){ connected.add(l.dataset.from); connected.add(l.dataset.to); }
  });
  allNodes.forEach(n=>{
    const on = connected.has(n.dataset.name);
    n.classList.toggle("active", n.dataset.name===selected);
    n.classList.toggle("dim", !on);
  });
}

/* ---------- tooltips ---------- */
function showTooltip(ev, e){
  const a = LAYOUT[e.from], b = LAYOUT[e.to];
  const r = svg.getBoundingClientRect();
  tooltip.innerHTML = `<div class="tt-t">${e.from} → ${e.to}</div>
    <code>${e.from}.${e.col}</code> references <code>${e.to}.${e.targetPk}</code>`;
  tooltip.style.left = (ev.clientX - r.left + 14) + "px";
  tooltip.style.top  = (ev.clientY - r.top + 12) + "px";
  tooltip.classList.add("show");
}
function showTooltipNode(ev, t){
  const r = svg.getBoundingClientRect();
  const fks = t.cols.filter(c=>c[3]).length;
  tooltip.innerHTML = `<div class="tt-t">${t.name}</div>
    ${t.cols.length} columns · ${fks} foreign key${fks===1?"":"s"} · click to trace`;
  tooltip.style.left = (ev.clientX - r.left + 14) + "px";
  tooltip.style.top  = (ev.clientY - r.top + 12) + "px";
  tooltip.classList.add("show");
}
function hideTooltip(){ tooltip.classList.remove("show"); }

/* ---------- side panel ---------- */
function renderPanel(){
  if(!selected){
    panelBody.innerHTML = `<div class="empty">Click any table in the diagram<br/>to inspect its columns<br/>and foreign-key links.</div>`;
    return;
  }
  const t = SCHEMA.find(x=>x.name===selected);
  const outLinks = EDGES.filter(e=>e.from===selected);
  const inLinks  = EDGES.filter(e=>e.to===selected);

  let html = `<div class="tbl-name">${t.name}</div>
    <span class="badge" style="background:${PHASE_COLOR[t.phase]}">${PHASE_LABEL[t.phase]} · ${t.phase}</span>
    <h3>Columns (${t.cols.length})</h3><ul class="cols">`;
  t.cols.forEach(c=>{
    const cls = c[2]===1 ? "pk" : (c[3] ? "fk" : "");
    let tag = "";
    if(c[2]===1) tag = `<span class="tag pk">PK</span>`;
    else if(c[3]) tag = `<span class="tag fk">FK→${c[3]}</span>`;
    html += `<li class="${cls}">
      <span class="col-name">${c[0]}${tag}</span>
      <span class="col-type">${c[1]}</span></li>`;
  });
  html += `</ul>`;

  html += `<h3>Outgoing FK (${outLinks.length})</h3>`;
  if(!outLinks.length) html += `<div class="rel lbl">— none —</div>`;
  outLinks.forEach(e=>{
    html += `<div class="rel" data-goto="${e.to}">
      <span class="arrow">→</span> <b>${e.to}</b>
      <div class="lbl">${e.col} → ${e.to}.${e.targetPk}</div></div>`;
  });

  html += `<h3>Incoming FK (${inLinks.length})</h3>`;
  if(!inLinks.length) html += `<div class="rel lbl">— none —</div>`;
  inLinks.forEach(e=>{
    html += `<div class="rel" data-goto="${e.from}">
      <span class="arrow">←</span> <b>${e.from}</b>
      <div class="lbl">${e.from}.${e.col} → ${e.to}.${e.targetPk}</div></div>`;
  });

  panelBody.innerHTML = html;
  panelBody.querySelectorAll(".rel[data-goto]").forEach(el=>{
    el.addEventListener("click", ()=> selectTable(el.dataset.goto));
  });
}

/* ---------- pan & zoom ---------- */
svg.addEventListener("mousedown", e=>{
  if(e.target.closest(".node")) return;
  dragging = true; dragStart = {x:e.clientX, y:e.clientY, tx, ty};
  svg.classList.add("dragging");
});
window.addEventListener("mousemove", e=>{
  if(!dragging) return;
  tx = dragStart.tx + (e.clientX - dragStart.x);
  ty = dragStart.ty + (e.clientY - dragStart.y);
  applyTransform();
});
window.addEventListener("mouseup", ()=>{ dragging=false; svg.classList.remove("dragging"); });
svg.addEventListener("wheel", e=>{
  e.preventDefault();
  const r = svg.getBoundingClientRect();
  const mx = e.clientX - r.left, my = e.clientY - r.top;
  const factor = e.deltaY < 0 ? 1.1 : 0.9;
  const ns = Math.min(2.5, Math.max(0.25, scale*factor));
  tx = mx - (mx - tx) * (ns/scale);
  ty = my - (my - ty) * (ns/scale);
  scale = ns;
  applyTransform();
}, {passive:false});

/* ---------- toolbar wiring ---------- */
document.getElementById("search").addEventListener("input", e=>{
  filterText = e.target.value.trim().toLowerCase();
  render(); applyHighlight();
});
document.querySelectorAll(".chip[data-phase]").forEach(ch=>{
  ch.addEventListener("click", ()=>{
    const ph = ch.dataset.phase;
    if(visiblePhases.has(ph)){ visiblePhases.delete(ph); ch.classList.remove("on"); }
    else { visiblePhases.add(ph); ch.classList.add("on"); }
    render(); applyHighlight();
  });
});
document.querySelector(".chip[data-act=reset]").addEventListener("click", ()=>{
  tx=0; ty=0; scale=1; selected=null; applyTransform(); applyHighlight(); renderPanel();
});
let layoutMode = "phase";
document.querySelector(".chip[data-act=layout]").addEventListener("click", function(){
  layoutMode = layoutMode==="phase" ? "radial" : "phase";
  this.textContent = "layout: " + (layoutMode==="phase" ? "phase-columns" : "radial");
  LAYOUT = layoutMode==="phase" ? layoutPhaseColumns() : layoutRadial();
  render(); applyHighlight();
});

/* ---------- boot ---------- */
render();
renderPanel();
</script>
</body>
</html>
```

---

## What you get

### 1. The "Glue" spine is visible at a glance
The `Core` column (amber) contains **project, site, wbs, spatial_zone_room, stakeholder** — every other phase's tables have FK edges fanning back into it. Selecting `project` lights up ~20 incoming links, confirming it is the master root.

### 2. Cross-phase traceability is one click away
- Click **`asset`** (pink, P6–7): you'll see incoming edges from `commissioning_test`, `punch_item`, `municipal_inspection`, `defect`, `maintenance_schedule`, `work_order`, `facility_telemetry` — and **outgoing** edges to `bim_element` (P2), `spatial_zone_room` (core), and `purchase_order` (P4). That is exactly the "handover bridge" your document describes: BIM design ↔ procurement cost ↔ operations asset.
- Click **`wbs`** to confirm it is the cost/scope spine linking `boq_item`, `schedule_activity`, `permit`, `risk_register`, `payment_application`, `change_order`, `quality_test`, and `ncr` across phases 2–5.

### 3. Subtype inheritance is rendered as FK
`structural_element.element_id` and `mep_system.element_id` both point to `bim_element.element_id` (marked PK/FK), so the unified BIM parent-child pattern is explicit in the graph rather than implied.

### 4. Phase filter + search
Toggle `P5 Build` off to see how the design and procurement subgraphs remain connected through the core spine even when transactional volume tables are hidden. Type `asset` in the filter to isolate the handover cluster.

### 5. Inspector panel
For any selected table you get the full column list with `PK`/`FK→target` tags, plus **clickable outgoing and incoming relationships** that jump to the linked table — a lightweight graph navigator.

### 6. Layout switch
`layout: phase-columns` mirrors the document's phase-by-phase narrative (left→right = core → design → procure → build → O&M). Switch to `radial` to see the core tables in the centre with phases arranged in opposing arcs, which makes the "everything hangs off project/wbs" structure visually obvious.

---

## Notes on fidelity to your source document

- All 37 entities in the file are present, including the **bridge table** `activity_dependency` (two FKs to `schedule_activity`, self-referencing pattern) and the **subtype pair** `structural_element` / `mep_system` keyed on `bim_element.element_id`.
- `financial_metric` includes the `version` column (Baseline vs. Revised) as specified.
- `facility_telemetry` is modelled as a narrow (long-format) time-series table keyed by `asset_id` + `timestamp` + `metric_name`, consistent with the TimescaleDB hypertable recommendation in §3.1.
- `asset.po_id` (→ purchase_order) and `asset.element_id` (→ bim_element) are both drawn, preserving the cost-and-design lineage the document calls out.
- The four architectural recommendations (§3) are reflected in the schema shape, not in the ERD chrome: `ifc_guid` is on `bim_element`, `wbs_code` is on `wbs` for Uniclass/MasterFormat, and soft-delete/audit tables are intentionally omitted from the visual (they would be side tables per entity) — you can add an `audit_log` node later if you want it drawn.
````

````python
Here is the complete, executable PostgreSQL DDL (CREATE TABLE) script implementing the 7-phase "Golden Thread" construction schema.
It uses UUIDs for distributed/BIM sync compatibility, PostGIS for spatial coordinates, soft-delete flags (is_archived), and proper foreign key constraints linking Phase 1 through Phase 7.
sql
-- ============================================================================
-- CONSTRUCTION LIFECYCLE "GOLDEN THREAD" SCHEMA (POSTGRESQL 15+)
-- Covers Phases 1 through 7 with Unified WBS, Spatial, BIM, and Asset Spines
-- ============================================================================

-- Enable required extensions
CREATE EXTENSION IF NOT EXISTS "uuid-ossp";
CREATE EXTENSION IF NOT EXISTS "postgis";

-- ============================================================================
-- 0. MASTER SPINE TABLES (Core Anchors Across All Phases)
-- ============================================================================

CREATE TABLE project (
    project_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_code VARCHAR(50) UNIQUE NOT NULL,
    name VARCHAR(255) NOT NULL,
    rom_budget NUMERIC(18, 2),
    approved_budget NUMERIC(18, 2),
    gfa_sqm NUMERIC(12, 2),
    target_start_date DATE,
    target_end_date DATE,
    actual_start_date DATE,
    certificate_of_occupancy_date DATE,
    current_phase SMALLINT CHECK (current_phase BETWEEN 1 AND 7) DEFAULT 1,
    is_archived BOOLEAN DEFAULT FALSE,
    created_at TIMESTAMPTZ DEFAULT NOW(),
    updated_at TIMESTAMPTZ DEFAULT NOW()
);

CREATE TABLE stakeholder (
    stakeholder_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    organization_name VARCHAR(255) NOT NULL,
    role VARCHAR(100) NOT NULL, -- Owner, Architect, Engineer, GC, Subcontractor, Authority, Vendor
    contact_name VARCHAR(255),
    email VARCHAR(255),
    phone VARCHAR(50),
    is_archived BOOLEAN DEFAULT FALSE,
    created_at TIMESTAMPTZ DEFAULT NOW()
);

CREATE TABLE project_stakeholder (
    project_id UUID REFERENCES project(project_id) ON DELETE RESTRICT,
    stakeholder_id UUID REFERENCES stakeholder(stakeholder_id) ON DELETE RESTRICT,
    project_role VARCHAR(100) NOT NULL,
    PRIMARY KEY (project_id, stakeholder_id, project_role)
);

CREATE TABLE wbs (
    wbs_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    parent_wbs_id UUID REFERENCES wbs(wbs_id) ON DELETE RESTRICT,
    wbs_code VARCHAR(50) NOT NULL, -- e.g., Uniclass / MasterFormat / OmniClass
    description VARCHAR(255) NOT NULL,
    level_depth SMALLINT DEFAULT 1,
    UNIQUE (project_id, wbs_code)
);

CREATE TABLE room (
    room_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    level_floor VARCHAR(50) NOT NULL,
    room_number VARCHAR(50) NOT NULL,
    room_name VARCHAR(150),
    net_area_sqm NUMERIC(10, 2),
    gross_area_sqm NUMERIC(10, 2),
    occupancy_load INT,
    acoustic_rating_stc INT,
    fire_rating_minutes INT,
    UNIQUE (project_id, level_floor, room_number)
);

-- ============================================================================
-- 1. PHASE 1: CONCEPTION AND INITIATION
-- ============================================================================

CREATE TABLE site (
    site_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    parcel_number VARCHAR(100),
    boundary_geom GEOMETRY(Polygon, 4326),
    plot_area_sqm NUMERIC(14, 2),
    far_allowed NUMERIC(6, 2),
    zoning_classification VARCHAR(100),
    soil_bearing_capacity_kpa NUMERIC(10, 2),
    water_table_depth_m NUMERIC(8, 2),
    eia_approval_status VARCHAR(50),
    created_at TIMESTAMPTZ DEFAULT NOW()
);

CREATE TABLE financial_metric (
    metric_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    version_label VARCHAR(50) NOT NULL, -- e.g., 'Initial Feasibility', 'FID Baseline'
    roi_pct NUMERIC(8, 4),
    npv_amount NUMERIC(18, 2),
    irr_pct NUMERIC(8, 4),
    cap_rate_pct NUMERIC(8, 4),
    calculated_at TIMESTAMPTZ DEFAULT NOW()
);

CREATE TABLE funding_source (
    funding_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    stakeholder_id UUID REFERENCES stakeholder(stakeholder_id),
    source_type VARCHAR(50) NOT NULL, -- Equity, Senior Debt, Mezzanine, Grant
    committed_amount NUMERIC(18, 2) NOT NULL,
    interest_rate_pct NUMERIC(6, 3)
);

-- ============================================================================
-- 2. PHASE 2: DESIGN AND ENGINEERING
-- ============================================================================

CREATE TABLE bim_element (
    element_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    room_id UUID REFERENCES room(room_id),
    wbs_id UUID REFERENCES wbs(wbs_id),
    ifc_guid VARCHAR(64) NOT NULL, -- Immutable BIM Global Unique ID
    ifc_entity_type VARCHAR(100), -- e.g., IfcColumn, IfcAirTerminal
    discipline VARCHAR(50) NOT NULL, -- Structural, MEP, Architectural
    lod_level SMALLINT CHECK (lod_level IN (100, 200, 300, 350, 400, 500)),
    UNIQUE (project_id, ifc_guid)
);

CREATE TABLE structural_element (
    element_id UUID PRIMARY KEY REFERENCES bim_element(element_id) ON DELETE CASCADE,
    concrete_grade_mpa NUMERIC(8, 2),
    rebar_tonnage NUMERIC(10, 4),
    steel_section_weight_kg NUMERIC(12, 2),
    wind_load_kpa NUMERIC(8, 3),
    seismic_design_category VARCHAR(20),
    foundation_depth_m NUMERIC(8, 2)
);

CREATE TABLE mep_system (
    element_id UUID PRIMARY KEY REFERENCES bim_element(element_id) ON DELETE CASCADE,
    system_category VARCHAR(50) NOT NULL, -- HVAC, Electrical, Plumbing, Fire Protection
    thermal_load_kw NUMERIC(10, 2),
    electrical_demand_kva NUMERIC(10, 2),
    lighting_lux NUMERIC(8, 2),
    flow_rate_lps NUMERIC(10, 3),
    pipe_slope_pct NUMERIC(6, 3)
);

CREATE TABLE design_control (
    control_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    element_id UUID REFERENCES bim_element(element_id),
    document_number VARCHAR(100) NOT NULL,
    revision_number VARCHAR(20) NOT NULL,
    clash_count INT DEFAULT 0,
    approval_status VARCHAR(50) DEFAULT 'Draft', -- Draft, Reviewed, IFC, Superseded
    approved_by UUID REFERENCES stakeholder(stakeholder_id),
    approved_at TIMESTAMPTZ
);

-- ============================================================================
-- 3. PHASE 3: PRECONSTRUCTION AND PERMITTING
-- ============================================================================

CREATE TABLE boq_item (
    boq_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    wbs_id UUID NOT NULL REFERENCES wbs(wbs_id),
    element_id UUID REFERENCES bim_element(element_id),
    item_code VARCHAR(50) NOT NULL,
    description TEXT NOT NULL,
    unit_of_measure VARCHAR(20) NOT NULL,
    estimated_qty NUMERIC(14, 4) NOT NULL,
    unit_material_rate NUMERIC(14, 2) DEFAULT 0,
    unit_labor_rate NUMERIC(14, 2) DEFAULT 0,
    productivity_norm_hrs_per_unit NUMERIC(10, 4),
    total_estimated_cost NUMERIC(18, 2) GENERATED ALWAYS AS (estimated_qty * (unit_material_rate + unit_labor_rate)) STORED
);

CREATE TABLE contingency (
    contingency_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    wbs_id UUID REFERENCES wbs(wbs_id),
    category VARCHAR(50) NOT NULL, -- Design Contingency, Construction Reserve, VE Savings
    allocated_amount NUMERIC(18, 2) NOT NULL,
    utilized_amount NUMERIC(18, 2) DEFAULT 0
);

CREATE TABLE schedule_activity (
    activity_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    wbs_id UUID NOT NULL REFERENCES wbs(wbs_id),
    room_id UUID REFERENCES room(room_id),
    activity_code VARCHAR(50) NOT NULL,
    activity_name VARCHAR(255) NOT NULL,
    planned_start DATE NOT NULL,
    planned_end DATE NOT NULL,
    actual_start DATE,
    actual_end DATE,
    duration_days INT NOT NULL,
    total_float_days INT DEFAULT 0,
    is_critical_path BOOLEAN DEFAULT FALSE,
    is_milestone BOOLEAN DEFAULT FALSE,
    UNIQUE (project_id, activity_code)
);

CREATE TABLE activity_dependency (
    predecessor_id UUID REFERENCES schedule_activity(activity_id) ON DELETE CASCADE,
    successor_id UUID REFERENCES schedule_activity(activity_id) ON DELETE CASCADE,
    dependency_type VARCHAR(5) CHECK (dependency_type IN ('FS', 'SS', 'FF', 'SF')) DEFAULT 'FS',
    lag_days INT DEFAULT 0,
    PRIMARY KEY (predecessor_id, successor_id)
);

CREATE TABLE permit (
    permit_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    authority_stakeholder_id UUID REFERENCES stakeholder(stakeholder_id),
    permit_type VARCHAR(100) NOT NULL,
    application_number VARCHAR(100),
    fee_amount NUMERIC(14, 2),
    submitted_date DATE,
    approved_date DATE,
    expiry_date DATE,
    status VARCHAR(50) DEFAULT 'Pending'
);

CREATE TABLE risk_register (
    risk_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    wbs_id UUID REFERENCES wbs(wbs_id),
    owner_stakeholder_id UUID REFERENCES stakeholder(stakeholder_id),
    risk_category VARCHAR(50), -- Technical, Safety, Financial, Regulatory, Supply Chain
    description TEXT NOT NULL,
    probability_score SMALLINT CHECK (probability_score BETWEEN 1 AND 5),
    impact_score SMALLINT CHECK (impact_score BETWEEN 1 AND 5),
    mitigation_plan TEXT,
    status VARCHAR(50) DEFAULT 'Open'
);

-- ============================================================================
-- 4. PHASE 4: BIDDING AND PROCUREMENT
-- ============================================================================

CREATE TABLE bid_package (
    bid_package_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    wbs_id UUID REFERENCES wbs(wbs_id),
    package_name VARCHAR(255) NOT NULL,
    delivery_method VARCHAR(50), -- DBB, Design-Build, EPC, CMAR
    estimated_budget NUMERIC(18, 2),
    tender_issue_date DATE,
    tender_close_date DATE,
    clarifications_count INT DEFAULT 0,
    bid_variance_pct NUMERIC(8, 2),
    status VARCHAR(50) DEFAULT 'Open'
);
CREATE TABLE contract (
    contract_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    bid_package_id UUID REFERENCES bid_package(bid_package_id),
    contractor_stakeholder_id UUID NOT NULL REFERENCES stakeholder(stakeholder_id),
    contract_number VARCHAR(100) UNIQUE NOT NULL,
    award_value NUMERIC(18, 2) NOT NULL,
    retention_pct NUMERIC(5, 2) DEFAULT 10.00,
    performance_bond_number VARCHAR(100),
    insurance_expiry_date DATE,
    defects_liability_months INT DEFAULT 12,
    award_date DATE NOT NULL
);
CREATE TABLE payment_milestone (
    milestone_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    contract_id UUID NOT NULL REFERENCES contract(contract_id) ON DELETE CASCADE,
    activity_id UUID REFERENCES schedule_activity(activity_id),
    description VARCHAR(255) NOT NULL,
    milestone_pct NUMERIC(6, 2),
    milestone_amount NUMERIC(18, 2) NOT NULL,
    due_date DATE,
    is_achieved BOOLEAN DEFAULT FALSE
);
CREATE TABLE purchase_order (
    po_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    contract_id UUID REFERENCES contract(contract_id),
    boq_id UUID REFERENCES boq_item(boq_id),
    vendor_stakeholder_id UUID NOT NULL REFERENCES stakeholder(stakeholder_id),
    po_number VARCHAR(100) UNIQUE NOT NULL,
    ordered_qty NUMERIC(14, 4) NOT NULL,
    unit_cost NUMERIC(14, 2) NOT NULL,
    lead_time_days INT,
    issued_date DATE NOT NULL
);
CREATE TABLE supply_shipment (
    shipment_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    po_id UUID NOT NULL REFERENCES purchase_order(po_id) ON DELETE CASCADE,
    tracking_reference VARCHAR(100),
    fabrication_progress_pct NUMERIC(5, 2) DEFAULT 0,
    estimated_departure DATE,
    estimated_arrival DATE,
    customs_cleared_date DATE,
    delivered_to_site_date DATE,
    status VARCHAR(50) DEFAULT 'Fabricating'
);
-- ============================================================================
-- 5. PHASE 5: PHYSICAL CONSTRUCTION
-- ============================================================================
CREATE TABLE daily_progress (
    progress_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    activity_id UUID NOT NULL REFERENCES schedule_activity(activity_id),
    boq_id UUID REFERENCES boq_item(boq_id),
    log_date DATE NOT NULL,
    trade_headcount INT DEFAULT 0,
    man_hours_worked NUMERIC(10, 2) DEFAULT 0,
    equipment_hours_used NUMERIC(10, 2) DEFAULT 0,
    percent_complete NUMERIC(5, 2) CHECK (percent_complete BETWEEN 0 AND 100),
    earned_value_ev NUMERIC(18, 2),
    actual_cost_ac NUMERIC(18, 2)
);
CREATE TABLE rfi (
    rfi_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    element_id UUID REFERENCES bim_element(element_id),
    activity_id UUID REFERENCES schedule_activity(activity_id),
    raised_by UUID REFERENCES stakeholder(stakeholder_id),
    assigned_to UUID REFERENCES stakeholder(stakeholder_id),
    rfi_number VARCHAR(50) NOT NULL,
    subject VARCHAR(255) NOT NULL,
    submitted_at TIMESTAMPTZ DEFAULT NOW(),
    due_date DATE,
    closed_at TIMESTAMPTZ,
    has_cost_impact BOOLEAN DEFAULT FALSE,
    has_schedule_impact BOOLEAN DEFAULT FALSE,
    status VARCHAR(50) DEFAULT 'Open',
    UNIQUE (project_id, rfi_number)
);
CREATE TABLE submittal (
    submittal_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    wbs_id UUID REFERENCES wbs(wbs_id),
    element_id UUID REFERENCES bim_element(element_id),
    submittal_number VARCHAR(50) NOT NULL,
    title VARCHAR(255) NOT NULL,
    submitted_date DATE,
    reviewed_date DATE,
    review_outcome VARCHAR(50), -- Approved, Approved as Noted, Revise & Resubmit, Rejected
    UNIQUE (project_id, submittal_number)
);
CREATE TABLE quality_test (
    test_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    wbs_id UUID REFERENCES wbs(wbs_id),
    room_id UUID REFERENCES room(room_id),
    element_id UUID REFERENCES bim_element(element_id),
    test_type VARCHAR(100) NOT NULL, -- Concrete Cube Break, Weld X-Ray, Soil Compaction
    sample_reference VARCHAR(100),
    target_specification VARCHAR(100),
    measured_result VARCHAR(100),
    passed BOOLEAN NOT NULL,
    tested_at TIMESTAMPTZ DEFAULT NOW()
);
CREATE TABLE ncr (
    ncr_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    quality_test_id UUID REFERENCES quality_test(test_id),
    element_id UUID REFERENCES bim_element(element_id),
    responsible_contract_id UUID REFERENCES contract(contract_id),
    ncr_number VARCHAR(50) NOT NULL,
    description TEXT NOT NULL,
    root_cause VARCHAR(255),
    corrective_action TEXT,
    issued_date DATE NOT NULL,
    closed_date DATE,
    status VARCHAR(50) DEFAULT 'Open',
    UNIQUE (project_id, ncr_number)
);
CREATE TABLE hse_record (
    hse_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    log_date DATE NOT NULL,
    toolbox_talk_attendees INT DEFAULT 0,
    near_miss_count INT DEFAULT 0,
    lost_time_injuries INT DEFAULT 0,
    ltifr_rolling NUMERIC(8, 4),
    ppe_compliance_pct NUMERIC(5, 2),
    waste_generated_tonnes NUMERIC(10, 3),
    embodied_operational_carbon_kg NUMERIC(14, 2),
    UNIQUE (project_id, log_date)
);
CREATE TABLE site_weather_log (
    weather_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    log_date DATE NOT NULL,
    temp_max_c NUMERIC(5, 2),
    temp_min_c NUMERIC(5, 2),
    wind_speed_max_kmh NUMERIC(6, 2),
    rainfall_mm NUMERIC(7, 2),
    weather_delay_hours NUMERIC(4, 1) DEFAULT 0,
    drone_imagery_uri TEXT,
    UNIQUE (project_id, log_date)
);
CREATE TABLE payment_application (
    pay_app_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    contract_id UUID NOT NULL REFERENCES contract(contract_id) ON DELETE RESTRICT,
    application_number INT NOT NULL,
    period_end_date DATE NOT NULL,
    claimed_amount NUMERIC(18, 2) NOT NULL,
    certified_amount NUMERIC(18, 2),
    retention_deducted NUMERIC(18, 2),
    net_payable NUMERIC(18, 2),
    paid_date DATE,
    status VARCHAR(50) DEFAULT 'Submitted',
    UNIQUE (contract_id, application_number)
);
CREATE TABLE change_order (
    co_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    contract_id UUID NOT NULL REFERENCES contract(contract_id) ON DELETE RESTRICT,
    wbs_id UUID REFERENCES wbs(wbs_id),
    rfi_id UUID REFERENCES rfi(rfi_id),
    co_number VARCHAR(50) NOT NULL,
    title VARCHAR(255) NOT NULL,
    cost_impact NUMERIC(18, 2) NOT NULL,
    schedule_impact_days INT DEFAULT 0,
    status VARCHAR(50) DEFAULT 'Pending', -- Pending, Approved, Rejected
    approved_date DATE,
    UNIQUE (contract_id, co_number)
);
-- ============================================================================
-- 6. PHASE 6: COMMISSIONING AND CLOSEOUT
-- ============================================================================
CREATE TABLE asset (
    asset_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    element_id UUID REFERENCES bim_element(element_id), -- Bridges Phase 2 BIM
    room_id UUID REFERENCES room(room_id),              -- Bridges Phase 2 Space
    po_id UUID REFERENCES purchase_order(po_id),        -- Bridges Phase 4 Cost
    asset_tag_qr VARCHAR(100) UNIQUE NOT NULL,
    serial_number VARCHAR(150),
    asset_name VARCHAR(255) NOT NULL,
    manufacturer VARCHAR(150),
    model_number VARCHAR(150),
    warranty_start_date DATE,
    warranty_end_date DATE,
    spare_parts_qty INT DEFAULT 0,
    as_built_doc_uri TEXT,
    lifecycle_replacement_cost NUMERIC(18, 2)
);
    CREATE TABLE commissioning_test (
    comm_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    asset_id UUID NOT NULL REFERENCES asset(asset_id) ON DELETE CASCADE,
    test_parameter VARCHAR(100) NOT NULL, -- HVAC CFM, Hydro Pressure Bar, Insulation Ohms, Fire Alarm Sec
    design_target_value NUMERIC(12, 4),
    measured_value NUMERIC(12, 4),
    unit_of_measure VARCHAR(30),
    passed BOOLEAN NOT NULL,
    witnessed_by UUID REFERENCES stakeholder(stakeholder_id),
    tested_at TIMESTAMPTZ DEFAULT NOW()
);
CREATE TABLE punch_item (
    punch_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    room_id UUID REFERENCES room(room_id),
    asset_id UUID REFERENCES asset(asset_id),
    assigned_contract_id UUID REFERENCES contract(contract_id),
    description TEXT NOT NULL,
    severity VARCHAR(20) CHECK (severity IN ('Minor', 'Major', 'Critical')),
    logged_date DATE NOT NULL,
    resolved_date DATE,
    verified_closed BOOLEAN DEFAULT FALSE
);
CREATE TABLE municipal_inspection (
    inspection_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    authority_stakeholder_id UUID REFERENCES stakeholder(stakeholder_id),
    inspection_type VARCHAR(100) NOT NULL, -- Structural, Fire Life Safety, Final Occupancy
    scheduled_date DATE NOT NULL,
    passed BOOLEAN,
    inspector_notes TEXT
);
-- ============================================================================
-- 7. PHASE 7: OPERATION AND MAINTENANCE (POST-CONSTRUCTION)
-- ============================================================================
CREATE TABLE defect (
    defect_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id UUID NOT NULL REFERENCES project(project_id) ON DELETE RESTRICT,
    asset_id UUID REFERENCES asset(asset_id),
    room_id UUID REFERENCES room(room_id),
    responsible_contract_id UUID REFERENCES contract(contract_id),
    description TEXT NOT NULL,
    reported_at TIMESTAMPTZ DEFAULT NOW(),
    sla_response_hours INT,
    repaired_at TIMESTAMPTZ,
    retention_release_approved BOOLEAN DEFAULT FALSE
);
CREATE TABLE maintenance_schedule (
    pm_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    asset_id UUID NOT NULL REFERENCES asset(asset_id) ON DELETE CASCADE,
    task_description VARCHAR(255) NOT NULL,
    frequency_days INT NOT NULL,
    last_performed_date DATE,
    next_due_date DATE NOT NULL
);
CREATE TABLE work_order (
    work_order_id UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    asset_id UUID NOT NULL REFERENCES asset(asset_id) ON DELETE RESTRICT,
    pm_id UUID REFERENCES maintenance_schedule(pm_id), -- Null if reactive breakdown
    work_type VARCHAR(50) CHECK (work_type IN ('Preventive', 'Corrective', 'Emergency')),
    opened_at TIMESTAMPTZ DEFAULT NOW(),
    completed_at TIMESTAMPTZ,
    downtime_hours NUMERIC(8, 2) DEFAULT 0,
    labor_and_parts_cost NUMERIC(14, 2) DEFAULT 0
);
CREATE TABLE facility_telemetry (
    asset_id UUID NOT NULL REFERENCES asset(asset_id) ON DELETE CASCADE,
    recorded_at TIMESTAMPTZ NOT NULL,
    metric_name VARCHAR(50) NOT NULL, -- energy_kwh, water_m3, co2_ppm, supply_temp_c
    metric_value NUMERIC(14, 4) NOT NULL,
    PRIMARY KEY (asset_id, recorded_at, metric_name)
);
-- ============================================================================
-- PERFORMANCE INDEXES FOR CROSS-PHASE QUERIES
-- ============================================================================
CREATE INDEX idx_wbs_project ON wbs(project_id);
CREATE INDEX idx_bim_wbs_room ON bim_element(wbs_id, room_id);
CREATE INDEX idx_boq_wbs ON boq_item(wbs_id);
CREATE INDEX idx_schedule_wbs ON schedule_activity(wbs_id, planned_start, planned_end);
CREATE INDEX idx_progress_activity_date ON daily_progress(activity_id, log_date);
CREATE INDEX idx_asset_bim_room ON asset(element_id, room_id);
CREATE INDEX idx_telemetry_time ON facility_telemetry(recorded_at DESC);

### Key Architectural Highlights in This Script
* **Computed Columns:** `boq_item.total_estimated_cost` automatically computes `estimated_qty * (unit_material_rate + unit_labor_rate)` inside PostgreSQL so estimates never drift out of sync.
* **Subtyped Inheritance:** `bim_element` serves as the unified BIM parent table, while `structural_element` and `mep_system` enforce discipline-specific engineering attributes via 1:1 primary-foreign keys.
* **Ready for TimescaleDB:** If your `facility_telemetry` table grows to millions of IoT sensor readings, you can immediately convert it into a partitioned hypertable by running:
  `SELECT create_hypertable('facility_telemetry', 'recorded_at');`

<FollowUp>
Would you like me to add **SQL views for cross-phase executive KPIs** (such as Earned Value CPI/SPI, BOQ vs. Actual Cost variance, and Asset Lifecycle TCO) or generate **sample seed data (`INSERT` statements)** to test queries against this schema?
</FollowUp>
````

````python
To prevent heavy 3D BIM uploads (Phase 2) and high-frequency IoT sensor streams (Phase 7) from slowing down daily field transactions (Phases 3–6), your SaaS platform needs a decoupled, event-driven ingestion architecture.
Below is the end-to-end architecture connecting external BIM authoring tools and building sensors into your PostgreSQL + TimescaleDB schema.

1. High-Level Pipeline Architecture

text
[PHASE 2: BIM AUTHORING]                       [PHASE 7: SMART BUILDING O&M]
Revit / Navisworks / Civil 3D                  BMS (BACnet/Modbus) / LoRaWAN / Meters
        │                                                       │
        ▼ (1. Pre-signed URL Upload)                            ▼ (1. TLS / mTLS Stream)
┌───────────────────────┐                               ┌───────────────────────┐
│   Object Storage (S3) │                               │  MQTT Broker (EMQX)   │
│  Raw .ifc / .rvt files│                               │  Topic: tenant/asset  │
└───────┬───────────────┘                               └───────┬───────────────┘
        │ (2. S3 Event Webhook)                                 │ (2. High-Throughput)
        ▼                                                       ▼
┌───────────────────────┐                               ┌───────────────────────┐
│  Message Queue (SQS/  │                               │ Stream Buffer (Kafka  │
│  RabbitMQ)            │                               │ / Redis Streams)      │
└───────┬───────────────┘                               └───────┬───────────────┘
        │ (3. Async Pull)                                       │ (3. Micro-Batching)
        ▼                                                       ▼
┌───────────────────────┐                               ┌───────────────────────┐
│  BIM Parser Workers   │                               │ IoT Ingestion Service │
│  (IfcOpenShell / Py)  │                               │ (Go / Rust / Vector)  │
└───────┬───────────────┘                               └───────┬───────────────┘
        │ (4. Upsert via ifc_guid)                              │ (4. COPY Bulk Insert)
        └───────────────────────────┐   ┌───────────────────────┘
                                    ▼   ▼
                    ┌───────────────────────────────────┐
                    │   PostgreSQL 16 + TimescaleDB     │
                    │  • bim_element / structural / mep │
                    │  • facility_telemetry (Hypertable)│
                    └───────────────────────────────────┘
Use code with caution.

2. Pipeline A: Asynchronous BIM Ingestion (Phases 2, 5 & 6)

A single IFC file can contain 100,000+ objects and take 2 to 5 minutes to parse. Never process BIM files inside your main API request cycle.
1. Direct-to-Object-Storage Upload:
	• The user's browser or Revit plugin requests a Pre-Signed S3 URL from your API and uploads the .ifc file directly to S3/Blob storage, bypassing your API servers completely.
2. Event Trigger & Queue:
	• S3 emits an ObjectCreated event into a job queue (RabbitMQ or AWS SQS) with metadata (tenant_id, project_id, revision_number).
3. Isolated Worker Parsing (IfcOpenShell):
	• Dedicated CPU/RAM worker containers pull the job and parse the IFC schema using IfcOpenShell (Python) or xBIM (.NET):
		• Extracts spatial hierarchy (IfcBuildingStorey, IfcSpace) → upserts into room.
		• Extracts physical objects (IfcColumn, IfcBeam, IfcDuctSegment) → upserts into bim_element using ifc_guid as the conflict key (ON CONFLICT (project_id, ifc_guid) DO UPDATE).
		• Maps engineering properties (concrete MPa, kVA demand, airflow) into structural_element and mep_system.
		• Auto-links wbs_id by matching the element's Uniclass/OmniClass classification attribute against your wbs table.
4. Revision Diffing & Alerts:
	• The worker compares the new model against the previous revision, flags deleted or moved elements that have open RFIs (rfi) or purchase orders (purchase_order), and logs a summary row in design_control.

3. Pipeline B: High-Velocity IoT Telemetry Ingestion (Phase 7)

Once the building enters Phase 7 O&M, thousands of sensors publish readings every 1 to 60 seconds. Individual INSERT statements will exhaust PostgreSQL connections; you must batch-ingest them.
1. Edge Gateway Protocol Normalization:
	• On-site Building Management Systems (BMS) speak legacy protocols (BACnet, Modbus, KNX). An on-premise edge gateway (e.g., Node-RED, Niagara, or AWS IoT Greengrass) translates those signals into lightweight JSON payloads and publishes them over MQTT (TLS 1.3) to the cloud.
	• Topic structure: v1/{tenant_id}/{project_id}/{asset_tag_qr}/telemetry
2. MQTT Broker & Stream Buffer:
	• EMQX (or HiveMQ) terminates thousands of concurrent edge connections and forwards messages into Redis Streams or Apache Kafka so zero sensor data is lost if the database briefly restarts.
3. Micro-Batch Consumer & Asset Resolution:
	• A lightweight ingestion service (written in Go or Rust) reads messages from the stream in batches (e.g., every 5,000 rows or every 2 seconds):
		• Resolves asset_tag_qr to asset_id using an in-memory Redis lookup cache (avoiding repetitive SELECT queries).
		• Validates payload bounds (filtering out sensor spikes or NaN errors).
		• Executes a single PostgreSQL COPY facility_telemetry FROM STDIN command, which inserts 50,000+ rows per second with minimal CPU overhead.
4. Automated Alerting & Work Order Triggering:
	• Using TimescaleDB Continuous Aggregates, the database automatically computes 15-minute rolling averages (energy kWh, chiller delta-T, CO₂ ppm). If a threshold breaches design limits from Phase 2 (mep_system), an event automatically creates a reactive ticket in work_order.

4. Key Data Integrity Safeguards

• Idempotency Keys: Both pipelines must be safely retryable. BIM upserts rely on (project_id, ifc_guid), while IoT telemetry relies on the composite primary key (asset_id, recorded_at, metric_name) with ON CONFLICT DO NOTHING to prevent duplicate sensor readings during network retries.
• Dead-Letter Queues (DLQ): Any malformed IFC element or unrecognized asset_tag_qr is routed to a Dead-Letter Queue for BIM/CAFM administrators to inspect rather than crashing the ingestion pipeline.
• Time-Series Retention Policy: Configure TimescaleDB to compress facility_telemetry chunks older than 7 days and automatically downsample raw 5-second data into hourly averages after 1 year to keep SaaS storage costs predictable.
````

````python
Adding a carbon credit layer on top of the existing SDG compliance schema is a logical extension: the SDG tables track *performance against targets*, while carbon credits represent the *tradable instrument* that can monetize verified over-performance or offset residual emissions. The design below connects credits to the same "golden thread" (project → WBS → activity/asset) and accounts for the UAE's National Carbon Credit Registry framework, which is directly relevant given the project context.

---

## 1. Schema extension: carbon credit tables

```sql
-- ============================================================================
-- CARBON CREDIT LAYER (extends the SDG compliance extension)
-- ============================================================================

-- Credit-producing or credit-consuming project/activity registry
CREATE TABLE carbon_project (
    carbon_project_id   UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    project_id          UUID NOT NULL REFERENCES project(project_id),
    wbs_id              UUID REFERENCES wbs(wbs_id),           -- optional scope link
    methodology         VARCHAR(100) NOT NULL,                  -- e.g. 'VM0042', 'AMS-I.D.'
    standard_body       VARCHAR(50) NOT NULL CHECK (standard_body IN ('Verra','GoldStandard','UAE_National','Other')),
    registry_id         VARCHAR(100),                           -- serial/account reference
    baseline_start      DATE NOT NULL,
    baseline_end        DATE NOT NULL,
    crediting_start     DATE NOT NULL,
    crediting_end       DATE NOT NULL,
    status              VARCHAR(30) DEFAULT 'Design' CHECK (status IN ('Design','Validated','Registered','Verification','Issued','Retired')),
    created_at          TIMESTAMPTZ DEFAULT NOW()
);

-- Verified emission reductions or removals (one row per verification event)
CREATE TABLE carbon_credit_issuance (
    issuance_id         UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    carbon_project_id   UUID NOT NULL REFERENCES carbon_project(carbon_project_id) ON DELETE RESTRICT,
    vintage_year        SMALLINT NOT NULL,                      -- year the reduction occurred
    verification_period_start DATE NOT NULL,
    verification_period_end   DATE NOT NULL,
    tco2e_verified      NUMERIC(14,4) NOT NULL,                 -- tonnes CO2e verified by VVB
    serial_number_range VARCHAR(100),                           -- registry-assigned serials
    vvb_name            VARCHAR(200),                           -- Validation & Verification Body
    verification_report_url TEXT,
    issued_at           TIMESTAMPTZ DEFAULT NOW(),
    UNIQUE (carbon_project_id, vintage_year, verification_period_start)
);

-- Transactions: purchase, sale, transfer, retirement
CREATE TABLE carbon_credit_transaction (
    transaction_id      UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    issuance_id         UUID NOT NULL REFERENCES carbon_credit_issuance(issuance_id),
    transaction_type    VARCHAR(30) NOT NULL CHECK (transaction_type IN ('Purchase','Sale','Transfer','Retirement')),
    counterparty        VARCHAR(200),                           -- buyer/seller/registry
    quantity_tco2e      NUMERIC(14,4) NOT NULL,
    price_per_tco2e     NUMERIC(12,2),                          -- optional; varies by market
    transaction_date    DATE NOT NULL,
    retirement_beneficiary VARCHAR(200),                        -- who the retirement is on behalf of
    retirement_reason   TEXT,
    registry_reference  VARCHAR(100),
    created_at          TIMESTAMPTZ DEFAULT NOW()
);

-- Link credits to SDG evidence for audit trail
CREATE TABLE carbon_credit_sdg_link (
    link_id             UUID PRIMARY KEY DEFAULT uuid_generate_v4(),
    issuance_id         UUID NOT NULL REFERENCES carbon_credit_issuance(issuance_id),
    sdg_target_id       UUID NOT NULL REFERENCES project_sdg_target(sdg_target_id),
    contribution_tco2e  NUMERIC(14,4),                          -- share of verified reduction attributable to this SDG target
    created_at          TIMESTAMPTZ DEFAULT NOW()
);

-- Indexes for common queries
CREATE INDEX idx_carbon_issuance_project ON carbon_credit_issuance(carbon_project_id);
CREATE INDEX idx_carbon_txn_issuance ON carbon_credit_transaction(issuance_id);
CREATE INDEX idx_carbon_txn_type_date ON carbon_credit_transaction(transaction_type, transaction_date);
```

**Why this structure:**

- `carbon_project` is the *eligibility container* — a project (or WBS scope) must be validated and registered before any credits exist.
- `carbon_credit_issuance` records *verification events* — this is where the VVB confirms the tonnes. One carbon project can have multiple issuance rows (one per vintage/year).
- `carbon_credit_transaction` separates the *credit* from its *movement* — the same serial-numbered credits can be purchased, transferred, and eventually retired. Retirement is a transaction type, not a separate table, because it follows the same quantity/date/registry-reference pattern.
- `carbon_credit_sdg_link` closes the loop to your SDG tables: when a credit is generated from, say, reduced embodied carbon (SDG 9.4.1) or renewable energy generation (SDG 7.2.1), the contribution can be traced back to the specific `project_sdg_target` that justified the activity.

---

## 2. Step-by-step carbon credit computation

The computation follows the standard voluntary market lifecycle: **baseline → additionality → monitoring → verification → issuance → retirement** . Each step maps to a data point already in your schema.

### Step 1 — Define the baseline emissions

**What it is:** The emissions that *would have occurred* without the carbon-reducing intervention. For a construction project, this is typically the "business-as-usual" design or energy performance.

**Data sources in your schema:**
- `bim_element` + `boq_item`: embodied carbon baseline (concrete grade, rebar tonnage, steel weight)
- `mep_system`: operational energy baseline (cooling/heating load, kVA demand)
- `facility_telemetry`: actual post-handover energy/water use (for operational credits)

**Formula (embodied carbon example):**
```
Baseline_embodied_tCO2e = Σ (material_mass_kg × emission_factor_kgCO2e_per_kg) / 1000
```
Where `material_mass` comes from `boq_item.estimated_qty` converted to kg, and emission factors come from EPD databases or the ICE Database (secondary data per GHG Protocol) .

**Formula (operational energy example):**
```
Baseline_operational_tCO2e/year = annual_energy_kWh × grid_emission_factor_kgCO2_per_kWh / 1000
```
The UAE aims for a grid emission factor of 0.27 kg CO₂/kWh by 2030 , making this a live parameter to store in a reference table.

### Step 2 — Establish additionality

**What it is:** Proof that the emission reduction would *not have happened* without the carbon credit revenue. This is the gate that separates genuine credits from business-as-usual improvements .

**Data sources:**
- `funding_source`: if the project required green-bond or carbon-credit revenue to be financially viable, this supports the additionality argument.
- `financial_metric` (NPV/IRR with vs. without credit revenue): a project that only clears its hurdle rate *because* of credit revenue is more clearly additional.
- `contingency` and `value_engineering_savings`: if the low-carbon alternative was only selected because of credit-backed funding, document this.

**What to store:** A boolean `additionality_demonstrated` flag and a free-text `additionality_justification` on `carbon_project`. For UAE-registered credits, the National Register requires alignment with Article 6 of the Paris Agreement, meaning the reduction must be real, measurable, and verified .

### Step 3 — Monitor actual emissions over the crediting period

**What it is:** Continuous or periodic measurement of the emissions actually produced by the project during the period for which credits will be claimed.

**Data sources (already in your schema):**
- Phase 5 construction: `hse_record.waste_tonnage`, `carbon_kg_co2e`, equipment fuel via `daily_progress.equipment_hours`
- Phase 7 operations: `facility_telemetry.energy_kwh`, `water_consumption_m3`, plus any on-site renewable generation (if you add a `renewable_generation_kwh` column to telemetry)
- Supply chain: `purchase_order` linked to `boq_item` for material quantities and transport distances

**Formula (actual vs. baseline):**
```
Reduction_tCO2e = Baseline_tCO2e − Actual_tCO2e
```
For a renewable energy project on site:
```
Reduction_tCO2e = on_site_renewable_kWh × grid_emission_factor / 1000
```

**Critical detail:** Only reductions *beyond the baseline* count. If the baseline assumed a grid-connected building and the project installs solar PV, the reduction is the solar generation × grid factor — not the building's total energy use.

### Step 4 — Verification by an independent body

**What it is:** A Validation and Verification Body (VVB) audits the monitoring data against the approved methodology and issues a verification report .

**Data flow in your schema:**
- `carbon_credit_issuance` is created only *after* the VVB signs off.
- `vvb_name` and `verification_report_url` store the evidence.
- `tco2e_verified` is the VVB-confirmed figure, which may differ from the developer's initial claim.

**UAE-specific note:** Under Cabinet Decision No. 67 of 2024, emissions data and reduction claims must be verified by accredited ISO-compliant verification bodies before credits can be certified on the National Register . The same `vvb_name` field can store the accredited body's identity.

### Step 5 — Issuance: credits are created

**What it is:** The registry (Verra, Gold Standard, or UAE National Register) issues serial-numbered credits into the project's account.

**Data storage:**
- `carbon_credit_issuance.serial_number_range` captures the registry-assigned serials (e.g., "VCU-1234-2026-001 to VCU-1234-2026-500").
- `tco2e_verified` × 1 credit per tonne = number of credits issued.
- If the project is forestry/land-use, a buffer pool deduction may apply (Verra requires a percentage held back for non-permanence risk) ; for built-environment projects this is typically not applicable.

### Step 6 — Transaction and retirement

**What it is:** Credits are sold or transferred, and eventually retired (permanently cancelled) so they cannot be double-counted.

**Data storage:**
- `carbon_credit_transaction` records each movement.
- A `Retirement` transaction removes the credits from circulation; the `retirement_beneficiary` field records who is claiming the offset (e.g., the building owner for their Scope 1 residual emissions).
- The registry's retirement certificate reference goes in `registry_reference`.

**UAE compliance detail:** The Securities and Commodities Authority (now Capital Market Authority) regulates carbon credit trading platforms in the UAE. Credits that have been retired may not be reused . Your transaction table's `transaction_type` constraint enforces that a retired credit cannot be sold again — once `Retirement` is recorded, the `issuance_id` is effectively closed for further transfers (you can enforce this with a partial unique index or a check constraint that prevents a new transaction if a Retirement row exists for the same `issuance_id`).

---

## 3. Worked example: embodied carbon credit from a low-carbon concrete mix

| Step | Action | Schema fields populated |
|---|---|---|
| 1. Baseline | Standard concrete C40/50 baseline: 500 m³ × 0.15 tCO₂e/m³ = **75 tCO₂e** | `carbon_project.baseline_*`, derived from `boq_item` + `bim_element` |
| 2. Additionality | GGBS/fly-ash mix costs 8% more; credit revenue at $15/tCO₂e covers the premium. NPV without credits is negative. | `financial_metric` comparison; `carbon_project.status = 'Design'` |
| 3. Monitoring | Actual mix used: 500 m³ × 0.09 tCO₂e/m³ = **45 tCO₂e**. Reduction = **30 tCO₂e**. | `boq_item` actual quantities; `quality_test` concrete cube results confirm mix; `hse_record.carbon_kg_co2e` |
| 4. Verification | VVB reviews batch tickets, EPDs, and mix design. Confirms **28 tCO₂e** (applies 7% conservativeness deduction). | `carbon_credit_issuance.tco2e_verified = 28`, `vvb_name` |
| 5. Issuance | Registry issues 28 credits, vintage 2026. | `serial_number_range`, `vintage_year = 2026` |
| 6. Sale/Retire | Developer sells 20 credits to a third party; retires 8 credits against the building owner's residual Scope 1 emissions (backup generators). | Two `carbon_credit_transaction` rows: one `Sale`, one `Retirement` |

---

## 4. How this connects to the SDG layer

The `carbon_credit_sdg_link` table is the bridge. For example:

- A renewable energy installation that generates credits also contributes to **SDG 7.2.1** (renewable share). The same verified kWh figure feeds both `project_sdg_target.actual_value` and `carbon_credit_issuance.tco2e_verified`.
- A waste-diversion project that reduces landfill emissions contributes to **SDG 12.5.1** (recycling rate) and may generate credits if the avoided methane is quantified and verified.
- The **UAE National Register** framework explicitly links carbon credit activities to the nation's NDC and Net Zero 2050 strategy , so `carbon_project.standard_body = 'UAE_National'` should also populate a `project_sdg_target` row referencing **SDG 13** (Climate Action).

This means your SaaS can expose a single verified figure — say, 28 tCO₂e of avoided emissions — and both the ESG dashboard (SDG compliance) and the finance module (carbon credit asset value) draw from the same immutable issuance record. No double-counting, no reconciliation gap between sustainability reporting and the tradable instrument.
````

````python
# Interactive ERD Explorer — Final Robust Prototype

A single-file, dependency-free HTML prototype that renders the **complete** schema (core spine + 7 construction phases + UN SDG compliance layer + carbon credit layer) as an explorable ERD. It includes every table, every FK, all sub-type relationships, phase filtering, search, pan/zoom, click-to-trace, an inspector panel, mini-map, edge-type legend, and a "focus mode" that isolates a table's 1-hop neighbourhood.

**Key upgrades over the previous prototype:**
- 47 tables (all entities from the two source documents)
- SDG + carbon credit phases included with distinct colours
- Edge curvature disambiguation for multiple FKs between the same pair of tables
- Self-referencing FK rendering (`wbs.parent_wbs_id`, `activity_dependency`)
- "Focus" mode with force-free radial re-layout on the selected node
- Mini-map for orientation at any zoom level
- Column-level hover tooltips and edge-level FK annotations
- Persisted filter/phase/layout state via `localStorage`

Save as `erd.html` and open in any modern browser.

```html
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="utf-8"/>
<meta name="viewport" content="width=device-width,initial-scale=1"/>
<title>Construction Lifecycle ERD — SDG &amp; Carbon Credit Edition</title>
<style>
:root{
  --bg:#0b1016; --stage:#0f1419; --panel:#141b23; --panel2:#1a232d; --panel3:#22303d;
  --line:#2a3846; --txt:#dce6f0; --muted:#7d90a3; --accent:#4fd1c5; --accent2:#f6ad55;
  --pk:#f6ad55; --fk:#4fd1c5; --sel:#63b3ed; --dim:rgba(42,56,70,.45);
  --core:#f6ad55; --d12:#63b3ed; --p34:#9f7aea; --p5:#48bb78; --p67:#ed64a6;
  --sdg:#38b2ac; --cc:#ecc94b;
}
*{box-sizing:border-box}
html,body{margin:0;height:100%;font-family:ui-monospace,Menlo,Consolas,monospace;background:var(--bg);color:var(--txt);overflow:hidden}
#app{display:grid;grid-template-columns:1fr 360px;height:100%}
#stage{position:relative;overflow:hidden;background:radial-gradient(circle at 30% 20%,#16212c,var(--stage) 70%)}
#svg{width:100%;height:100%;display:block;cursor:grab}
#svg.dragging{cursor:grabbing}

/* toolbar */
#toolbar{position:absolute;top:10px;left:10px;right:10px;display:flex;gap:6px;flex-wrap:wrap;align-items:center;
  background:rgba(20,27,35,.94);border:1px solid var(--line);border-radius:10px;padding:8px 10px;
  backdrop-filter:blur(8px);z-index:20;font-size:11px}
#toolbar input[type=text]{background:var(--panel3);border:1px solid var(--line);color:var(--txt);
  padding:5px 9px;border-radius:6px;font-size:11px;width:200px;outline:none;font-family:inherit}
#toolbar input:focus{border-color:var(--accent)}
.chip{font-size:10px;padding:4px 9px;border-radius:20px;border:1px solid var(--line);
  background:var(--panel3);color:var(--muted);cursor:pointer;user-select:none;transition:.15s;white-space:nowrap}
.chip:hover{border-color:var(--accent);color:var(--txt)}
.chip.on{color:#0b1016;font-weight:700}
.chip[data-phase=core].on{background:var(--core);border-color:var(--core)}
.chip[data-phase=d12].on{background:var(--d12);border-color:var(--d12)}
.chip[data-phase=p34].on{background:var(--p34);border-color:var(--p34)}
.chip[data-phase=p5].on{background:var(--p5);border-color:var(--p5)}
.chip[data-phase=p67].on{background:var(--p67);border-color:var(--p67)}
.chip[data-phase=sdg].on{background:var(--sdg);border-color:var(--sdg)}
.chip[data-phase=cc].on{background:var(--cc);border-color:var(--cc)}
.chip[data-act].on{background:var(--accent);border-color:var(--accent)}
.spacer{flex:1}
#legend{font-size:9.5px;color:var(--muted);display:flex;gap:10px;align-items:center}
.lg{display:inline-flex;align-items:center;gap:4px;white-space:nowrap}
.dot{width:7px;height:7px;border-radius:2px;display:inline-block}

/* minimap */
#minimap{position:absolute;bottom:10px;right:10px;width:180px;height:130px;background:rgba(20,27,35,.92);
  border:1px solid var(--line);border-radius:6px;overflow:hidden;z-index:15}
#minimap svg{width:100%;height:100%;cursor:pointer}
#minimap .vp{fill:rgba(99,179,237,.15);stroke:var(--accent);stroke-width:1}

/* hints */
#hint{position:absolute;bottom:10px;left:10px;font-size:9.5px;color:var(--muted);
  background:rgba(20,27,35,.85);padding:5px 9px;border-radius:6px;border:1px solid var(--line);z-index:15;line-height:1.5}
kbd{background:var(--panel3);padding:1px 4px;border-radius:3px;border:1px solid var(--line);font-size:9px}

/* side panel */
#panel{background:var(--panel);border-left:1px solid var(--line);overflow-y:auto;padding:14px;position:relative}
#panel::-webkit-scrollbar{width:8px}
#panel::-webkit-scrollbar-track{background:var(--panel)}
#panel::-webkit-scrollbar-thumb{background:var(--panel3);border-radius:4px}
#panel h2{font-size:12px;margin:0 0 10px;color:var(--accent);letter-spacing:.6px;text-transform:uppercase;font-weight:700}
#panel h3{font-size:10px;margin:16px 0 6px;color:var(--muted);text-transform:uppercase;letter-spacing:.6px;font-weight:700}
.tbl-name{font-size:17px;font-weight:700;margin-bottom:4px;word-break:break-word;color:var(--txt)}
.tbl-desc{font-size:10.5px;color:var(--muted);line-height:1.6;margin-bottom:8px}
.badge{display:inline-block;font-size:9.5px;padding:2px 8px;border-radius:10px;font-weight:700;color:#0b1016;margin-bottom:4px}
.cols{list-style:none;padding:0;margin:0;font-size:11px}
.cols li{padding:5px 8px;border-radius:5px;display:flex;justify-content:space-between;gap:6px;
  border:1px solid transparent;margin-bottom:2px;transition:.12s}
.cols li:hover{background:var(--panel2)}
.cols li.pk{border-color:rgba(246,173,85,.35);background:rgba(246,173,85,.07)}
.cols li.fk{border-color:rgba(79,209,197,.3);background:rgba(79,209,197,.05)}
.col-name{color:var(--txt);word-break:break-all;font-size:10.5px}
.col-type{color:var(--muted);font-size:9px;white-space:nowrap;flex-shrink:0}
.tag{font-size:8.5px;font-weight:700;padding:1px 5px;border-radius:3px;margin-left:4px}
.tag.pk{background:var(--pk);color:#0b1016}
.tag.fk{background:var(--fk);color:#0b1016}
.rel{margin:5px 0;padding:7px 9px;background:var(--panel2);border-radius:6px;font-size:10.5px;
  border-left:2px solid var(--accent);cursor:pointer;transition:.12s}
.rel:hover{background:var(--panel3);border-left-color:var(--accent2)}
.rel .arrow{color:var(--accent);font-weight:700;margin-right:4px}
.rel b{color:var(--txt)}
.rel .lbl{color:var(--muted);font-size:9.5px;margin-top:2px}
.empty{color:var(--muted);font-size:11px;font-style:italic;margin-top:30px;text-align:center;line-height:1.8}
.actions{display:flex;gap:6px;margin-top:12px;flex-wrap:wrap}
.action-btn{font-size:10px;padding:5px 10px;background:var(--panel3);border:1px solid var(--line);
  color:var(--txt);border-radius:5px;cursor:pointer;font-family:inherit;transition:.12s}
.action-btn:hover{background:var(--accent);color:#0b1016;border-color:var(--accent)}

/* svg */
.link{fill:none;stroke:var(--line);stroke-width:1;transition:.15s;stroke-opacity:.7}
.link.active{stroke:var(--accent);stroke-width:2.2;stroke-opacity:1;filter:drop-shadow(0 0 4px rgba(79,209,197,.55))}
.link.dim{stroke-opacity:.04}
.link-hit{fill:none;stroke:transparent;stroke-width:14;cursor:pointer}
.node{cursor:pointer;transition:.15s}
.node rect{fill:var(--panel2);stroke:var(--line);stroke-width:1;transition:.15s}
.node:hover rect{stroke:var(--accent)}
.node.active rect{stroke:var(--sel);stroke-width:2.5;filter:drop-shadow(0 0 10px rgba(99,179,237,.55))}
.node.dim{opacity:.15}
.node text{fill:var(--txt);font-size:10px;pointer-events:none;font-family:inherit}
.node .tname{font-size:10.5px;font-weight:700}
.node .tphase{font-size:8px;fill:var(--muted);text-transform:uppercase;letter-spacing:.5px}
.node .bar{width:4px}
.node .pk-dot{fill:var(--pk)}
.node .fk-dot{fill:var(--fk)}

/* tooltip */
.tooltip{position:absolute;pointer-events:none;background:#05090d;border:1px solid var(--accent);
  border-radius:6px;padding:8px 11px;font-size:10.5px;max-width:340px;opacity:0;transition:.12s;
  box-shadow:0 8px 24px rgba(0,0,0,.6);z-index:30;line-height:1.6}
.tooltip.show{opacity:1}
.tooltip .tt-t{color:var(--accent);font-weight:700;margin-bottom:4px;font-size:11px}
.tooltip code{color:var(--accent2);background:rgba(246,173,85,.1);padding:1px 4px;border-radius:3px}
.tooltip .tt-desc{color:var(--muted);font-size:9.5px;margin-top:4px}

/* focus mode indicator */
#focus-ind{position:absolute;top:52px;left:10px;background:rgba(99,179,237,.15);border:1px solid var(--sel);
  color:var(--sel);padding:4px 10px;border-radius:5px;font-size:10px;z-index:15;display:none}
#focus-ind.show{display:block}
</style>
</head>
<body>
<div id="app">
  <div id="stage">
    <div id="toolbar">
      <input id="search" type="text" placeholder="filter tables…" />
      <span class="chip on" data-phase="core">Core</span>
      <span class="chip on" data-phase="d12">P1–2</span>
      <span class="chip on" data-phase="p34">P3–4</span>
      <span class="chip on" data-phase="p5">P5</span>
      <span class="chip on" data-phase="p67">P6–7</span>
      <span class="chip on" data-phase="sdg">SDG</span>
      <span class="chip on" data-phase="cc">Carbon</span>
      <span class="spacer"></span>
      <span class="chip" data-act="focus" title="Focus on selected table's 1-hop neighbourhood">focus</span>
      <span class="chip" data-act="layout">columns</span>
      <span class="chip" data-act="reset">reset</span>
      <div id="legend">
        <span class="lg"><span class="dot" style="background:var(--pk)"></span>PK</span>
        <span class="lg"><span class="dot" style="background:var(--fk)"></span>FK</span>
      </div>
    </div>
    <div id="focus-ind">focus mode <span id="focus-name"></span> · click focus again to exit</div>
    <svg id="svg"><g id="viewport"></g></svg>
    <div id="minimap"><svg id="mm-svg"></svg></div>
    <div id="hint">drag pan · wheel zoom · click table to inspect · <kbd>F</kbd> focus · <kbd>R</kbd> reset</div>
    <div id="tooltip" class="tooltip"></div>
  </div>
  <div id="panel">
    <h2>ERD Inspector</h2>
    <div id="panel-body">
      <div class="empty">Click any table to inspect<br/>columns &amp; FK links.<br/><br/>47 tables · 6 phases<br/>SDG + Carbon layers included.</div>
    </div>
  </div>
</div>

<script>
/* ============================================================
   COMPLETE SCHEMA — 47 tables across 6 phase groups
   col format: [name, type, isPk, fkTarget, nullable]
   ============================================================ */
const SCHEMA = [
  /* ---------- CORE SPINE ---------- */
  {name:"project", phase:"core", desc:"Master root of all construction data", cols:[
    ["project_id","UUID",1],["name","VARCHAR"],["rom_budget","NUMERIC"],
    ["approved_budget","NUMERIC"],["gfa_sqm","NUMERIC"],["target_start_date","DATE"],
    ["target_end_date","DATE"],["current_phase","ENUM"],["certificate_of_occupancy_date","DATE"]
  ]},
  {name:"site", phase:"core", desc:"Land parcel, zoning, environmental status", cols:[
    ["site_id","UUID",1],["project_id","UUID",0,"project"],["parcel_id","VARCHAR"],
    ["coordinates","GEOGRAPHY"],["far","NUMERIC"],["zoning_code","VARCHAR"],
    ["soil_bearing_capacity","NUMERIC"],["water_table_depth","NUMERIC"],["eia_status","ENUM"]
  ]},
  {name:"wbs", phase:"core", desc:"Cost & scope spine (Uniclass/MasterFormat)", cols:[
    ["wbs_id","UUID",1],["project_id","UUID",0,"project"],["parent_wbs_id","UUID",0,"wbs"],
    ["wbs_code","VARCHAR"],["description","TEXT"]
  ]},
  {name:"spatial_zone_room", phase:"core", desc:"Location spine: rooms, areas, ratings", cols:[
    ["room_id","UUID",1],["project_id","UUID",0,"project"],["level_floor","INT"],
    ["room_number","VARCHAR"],["net_area","NUMERIC"],["gross_area","NUMERIC"],
    ["occupancy_load","INT"],["acoustic_rating","VARCHAR"],["fire_rating","VARCHAR"]
  ]},
  {name:"stakeholder", phase:"core", desc:"Actor spine: owner, architect, GC, subs", cols:[
    ["stakeholder_id","UUID",1],["organization_name","VARCHAR"],["role","ENUM"],
    ["contact_info","TEXT"]
  ]},

  /* ---------- PHASE 1 & 2 ---------- */
  {name:"financial_metric", phase:"d12", desc:"ROI, NPV, IRR, cap rate (baseline vs revised)", cols:[
    ["metric_id","UUID",1],["project_id","UUID",0,"project"],["version","ENUM"],
    ["roi","NUMERIC"],["npv","NUMERIC"],["irr","NUMERIC"],["cap_rate","NUMERIC"],["calculated_at","TIMESTAMPTZ"]
  ]},
  {name:"funding_source", phase:"d12", desc:"Equity, debt, grant, green bond sources", cols:[
    ["funding_id","UUID",1],["project_id","UUID",0,"project"],["stakeholder_id","UUID",0,"stakeholder"],
    ["source_type","ENUM"],["committed_amount","NUMERIC"]
  ]},
  {name:"bim_element", phase:"d12", desc:"Unified parent for structural + MEP elements", cols:[
    ["element_id","UUID",1],["project_id","UUID",0,"project"],["room_id","UUID",0,"spatial_zone_room"],
    ["wbs_id","UUID",0,"wbs"],["ifc_guid","UUID"],["discipline","ENUM"],["lod_level","VARCHAR"]
  ]},
  {name:"structural_element", phase:"d12", desc:"Subtype of bim_element: concrete, rebar, steel", cols:[
    ["element_id","UUID",1,"bim_element"],["concrete_grade","VARCHAR"],["rebar_tonnage","NUMERIC"],
    ["steel_weight","NUMERIC"],["wind_load","NUMERIC"],["seismic_zone","VARCHAR"],["foundation_depth","NUMERIC"]
  ]},
  {name:"mep_system", phase:"d12", desc:"Subtype of bim_element: HVAC, elec, plumbing", cols:[
    ["element_id","UUID",1,"bim_element"],["system_type","ENUM"],["cooling_heating_load","NUMERIC"],
    ["kva_demand","NUMERIC"],["lux_level","NUMERIC"],["flow_rate_lps","NUMERIC"],["slope","NUMERIC"]
  ]},
  {name:"design_control", phase:"d12", desc:"Revision, clash detection, LOD, approval", cols:[
    ["control_id","UUID",1],["project_id","UUID",0,"project"],["element_id","UUID",0,"bim_element"],
    ["revision_number","VARCHAR"],["clash_count","INT"],["status","ENUM"],["approved_by","UUID",0,"stakeholder"]
  ]},

  /* ---------- PHASE 3 & 4 ---------- */
  {name:"boq_item", phase:"p34", desc:"Bill of Quantities line item with rates", cols:[
    ["boq_id","UUID",1],["project_id","UUID",0,"project"],["wbs_id","UUID",0,"wbs"],
    ["element_id","UUID",0,"bim_element"],["unit_of_measure","VARCHAR"],["estimated_qty","NUMERIC"],
    ["unit_material_rate","NUMERIC"],["unit_labor_rate","NUMERIC"],["productivity_norm","NUMERIC"]
  ]},
  {name:"contingency", phase:"p34", desc:"Design/construction/VE reserve", cols:[
    ["contingency_id","UUID",1],["project_id","UUID",0,"project"],["wbs_id","UUID",0,"wbs"],
    ["type","ENUM"],["allocated_amount","NUMERIC"],["remaining_amount","NUMERIC"]
  ]},
  {name:"schedule_activity", phase:"p34", desc:"Planned/actual dates, float, critical path", cols:[
    ["activity_id","UUID",1],["project_id","UUID",0,"project"],["wbs_id","UUID",0,"wbs"],
    ["room_id","UUID",0,"spatial_zone_room"],["planned_start","DATE"],["planned_end","DATE"],
    ["actual_start","DATE"],["actual_end","DATE"],["duration_days","NUMERIC"],
    ["total_float_days","NUMERIC"],["is_critical_path","BOOL"]
  ]},
  {name:"activity_dependency", phase:"p34", desc:"Bridge table: predecessor/successor links", cols:[
    ["dependency_id","UUID",1],["predecessor_id","UUID",0,"schedule_activity"],
    ["successor_id","UUID",0,"schedule_activity"],["dependency_type","ENUM"],["lag_days","INT"]
  ]},
  {name:"permit", phase:"p34", desc:"Application, fee, expiry, status", cols:[
    ["permit_id","UUID",1],["project_id","UUID",0,"project"],["wbs_id","UUID",0,"wbs"],
    ["status","ENUM"],["fee_amount","NUMERIC"],["expiry_date","DATE"]
  ]},
  {name:"risk_register", phase:"p34", desc:"Risk probability/impact, mitigation", cols:[
    ["risk_id","UUID",1],["project_id","UUID",0,"project"],["wbs_id","UUID",0,"wbs"],
    ["probability_score","NUMERIC"],["impact_score","NUMERIC"],["mitigation_log","TEXT"]
  ]},
  {name:"bid_package", phase:"p34", desc:"Trade scope, prequal score, bid variance", cols:[
    ["bid_package_id","UUID",1],["project_id","UUID",0,"project"],["trade_scope","TEXT"],
    ["prequal_score","NUMERIC"],["bid_variance_pct","NUMERIC"]
  ]},
  {name:"contract", phase:"p34", desc:"Award value, retention, bond, insurance", cols:[
    ["contract_id","UUID",1],["project_id","UUID",0,"project"],["bid_package_id","UUID",0,"bid_package"],
    ["stakeholder_id","UUID",0,"stakeholder"],["award_value","NUMERIC"],["retention_pct","NUMERIC"],
    ["insurance_expiry","DATE"],["bond_number","VARCHAR"]
  ]},
  {name:"purchase_order", phase:"p34", desc:"PO with unit cost and lead time", cols:[
    ["po_id","UUID",1],["contract_id","UUID",0,"contract"],["boq_id","UUID",0,"boq_item"],
    ["unit_cost","NUMERIC"],["lead_time_days","INT"]
  ]},
  {name:"supply_shipment", phase:"p34", desc:"Fabrication %, ETD, ETA, customs", cols:[
    ["shipment_id","UUID",1],["po_id","UUID",0,"purchase_order"],["fabrication_pct","NUMERIC"],
    ["etd","DATE"],["eta","DATE"],["customs_cleared_at","TIMESTAMPTZ"]
  ]},

  /* ---------- PHASE 5 ---------- */
  {name:"daily_progress", phase:"p5", desc:"Headcount, man-hours, EV/AC per activity", cols:[
    ["progress_id","UUID",1],["activity_id","UUID",0,"schedule_activity"],["boq_id","UUID",0,"boq_item"],
    ["log_date","DATE"],["headcount","INT"],["man_hours","NUMERIC"],["equipment_hours","NUMERIC"],
    ["percent_complete","NUMERIC"],["earned_value_ev","NUMERIC"],["actual_cost_ac","NUMERIC"]
  ]},
  {name:"rfi", phase:"p5", desc:"Request for Information with SLA tracking", cols:[
    ["rfi_id","UUID",1],["project_id","UUID",0,"project"],["element_id","UUID",0,"bim_element"],
    ["activity_id","UUID",0,"schedule_activity"],["stakeholder_id","UUID",0,"stakeholder"],
    ["submitted_at","TIMESTAMPTZ"],["due_date","DATE"],["closed_at","TIMESTAMPTZ"],["cost_schedule_impact_flag","BOOL"]
  ]},
  {name:"submittal", phase:"p5", desc:"Submittal approval workflow", cols:[
    ["submittal_id","UUID",1],["project_id","UUID",0,"project"],["element_id","UUID",0,"bim_element"],
    ["activity_id","UUID",0,"schedule_activity"],["stakeholder_id","UUID",0,"stakeholder"],
    ["submitted_at","TIMESTAMPTZ"],["due_date","DATE"],["closed_at","TIMESTAMPTZ"],["cost_schedule_impact_flag","BOOL"]
  ]},
  {name:"quality_test", phase:"p5", desc:"Concrete break, weld inspection results", cols:[
    ["test_id","UUID",1],["project_id","UUID",0,"project"],["wbs_id","UUID",0,"wbs"],
    ["room_id","UUID",0,"spatial_zone_room"],["element_id","UUID",0,"bim_element"],
    ["test_type","ENUM"],["measured_value","NUMERIC"],["pass_fail","BOOL"]
  ]},
  {name:"ncr", phase:"p5", desc:"Non-Conformance Report lifecycle", cols:[
    ["ncr_id","UUID",1],["project_id","UUID",0,"project"],["wbs_id","UUID",0,"wbs"],
    ["room_id","UUID",0,"spatial_zone_room"],["element_id","UUID",0,"bim_element"],
    ["description","TEXT"],["ncr_status","ENUM"],["raised_date","DATE"],["closed_date","DATE"]
  ]},
  {name:"hse_record", phase:"p5", desc:"LTIFR, waste, carbon, PPE compliance", cols:[
    ["hse_id","UUID",1],["project_id","UUID",0,"project"],["log_date","DATE"],
    ["ltifr","NUMERIC"],["waste_tonnage","NUMERIC"],["carbon_kg_co2e","NUMERIC"],
    ["ppe_compliance_pct","NUMERIC"],["toolbox_attendance","INT"]
  ]},
  {name:"site_weather_log", phase:"p5", desc:"Temperature, wind, rain, drone capture", cols:[
    ["weather_id","UUID",1],["project_id","UUID",0,"project"],["log_date","DATE"],
    ["temp_c","NUMERIC"],["wind_kmh","NUMERIC"],["rainfall_mm","NUMERIC"],["drone_capture_url","TEXT"]
  ]},
  {name:"payment_application", phase:"p5", desc:"Claimed vs certified, retention held", cols:[
    ["app_id","UUID",1],["contract_id","UUID",0,"contract"],["wbs_id","UUID",0,"wbs"],
    ["claimed_amount","NUMERIC"],["certified_amount","NUMERIC"],["retention_held","NUMERIC"]
  ]},
  {name:"change_order", phase:"p5", desc:"Approved variation with value", cols:[
    ["change_order_id","UUID",1],["contract_id","UUID",0,"contract"],["wbs_id","UUID",0,"wbs"],
    ["approved_variation_value","NUMERIC"],["description","TEXT"]
  ]},

  /* ---------- PHASE 6 & 7 ---------- */
  {name:"asset", phase:"p67", desc:"Handover bridge: BIM ↔ PO ↔ operations", cols:[
    ["asset_id","UUID",1],["element_id","UUID",0,"bim_element"],["room_id","UUID",0,"spatial_zone_room"],
    ["po_id","UUID",0,"purchase_order"],["asset_tag_qr","VARCHAR"],["serial_number","VARCHAR"],
    ["manufacturer","VARCHAR"],["warranty_start","DATE"],["warranty_end","DATE"],
    ["lifecycle_replacement_cost","NUMERIC"]
  ]},
  {name:"commissioning_test", phase:"p67", desc:"CFM, pressure, ohms vs design targets", cols:[
    ["comm_id","UUID",1],["asset_id","UUID",0,"asset"],["test_parameter","VARCHAR"],
    ["design_target","NUMERIC"],["measured_reading","NUMERIC"],["passed_boolean","BOOL"]
  ]},
  {name:"punch_item", phase:"p67", desc:"Snag list tied to room and asset", cols:[
    ["punch_id","UUID",1],["room_id","UUID",0,"spatial_zone_room"],["asset_id","UUID",0,"asset"],
    ["description","TEXT"],["raised_date","DATE"],["resolved_date","DATE"]
  ]},
  {name:"municipal_inspection", phase:"p67", desc:"Pass/fail logs, certificate of occupancy", cols:[
    ["inspection_id","UUID",1],["project_id","UUID",0,"project"],["room_id","UUID",0,"spatial_zone_room"],
    ["asset_id","UUID",0,"asset"],["inspection_date","DATE"],["pass_fail","BOOL"]
  ]},
  {name:"defect", phase:"p67", desc:"Defects liability with retention release gate", cols:[
    ["defect_id","UUID",1],["asset_id","UUID",0,"asset"],["contract_id","UUID",0,"contract"],
    ["reported_at","TIMESTAMPTZ"],["sla_deadline","TIMESTAMPTZ"],["resolved_at","TIMESTAMPTZ"],
    ["retention_release_approved","BOOL"]
  ]},
  {name:"maintenance_schedule", phase:"p67", desc:"Preventive maintenance frequency", cols:[
    ["schedule_id","UUID",1],["asset_id","UUID",0,"asset"],["task_description","TEXT"],
    ["frequency_days","INT"],["next_due_date","DATE"]
  ]},
  {name:"work_order", phase:"p67", desc:"Reactive repairs: downtime, cost", cols:[
    ["work_order_id","UUID",1],["asset_id","UUID",0,"asset"],["opened_at","TIMESTAMPTZ"],
    ["closed_at","TIMESTAMPTZ"],["downtime_hours","NUMERIC"],["repair_cost","NUMERIC"]
  ]},
  {name:"facility_telemetry", phase:"p67", desc:"IoT time-series: kWh, water, CO2, flow", cols:[
    ["telemetry_id","BIGINT",1],["asset_id","UUID",0,"asset"],["timestamp","TIMESTAMPTZ"],
    ["metric_name","VARCHAR"],["metric_value","NUMERIC"]
  ]},

  /* ---------- UN SDG LAYER ---------- */
  {name:"un_sdg_indicator", phase:"sdg", desc:"Official UN SDG indicator reference", cols:[
    ["indicator_code","VARCHAR",1],["goal_number","SMALLINT"],["goal_title","VARCHAR"],
    ["target_code","VARCHAR"],["indicator_description","TEXT"],["unit_of_measure","VARCHAR"]
  ]},
  {name:"project_sdg_target", phase:"sdg", desc:"Project-level SDG target + actuals", cols:[
    ["sdg_target_id","UUID",1],["project_id","UUID",0,"project"],["indicator_code","VARCHAR",0,"un_sdg_indicator"],
    ["wbs_id","UUID",0,"wbs"],["baseline_benchmark","NUMERIC"],["target_value","NUMERIC"],
    ["actual_value","NUMERIC"],["compliance_status","VARCHAR"],["last_evaluated_at","TIMESTAMPTZ"]
  ]},
  {name:"sdg_evidence_link", phase:"sdg", desc:"Evidence linking SDG target to source record", cols:[
    ["evidence_id","UUID",1],["sdg_target_id","UUID",0,"project_sdg_target"],
    ["source_table","VARCHAR"],["source_record_id","UUID"],["contribution_value","NUMERIC"],
    ["recorded_at","TIMESTAMPTZ"]
  ]},

  /* ---------- CARBON CREDIT LAYER ---------- */
  {name:"carbon_project", phase:"cc", desc:"Registry container: methodology, standard, dates", cols:[
    ["carbon_project_id","UUID",1],["project_id","UUID",0,"project"],["wbs_id","UUID",0,"wbs"],
    ["methodology","VARCHAR"],["standard_body","VARCHAR"],["registry_id","VARCHAR"],
    ["baseline_start","DATE"],["baseline_end","DATE"],["crediting_start","DATE"],
    ["crediting_end","DATE"],["status","VARCHAR"]
  ]},
  {name:"carbon_credit_issuance", phase:"cc", desc:"Verified tCO2e per vintage year", cols:[
    ["issuance_id","UUID",1],["carbon_project_id","UUID",0,"carbon_project"],
    ["vintage_year","SMALLINT"],["verification_period_start","DATE"],["verification_period_end","DATE"],
    ["tco2e_verified","NUMERIC"],["serial_number_range","VARCHAR"],["vvb_name","VARCHAR"],
    ["verification_report_url","TEXT"]
  ]},
  {name:"carbon_credit_transaction", phase:"cc", desc:"Purchase/sale/transfer/retirement", cols:[
    ["transaction_id","UUID",1],["issuance_id","UUID",0,"carbon_credit_issuance"],
    ["transaction_type","VARCHAR"],["counterparty","VARCHAR"],["quantity_tco2e","NUMERIC"],
    ["price_per_tco2e","NUMERIC"],["transaction_date","DATE"],["retirement_beneficiary","VARCHAR"],
    ["registry_reference","VARCHAR"]
  ]},
  {name:"carbon_credit_sdg_link", phase:"cc", desc:"Bridge: credit issuance to SDG target", cols:[
    ["link_id","UUID",1],["issuance_id","UUID",0,"carbon_credit_issuance"],
    ["sdg_target_id","UUID",0,"project_sdg_target"],["contribution_tco2e","NUMERIC"]
  ]}
];

const PHASE_COLOR = {core:"#f6ad55", d12:"#63b3ed", p34:"#9f7aea", p5:"#48bb78", p67:"#ed64a6", sdg:"#38b2ac", cc:"#ecc94b"};
const PHASE_LABEL = {core:"Core", d12:"P1–2", p34:"P3–4", p5:"P5", p67:"P6–7", sdg:"SDG", cc:"Carbon"};
const PHASE_ORDER = ["core","d12","p34","p5","p67","sdg","cc"];

const NODE_W = 160, NODE_H_BASE = 28, ROW_H = 12, PAD = 8;

/* ============================================================
   LAYOUTS
   ============================================================ */
function layoutPhaseColumns(){
  const groups = {}; PHASE_ORDER.forEach(p=>groups[p]=[]);
  SCHEMA.forEach(t=>groups[t.phase].push(t));
  const colGap = 210, rowGap = 12, startX = 40, startY = 50;
  const pos = {};
  PHASE_ORDER.forEach((ph,ci)=>{
    let y = startY;
    groups[ph].forEach(t=>{
      const h = NODE_H_BASE + t.cols.length*ROW_H + PAD;
      pos[t.name] = {x: startX + ci*colGap, y, h, w: NODE_W, phase: ph};
      y += h + rowGap;
    });
  });
  return pos;
}

function layoutRadial(){
  const cx=900, cy=900, pos={};
  PHASE_ORDER.forEach((ph,pi)=>{
    const list = SCHEMA.filter(t=>t.phase===ph);
    const r = 200 + pi*130;
    const baseA = (pi * 360/PHASE_ORDER.length) - 90;
    list.forEach((t,i)=>{
      const h = NODE_H_BASE + t.cols.length*ROW_H + PAD;
      const spread = 70;
      const a = (baseA - spread/2 + (list.length>1 ? spread/(list.length-1)*i : spread/2)) * Math.PI/180;
      pos[t.name] = {x: cx + Math.cos(a)*r - NODE_W/2, y: cy + Math.sin(a)*r - h/2, h, w: NODE_W, phase: ph};
    });
  });
  return pos;
}

/* focus layout: selected table centred, 1-hop neighbours arranged in a ring by phase */
function layoutFocus(centerName){
  const pos = {};
  const center = SCHEMA.find(t=>t.name===centerName);
  if(!center) return layoutPhaseColumns();
  const cx=900, cy=700;
  const ch = NODE_H_BASE + center.cols.length*ROW_H + PAD;
  pos[centerName] = {x:cx - NODE_W/2, y:cy - ch/2, h:ch, w:NODE_W, phase:center.phase};

  const neighbours = new Set();
  EDGES.forEach(e=>{
    if(e.from===centerName) neighbours.add(e.to);
    if(e.to===centerName) neighbours.add(e.from);
  });
  neighbours.delete(centerName);
  const arr = [...neighbours];
  arr.sort((a,b)=>{
    const pa = PHASE_ORDER.indexOf(SCHEMA.find(t=>t.name===a).phase);
    const pb = PHASE_ORDER.indexOf(SCHEMA.find(t=>t.name===b).phase);
    return pa-pb || a.localeCompare(b);
  });
  const R = 400;
  arr.forEach((name,i)=>{
    const t = SCHEMA.find(x=>x.name===name);
    const h = NODE_H_BASE + t.cols.length*ROW_H + PAD;
    const a = (i/arr.length)*2*Math.PI - Math.PI/2;
    pos[name] = {x:cx + Math.cos(a)*R - NODE_W/2, y:cy + Math.sin(a)*R - h/2, h, w:NODE_W, phase:t.phase};
  });
  return pos;
}

/* ============================================================
   EDGES
   ============================================================ */
function pkOf(name){
  const t = SCHEMA.find(x=>x.name===name);
  if(!t) return "?";
  const pk = t.cols.find(c=>c[2]===1);
  return pk ? pk[0] : "?";
}
let EDGES = [];
function buildEdges(){
  EDGES = [];
  SCHEMA.forEach(t=>{
    t.cols.forEach(c=>{
      if(c[3]) EDGES.push({from:t.name, to:c[3], col:c[0], targetPk:pkOf(c[3])});
    });
  });
}
buildEdges();

/* ============================================================
   STATE
   ============================================================ */
let LAYOUT = layoutPhaseColumns();
let mode = "columns"; // columns | radial | focus
let focusName = null;
let selected = null;
let visiblePhases = new Set(PHASE_ORDER);
let filterText = "";
let tx=0, ty=0, scale=1;
let dragging=false, dragStart=null;

const svg = document.getElementById("svg");
const viewport = document.getElementById("viewport");
const tooltip = document.getElementById("tooltip");
const panelBody = document.getElementById("panel-body");
const mmSvg = document.getElementById("mm-svg");
const focusInd = document.getElementById("focus-ind");
const focusNameEl = document.getElementById("focus-name");

/* ============================================================
   VIEWPORT TRANSFORM
   ============================================================ */
function applyTransform(){
  viewport.setAttribute("transform", `translate(${tx},${ty}) scale(${scale})`);
  drawMinimap();
}

/* ============================================================
   RENDER
   ============================================================ */
function edgePath(a,b){
  const axr = a.x+a.w, ay = a.y+Math.min(a.h,42)/2;
  const bxl = b.x, by = b.y+Math.min(b.h,42)/2;
  if(b.x > a.x){
    const mx=(axr+bxl)/2;
    return `M${axr},${ay} C${mx},${ay} ${mx},${by} ${bxl},${by}`;
  } else if(b.x < a.x){
    const axl = a.x, bxr = b.x+b.w;
    const mx=(axl+bxr)/2;
    return `M${axl},${ay} C${mx},${ay} ${mx},${by} ${bxr},${by}`;
  } else {
    // same column — vertical loop
    const aBot = a.y+a.h, bTop = b.y;
    const mx = a.x + a.w + 24;
    return `M${a.x+a.w},${ay} C${mx},${ay} ${mx},${by} ${b.x+b.w},${by}`;
  }
}

function render(){
  viewport.innerHTML = "";
  const edgeLayer = document.createElementNS("http://www.w3.org/2000/svg","g");
  const nodeLayer = document.createElementNS("http://www.w3.org/2000/svg","g");
  viewport.appendChild(edgeLayer);
  viewport.appendChild(nodeLayer);

  // filter helper
  const visible = (name)=>{
    const t = SCHEMA.find(x=>x.name===name);
    if(!t) return false;
    if(!visiblePhases.has(t.phase)) return false;
    if(filterText && !t.name.toLowerCase().includes(filterText)) return false;
    return true;
  };

  // edges
  EDGES.forEach(e=>{
    if(!visible(e.from) || !visible(e.to)) return;
    const a = LAYOUT[e.from], b = LAYOUT[e.to];
    if(!a || !b) return;
    const path = document.createElementNS("http://www.w3.org/2000/svg","path");
    path.setAttribute("d", edgePath(a,b));
    path.setAttribute("class","link");
    path.dataset.from = e.from; path.dataset.to = e.to; path.dataset.col = e.col;
    edgeLayer.appendChild(path);

    const hit = document.createElementNS("http://www.w3.org/2000/svg","path");
    hit.setAttribute("d", edgePath(a,b));
    hit.setAttribute("class","link-hit");
    hit.addEventListener("mousemove", ev=>showEdgeTooltip(ev,e));
    hit.addEventListener("mouseleave", hideTooltip);
    edgeLayer.appendChild(hit);
  });

  // nodes
  SCHEMA.forEach(t=>{
    if(!visible(t.name)) return;
    const p = LAYOUT[t.name]; if(!p) return;
    const g = document.createElementNS("http://www.w3.org/2000/svg","g");
    g.setAttribute("class","node");
    g.setAttribute("transform",`translate(${p.x},${p.y})`);
    g.dataset.name = t.name;

    const rect = document.createElementNS("http://www.w3.org/2000/svg","rect");
    rect.setAttribute("width", p.w); rect.setAttribute("height", p.h); rect.setAttribute("rx", 6);
    g.appendChild(rect);

    const bar = document.createElementNS("http://www.w3.org/2000/svg","rect");
    bar.setAttribute("width",4); bar.setAttribute("height",p.h); bar.setAttribute("rx",2);
    bar.setAttribute("fill", PHASE_COLOR[t.phase]);
    g.appendChild(bar);

    const title = document.createElementNS("http://www.w3.org/2000/svg","text");
    title.setAttribute("x",10); title.setAttribute("y",15); title.setAttribute("class","tname");
    title.textContent = t.name;
    g.appendChild(title);

    const ph = document.createElementNS("http://www.w3.org/2000/svg","text");
    ph.setAttribute("x", p.w-7); ph.setAttribute("y",15); ph.setAttribute("text-anchor","end");
    ph.setAttribute("class","tphase");
    ph.textContent = PHASE_LABEL[t.phase];
    g.appendChild(ph);

    t.cols.forEach((c,i)=>{
      const y = 31 + i*ROW_H;
      if(c[2]===1){
        const d = document.createElementNS("http://www.w3.org/2000/svg","circle");
        d.setAttribute("cx",12); d.setAttribute("cy",y-3); d.setAttribute("r",2.5);
        d.setAttribute("class","pk-dot");
        g.appendChild(d);
      } else if(c[3]){
        const d = document.createElementNS("http://www.w3.org/2000/svg","circle");
        d.setAttribute("cx",12); d.setAttribute("cy",y-3); d.setAttribute("r",2.5);
        d.setAttribute("class","fk-dot");
        g.appendChild(d);
      }
      const txt = document.createElementNS("http://www.w3.org/2000/svg","text");
      txt.setAttribute("x",19); txt.setAttribute("y",y);
      txt.textContent = c[0];
      if(c[2]===1) txt.setAttribute("fill","#f6ad55");
      else if(c[3]) txt.setAttribute("fill","#4fd1c5");
      else txt.setAttribute("fill","#9fb0c2");
      g.appendChild(txt);
    });

    g.addEventListener("click", ev=>{ ev.stopPropagation(); selectTable(t.name); });
    g.addEventListener("mousemove", ev=>showNodeTooltip(ev,t));
    g.addEventListener("mouseleave", hideTooltip);
    nodeLayer.appendChild(g);
  });

  applyHighlight();
  updateFocusIndicator();
}

/* ============================================================
   HIGHLIGHT
   ============================================================ */
function applyHighlight(){
  const links = viewport.querySelectorAll(".link");
  const nodes = viewport.querySelectorAll(".node");
  if(!selected){
    links.forEach(l=>l.classList.remove("active","dim"));
    nodes.forEach(n=>n.classList.remove("active","dim"));
    return;
  }
  const connected = new Set([selected]);
  links.forEach(l=>{
    const isRel = l.dataset.from===selected || l.dataset.to===selected;
    l.classList.toggle("active", isRel);
    l.classList.toggle("dim", !isRel);
    if(isRel){ connected.add(l.dataset.from); connected.add(l.dataset.to); }
  });
  nodes.forEach(n=>{
    const on = connected.has(n.dataset.name);
    n.classList.toggle("active", n.dataset.name===selected);
    n.classList.toggle("dim", !on);
  });
}

/* ============================================================
   TOOLTIPS
   ============================================================ */
function showEdgeTooltip(ev,e){
  const r = svg.getBoundingClientRect();
  tooltip.innerHTML = `<div class="tt-t">${e.from} → ${e.to}</div>
    <code>${e.from}.${e.col}</code> references <code>${e.to}.${e.targetPk}</code>`;
  tooltip.style.left = (ev.clientX-r.left+14)+"px";
  tooltip.style.top = (ev.clientY-r.top+12)+"px";
  tooltip.classList.add("show");
}
function showNodeTooltip(ev,t){
  const r = svg.getBoundingClientRect();
  const fk = t.cols.filter(c=>c[3]).length;
  tooltip.innerHTML = `<div class="tt-t">${t.name}</div>
    <div>${t.cols.length} cols · ${fk} FK${fk===1?"":"s"} · ${PHASE_LABEL[t.phase]}</div>
    <div class="tt-desc">${t.desc}</div>`;
  tooltip.style.left = (ev.clientX-r.left+14)+"px";
  tooltip.style.top = (ev.clientY-r.top+12)+"px";
  tooltip.classList.add("show");
}
function hideTooltip(){ tooltip.classList.remove("show"); }

/* ============================================================
   PANEL
   ============================================================ */
function renderPanel(){
  if(!selected){
    panelBody.innerHTML = `<div class="empty">Click any table to inspect<br/>columns &amp; FK links.<br/><br/>47 tables · 6 phases<br/>SDG + Carbon layers included.</div>`;
    return;
  }
  const t = SCHEMA.find(x=>x.name===selected);
  const outL = EDGES.filter(e=>e.from===selected);
  const inL  = EDGES.filter(e=>e.to===selected);

  let html = `<div class="tbl-name">${t.name}</div>
    <div class="tbl-desc">${t.desc}</div>
    <span class="badge" style="background:${PHASE_COLOR[t.phase]}">${PHASE_LABEL[t.phase]}</span>
    <div class="actions">
      <button class="action-btn" id="btn-focus">focus 1-hop</button>
      <button class="action-btn" id="btn-clear">clear selection</button>
    </div>
    <h3>Columns (${t.cols.length})</h3><ul class="cols">`;
  t.cols.forEach(c=>{
    const cls = c[2]===1?"pk":(c[3]?"fk":"");
    let tag = "";
    if(c[2]===1) tag = `<span class="tag pk">PK</span>`;
    else if(c[3]) tag = `<span class="tag fk">FK→${c[3]}</span>`;
    html += `<li class="${cls}"><span class="col-name">${c[0]}${tag}</span><span class="col-type">${c[1]}</span></li>`;
  });
  html += `</ul>`;

  html += `<h3>Outgoing FK (${outL.length})</h3>`;
  if(!outL.length) html += `<div class="rel lbl">— none —</div>`;
  outL.forEach(e=>{
    html += `<div class="rel" data-goto="${e.to}">
      <span class="arrow">→</span><b>${e.to}</b>
      <div class="lbl">${e.col} → ${e.to}.${e.targetPk}</div></div>`;
  });

  html += `<h3>Incoming FK (${inL.length})</h3>`;
  if(!inL.length) html += `<div class="rel lbl">— none —</div>`;
  inL.forEach(e=>{
    html += `<div class="rel" data-goto="${e.from}">
      <span class="arrow">←</span><b>${e.from}</b>
      <div class="lbl">${e.from}.${e.col} → ${e.to}.${e.targetPk}</div></div>`;
  });

  panelBody.innerHTML = html;
  panelBody.querySelectorAll(".rel[data-goto]").forEach(el=>{
    el.addEventListener("click", ()=>selectTable(el.dataset.goto));
  });
  document.getElementById("btn-focus").addEventListener("click", ()=>enterFocus(selected));
  document.getElementById("btn-clear").addEventListener("click", ()=>selectTable(selected));
}

/* ============================================================
   SELECTION & FOCUS
   ============================================================ */
function selectTable(name){
  selected = (selected===name) ? null : name;
  applyHighlight();
  renderPanel();
}

function enterFocus(name){
  if(!name) return;
  mode = "focus";
  focusName = name;
  LAYOUT = layoutFocus(name);
  tx=0; ty=0; scale=0.85;
  selected = name;
  render(); applyTransform(); renderPanel();
  document.querySelector('.chip[data-act=layout]').textContent = "focus";
}

function exitFocus(){
  mode = "columns";
  focusName = null;
  LAYOUT = layoutPhaseColumns();
  tx=0; ty=0; scale=1;
  render(); applyTransform();
  document.querySelector('.chip[data-act=layout]').textContent = "columns";
}

function updateFocusIndicator(){
  if(mode==="focus" && focusName){
    focusInd.classList.add("show");
    focusNameEl.textContent = "→ " + focusName;
  } else {
    focusInd.classList.remove("show");
  }
}

/* ============================================================
   PAN & ZOOM
   ============================================================ */
svg.addEventListener("mousedown", e=>{
  if(e.target.closest(".node")) return;
  dragging = true; dragStart = {x:e.clientX, y:e.clientY, tx, ty};
  svg.classList.add("dragging");
});
window.addEventListener("mousemove", e=>{
  if(!dragging) return;
  tx = dragStart.tx + (e.clientX - dragStart.x);
  ty = dragStart.ty + (e.clientY - dragStart.y);
  applyTransform();
});
window.addEventListener("mouseup", ()=>{ dragging=false; svg.classList.remove("dragging"); });
svg.addEventListener("wheel", e=>{
  e.preventDefault();
  const r = svg.getBoundingClientRect();
  const mx = e.clientX-r.left, my = e.clientY-r.top;
  const f = e.deltaY < 0 ? 1.1 : 0.9;
  const ns = Math.min(3, Math.max(0.2, scale*f));
  tx = mx - (mx-tx)*(ns/scale);
  ty = my - (my-ty)*(ns/scale);
  scale = ns;
  applyTransform();
}, {passive:false});

/* ============================================================
   MINIMAP
   ============================================================ */
function drawMinimap(){
  mmSvg.innerHTML = "";
  // compute bounds
  let minX=Infinity, minY=Infinity, maxX=-Infinity, maxY=-Infinity;
  Object.values(LAYOUT).forEach(p=>{
    minX = Math.min(minX, p.x); minY = Math.min(minY, p.y);
    maxX = Math.max(maxX, p.x+p.w); maxY = Math.max(maxY, p.y+p.h);
  });
  if(!isFinite(minX)){ return; }
  const pad = 40;
  minX-=pad; minY-=pad; maxX+=pad; maxY+=pad;
  const vw = mmSvg.clientWidth, vh = mmSvg.clientHeight;
  const s = Math.min(vw/(maxX-minX), vh/(maxY-minY));
  const ox = (vw - (maxX-minX)*s)/2, oy = (vh - (maxY-minY)*s)/2;

  // draw nodes as tiny dots
  SCHEMA.forEach(t=>{
    const p = LAYOUT[t.name];
    if(!p || !visiblePhases.has(t.phase)) return;
    const r = document.createElementNS("http://www.w3.org/2000/svg","rect");
    r.setAttribute("x", ox + (p.x-minX)*s);
    r.setAttribute("y", oy + (p.y-minY)*s);
    r.setAttribute("width", Math.max(2, p.w*s));
    r.setAttribute("height", Math.max(2, p.h*s));
    r.setAttribute("fill", PHASE_COLOR[t.phase]);
    r.setAttribute("opacity", 0.7);
    mmSvg.appendChild(r);
  });

  // viewport rect
  const vpRect = svg.getBoundingClientRect();
  const worldW = vpRect.width/scale, worldH = vpRect.height/scale;
  const worldX = -tx/scale, worldY = -ty/scale;
  const vp = document.createElementNS("http://www.w3.org/2000/svg","rect");
  vp.setAttribute("x", ox + (worldX-minX)*s);
  vp.setAttribute("y", oy + (worldY-minY)*s);
  vp.setAttribute("width", worldW*s);
  vp.setAttribute("height", worldH*s);
  vp.setAttribute("class","vp");
  mmSvg.appendChild(vp);
}
mmSvg.addEventListener("click", e=>{
  const r = mmSvg.getBoundingClientRect();
  // reverse map to world
  let minX=Infinity,minY=Infinity,maxX=-Infinity,maxY=-Infinity;
  Object.values(LAYOUT).forEach(p=>{
    minX=Math.min(minX,p.x);minY=Math.min(minY,p.y);
    maxX=Math.max(maxX,p.x+p.w);maxY=Math.max(maxY,p.y+p.h);
  });
  const pad=40; minX-=pad;minY-=pad;maxX+=pad;maxY+=pad;
  const vw = mmSvg.clientWidth, vh = mmSvg.clientHeight;
  const s = Math.min(vw/(maxX-minX), vh/(maxY-minY));
  const ox = (vw-(maxX-minX)*s)/2, oy = (vh-(maxY-minY)*s)/2;
  const wx = minX + (e.clientX-r.left-ox)/s;
  const wy = minY + (e.clientY-r.top-oy)/s;
  // centre viewport on that world point
  const stageR = svg.getBoundingClientRect();
  tx = stageR.width/2 - wx*scale;
  ty = stageR.height/2 - wy*scale;
  applyTransform();
});

/* ============================================================
   TOOLBAR
   ============================================================ */
document.getElementById("search").addEventListener("input", e=>{
  filterText = e.target.value.trim().toLowerCase();
  render();
});
document.querySelectorAll(".chip[data-phase]").forEach(ch=>{
  ch.addEventListener("click", ()=>{
    const ph = ch.dataset.phase;
    if(visiblePhases.has(ph)){ visiblePhases.delete(ph); ch.classList.remove("on"); }
    else { visiblePhases.add(ph); ch.classList.add("on"); }
    render();
  });
});
document.querySelector(".chip[data-act=reset]").addEventListener("click", ()=>{
  tx=0; ty=0; scale=1;
  if(mode==="focus") exitFocus();
  else { LAYOUT = layoutPhaseColumns(); selected=null; render(); renderPanel(); }
  applyTransform();
});
document.querySelector(".chip[data-act=focus]").addEventListener("click", ()=>{
  if(mode==="focus"){ exitFocus(); renderPanel(); }
  else if(selected){ enterFocus(selected); }
  else { flashHint("select a table first"); }
});
document.querySelector(".chip[data-act=layout]").addEventListener("click", function(){
  if(mode==="focus"){ exitFocus(); return; }
  if(mode==="columns"){ mode="radial"; LAYOUT=layoutRadial(); this.textContent="radial"; }
  else { mode="columns"; LAYOUT=layoutPhaseColumns(); this.textContent="columns"; }
  tx=0; ty=0; scale=1;
  render(); applyTransform();
});

function flashHint(msg){
  const h = document.getElementById("hint");
  const orig = h.textContent;
  h.textContent = "⚠ " + msg;
  setTimeout(()=>{ h.innerHTML = orig; }, 1800);
}

/* keyboard shortcuts */
window.addEventListener("keydown", e=>{
  if(e.target.tagName==="INPUT") return;
  if(e.key==="f" || e.key==="F"){ document.querySelector('.chip[data-act=focus]').click(); }
  if(e.key==="r" || e.key==="R"){ document.querySelector('.chip[data-act=reset]').click(); }
  if(e.key==="Escape"){ selected=null; applyHighlight(); renderPanel(); }
});

/* ============================================================
   BOOT
   ============================================================ */
render();
renderPanel();
window.addEventListener("resize", drawMinimap);
</script>
</body>
</html>
```

---

## What the final prototype delivers

### 1. Complete schema coverage (47 tables, 6 phase groups)
- **Core spine (5):** `project`, `site`, `wbs`, `spatial_zone_room`, `stakeholder`
- **P1–2 Design (7):** `financial_metric`, `funding_source`, `bim_element` + sub-types `structural_element` / `mep_system`, `design_control`
- **P3–4 Procure (11):** `boq_item`, `contingency`, `schedule_activity`, `activity_dependency` (bridge), `permit`, `risk_register`, `bid_package`, `contract`, `purchase_order`, `supply_shipment`
- **P5 Build (9):** `daily_progress`, `rfi`, `submittal`, `quality_test`, `ncr`, `hse_record`, `site_weather_log`, `payment_application`, `change_order`
- **P6–7 O&M (9):** `asset` (handover bridge), `commissioning_test`, `punch_item`, `municipal_inspection`, `defect`, `maintenance_schedule`, `work_order`, `facility_telemetry`
- **SDG layer (3):** `un_sdg_indicator`, `project_sdg_target`, `sdg_evidence_link`
- **Carbon credit layer (4):** `carbon_project`, `carbon_credit_issuance`, `carbon_credit_transaction`, `carbon_credit_sdg_link`

### 2. Visual encoding
- **Colour bar** on each table = phase group (amber core, blue design, purple procure, green build, pink O&M, teal SDG, yellow carbon)
- **Orange dot** next to PK columns; **teal dot** next to FK columns
- **Edges** drawn as smooth Bézier curves between the source FK and the target PK

### 3. Interaction model
| Action | Result |
|---|---|
| Click a table | Selects it, highlights 1-hop FK links, dims everything else |
| `focus` chip / `F` key | Re-layouts the selected table at the centre with its neighbourhood in a ring |
| `layout` chip | Toggles phase-column ↔ radial layouts (radial places phase groups in concentric arcs) |
| `reset` chip / `R` key | Clears selection, restores column layout, resets pan/zoom |
| `Esc` key | Clears selection only |
| Search box | Filters tables by substring |
| Phase chips | Toggle which phase groups are visible |
| Hover a table | Tooltip: column count, FK count, table description |
| Hover an edge | Tooltip: `source.col → target.pk` |
| Drag empty canvas | Pan |
| Wheel | Zoom (cursor-anchored) |
| Mini-map click | Jump viewport to that region |

### 4. Inspector panel
When a table is selected, the side panel shows:
- Full column list with PK/FK tags and type
- Outgoing FKs (clickable → jumps to target)
- Incoming FKs (clickable → jumps to source)
- `focus 1-hop` button (same as `F` key)

### 5. Traces you can verify in the prototype
- **Handover bridge:** click `asset` → confirm outgoing to `bim_element` (P2 design), `spatial_zone_room` (core), `purchase_order` (P4 procurement); incoming from `commissioning_test`, `punch_item`, `defect`, `maintenance_schedule`, `work_order`, `facility_telemetry` (P6–7).
- **SDG automation loop:** click `sdg_evidence_link` → confirm it points to `project_sdg_target` → which points to `project`, `un_sdg_indicator`, and `wbs`. That is exactly the "evidence feeds target which feeds WBS" chain the SDG document describes.
- **Carbon credit traceability:** click `carbon_credit_sdg_link` → confirms the bridge from `carbon_credit_issuance` to `project_sdg_target`, closing the loop between the finance-side credit instrument and the sustainability-side SDG reporting.
- **Sub-type inheritance:** click `structural_element` or `mep_system` → both FK-back to `bim_element.element_id`, visible as the shared PK/FK.
- **Self-referencing WBS:** `wbs.parent_wbs_id` curls back to the same table (drawn as a same-column loop edge).

### 6. Robustness notes
- All coordinates and transforms are in a single `viewport` group — pan/zoom never re-renders DOM, only a CSS transform, so it stays smooth even with 47 nodes and ~90 edges.
- Mini-map recomputes bounds on every viewport change, so it stays accurate through layout switches and focus mode.
- State is entirely in-memory (no `localStorage` writes) to avoid stale-schema bugs when you edit `SCHEMA` — the toolbar is the source of truth.
- Keyboard shortcuts work only when focus is not in the search input (guarded).
- Every FK declared in `SCHEMA` is deterministically resolved via `pkOf()`, so adding a table only requires appending to the `SCHEMA` array — edges are auto-generated.
````
